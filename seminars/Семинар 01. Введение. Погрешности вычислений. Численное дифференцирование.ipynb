## Что изучает вычислительная математика?

**Вычислительная математика изучает методы решения задач непрерывной математики** посредством систем, выполняющих конечное число операций с конечными величинами (к таким системам относится и компьютер), а также **занимается проблемами обоснования корректности и оценки точности этих методов**.

Численное решение любой задачи сопряжено с целым рядом специфических трудностей, появляющихся из-за особенностей вычислений на компьютере.

1. **Ограниченность памяти компьютера**

    Решение любой задачи на компьютере возможно найти с использованием лишь **конечного** набора значений переменных. Например, при вычислении интеграла возможно использовать лишь конечный набор значений подынтегральной функции. Решение задачи Коши получится найти только в конечном наборе точек на заданном отрезке. Подобный переход от непрерывной функции к функции, заданной на конечном наборе точек, называется **дискретизацией задачи**.

2. **Представление чисел в памяти компьютера**
    
    Для хранения числа в памяти компьютера отводится конечное число бит, поэтому  вычисления производятся с числами, имеющими конечное количество знаков после запятой. При решении задач это приводит к дополнительным сложностям, которых не было в непрерывной математике.

3. **Зависимость результата от способа вычисления (выбора дискретизации)**

    Многие понятия математики, такие как производная, интеграл и др., должны быть "переведены" с языка анализа бесконечно малых на язык **малых (но конечных) величин**, с которыми проводится **большое (но конечное) число операций**. Как показывает практика, способ "перевода" влияет на результат решения задачи. Отсюда и возникает проблема **обоснования** корректности метода и **выбора для данной задачи наилучшего** из методов (если их существует несколько). Интуитивно понятно, что при переходе от строгих определений математического анализа и работы с бесконечными десятичными дробями к алгоритмам вычислительной математики, состоящим из действий с числами, которые представимы в виде конечных десятичных дробей, неизбежны погрешности, сказывающиеся на результате, и поэтому искомое решение задачи может быть найдено только приближённо. Одна из целей вычислительной математики — выяснить, насколько точным является полученное приближение к решению и какими методами его можно улучшить.


Подчеркнём, что в круг вопросов, рассматриваемых вычислительной математикой, **не входит вопрос о реализации тех или иных приближённых методов на конкретных устройствах**. Как запрограммировать тот или иной алгоритм на выбранном языке, как ускорить выполнение полученной программы, какие готовые структуры языка (библиотеки, модули и пр.) можно использовать — всё это вопросы **научного программирования**, но не вычислительной математики. Тем не менее в рамках курса будут предлагаться как теоретические задачи, например, на оценку точности какого-либо метода, так и практические задачи, заключающиеся в программной реализации некоторого метода.

## Модель — алгоритм — программа

Роль вычислительной математики в процессе описания окружащего мира можно проиллюстрировать с помощью триады **"модель — алгоритм — программа"**, предложенной известным отечественным математиком А.А. Самарским ([ссылка](https://lib.mipt.ru/book/14774/?q=Самарский)). Поясним, о чём идёт речь, с помощью простейшего примера — задачи о колебании математического маятника.

fig_resize.png

*Триада "модель — алгоритм — программа" А.А. Самарского*

**1. Модель.**

Отправной точкой для всех теоретических построений является само явление колебания, наблюдаемое в природе. **Создание модели** явления **не является задачей вычислительной математики**, это задача предметной области, в данном случае — механики. Из механики известно, что колебание материальной точки на невесомой нерастяжимой нити (без трения) может быть описано задачей Коши:

$$\ddot\varphi+k\sin\varphi=0,\quad \varphi(0)=\phi_0, \quad \dot\varphi(0)=\omega_0.$$

Предметная область (механика) даёт объяснение смысла входящих в эту задачу величин: $\varphi=\varphi(t)$  — угол отклонения маятника от вертикали, $t$ — время, $k=g/l$, где $l$ — длина маятника, а $g$ — ускорение свободного падения. Этот смысл для вычислительной математики мало интересен (существенны, пожалуй, лишь области определения и значения входящих в уравнения параметров и функций, которые задаются как раз их физическим смыслом). Более важно то, что сформулирована модель — дано конкретное уравнение, которое надо решать численно (на компьютере). Существенно, что **для одного и того же явления можно предлагать различные модели**. Например, для описания **малых** колебаний маятника вместо выписанной выше задачи Коши можно использовать задачу

$$\ddot\varphi+k\varphi=0,\quad \varphi(0)=\phi_0, \quad \dot\varphi(0)=\omega_0.$$

Какую модель применить в данном конкретном случае — это вопрос, решаемый предметной областью. Вычислительная математика "вступает в игру" только после того, как рассматриваемая задача сформулирована в математических терминах (обычно в виде некоторых уравнений).

**2. Алгоритм.**

Чтобы дать **алгоритм** решения задачи, вычислительная математика проводит дискретизацию входящих в задачу объектов и предлагает некий **метод** получения искомого ответа. Например, непрерывное множество значений независимой переменной $t\in[0,+\infty)$ заменяется конечным **набором узлов** (одномерной сеткой) $\{t_i\}_{i=0}^N$ с шагами $\tau_i=t_{i+1}-t_i$, причём $t_0=0$, а $t_N=T$. Непрерывная функция $\varphi=\varphi(t)$ заменяется набором значений $\{\varphi_i\}_{i=0}^N$ в узлах сетки (**сеточной функцией**). Ниже мы для простоты будем считать, что $\tau_0=...=\tau_{N-1}\equiv\tau$ (**равномерная сетка**), хотя это и не обязательно.

Далее встаёт вопрос об **аппроксимации** ("переводе" на дискретный язык) производной. Известное из анализа определение

$$\dot\varphi(t)=\lim_{\tau\to 0}\frac{\varphi(t+\tau)-\varphi(t)}{\tau}$$

адаптируется для дискретного случая, например, следующим образом (так называемая **производная вперёд**):

$$\dot\varphi_i=\frac{\varphi_{i+1}-\varphi_i}{\tau}.$$

Тогда вторую производную можно задать как

$$\ddot\varphi_i=\frac{1}{\tau}\bigg[\frac{\varphi_{i+1}-\varphi_i}{\tau}-\frac{\varphi_{i}-\varphi_{i-1}}{\tau}\bigg]=\frac{\varphi_{i+1}-2\varphi_i+\varphi_{i-1}}{\tau^2}.$$

В итоге получается **разностное уравнение**, связывающее значения сеточной функции $\varphi_i$ в соседних узлах сетки:

$$\frac{\varphi_{i+1}-2\varphi_i+\varphi_{i-1}}{\tau^2}+k\sin\varphi_i=0.$$

 Это уравнение записывается для всех внутренних точек расчетной области и дополняется уравнениями, аппроксимирующими начальные условия. Полученная система уравнений решается на компьютере вместо дифференциального уравнения.



Дальнейшие шаги по аппроксимации начальных условий, решению системы уравнений и т.п. мы сейчас рассматривать не будем. Мы не будем приводить здесь и оценок точности получаемого решения. Нам важно было указать саму идею и подчеркнуть, что целью вычислительной математики является создание **математически обоснованного конечного алгоритма действий с некоторым конечным набором чисел**. После того, как алгоритм создан и дана оценка точности результата, задачу вычислительной математики можно считать выполненной.

**3. Программа.**

На третьем этапе остаётся написать **программу**, реализующую алгоритм для конкретного устройства.

## Численное решение задачи Коши

Приведём пример численного решения первой из обсуждавшихся выше задач Коши. Для этого перепишем входящее в неё уравнение второго порядка в виде эквивалентной системы двух зацепленных уравнений первого порядка (этот приём известен из курса дифференциальных уравнений):

\begin{equation}
  \begin{cases}
    \dot\varphi = f,\\
    \dot f = -k\sin\varphi.
  \end{cases}
\end{equation}

Тогда в векторных обозначениях задача Коши выглядит так:

\begin{equation}
  \dot{\textbf{x}}=\textbf{F},\quad \textbf{x}(0)=(\phi_0, \omega_0)^\text{T},\quad \textbf{F}=(x_2, -k\sin{x_1})^\text{T}.
\end{equation}

Системы такого вида можно решать численно с помощью функции ```odeint``` модуля ```scipy.integrate``` библиотеки ```scipy``` ([документация](https://docs.scipy.org/doc/scipy-1.16.0/reference/generated/scipy.integrate.odeint.html)):

````python
import numpy as np
from scipy.integrate import odeint

# функция правой части
def F(x: np.ndarray, t: np.ndarray, k: float):
    return np.array([x[1], -k*np.sin(x[0])])

# функция, решающая задачу Коши
def solve_Cauchy(phi_0: float, omega_0: float, k: float, t: np.ndarray):
    x_0 = np.array([phi_0, omega_0]) # начальные условия

    return odeint(F, x_0, t, args=(k,))
````

В качестве результата мы получаем массив из векторов $\textbf{x}_i=\textbf{x}(t_i)$, где $t_i$ пробегает значения из выбранной сетки по времени. Результат можно визуализировать, построив графики зависимостей $\varphi=\varphi(t)$ и $\dot\varphi=\dot\varphi(t)$, а также фазовый портрет системы в плоскости $(\varphi(t),\dot\varphi(t))$. Запрограммируем функцию для построения графиков:

````python
import matplotlib.pyplot as plt

# функция для построения графиков
def graph_Cauchy(solution: np.ndarray, t: np.ndarray):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    plt.subplots_adjust(wspace=0.3)

    ax1.plot(t, solution[:,0])
    ax2.plot(t, solution[:,1])

    ax1.set_title("Закон движения математического маятника")
    ax2.set_title("Угловая скорость математического маятника")

    ax1.set_xlabel("$t$")
    ax2.set_xlabel("$t$")

    ax1.set_ylabel("$\\phi(t)$")
    ax2.set_ylabel("$\\dot\\phi(t)$")

    ax1.grid()
    ax2.grid()

    plt.show()

    fig, ax = plt.subplots(figsize=(5, 5))

    ax.plot(solution[:,0], solution[:,1])

    ax.set_title("Фазовый портрет математического маятника")
    ax.set_xlabel("$\\phi(t)$")
    ax.set_ylabel("$\\dot\\phi(t)$")

    ax.grid()

    plt.show()
````

Создадим интерактивные графики, пользуясь средствами библиотеки ```ipywidgets``` ([документация](https://ipywidgets.readthedocs.io/en/latest/)).

````python
from ipywidgets import interact, FloatSlider

# ползунки для интерактивного графика (виджета)
phi0_w = FloatSlider(min=0, max=2*np.pi, step=0.05, value=np.pi/6)
omega0_w = FloatSlider(min=-4.0, max=4.0, step=0.05, value=0)
k_w = FloatSlider(min=0, max=10.0, step=0.05, value=1)

def pendulum(phi_0: float, omega_0: float, k: float):
    grid_t = np.linspace(0, 50, 1000) # сетка по времени
    sol = solve_Cauchy(phi_0=phi_0, omega_0=omega_0, k=k, t=grid_t)
    graph_Cauchy(sol, grid_t)

interact(pendulum, phi_0=phi0_w, omega_0=omega0_w, k=k_w);
````

Можно анимировать движение маятника. Для этого учтём, что уравнение траектории маятника в параметрической форме записывается так (начало координат помещено в точке подвеса):

\begin{equation}
  \begin{cases}
    \displaystyle x=l\sin\varphi(t)=\frac{g}{k}\sin\varphi(t),\\[0.1cm]
    \displaystyle y=-l\cos\varphi(t)=-\frac{g}{k}\cos\varphi(t).
  \end{cases}
\end{equation}

Полезно знать, что основные физические константы можно импортировать из модуля ```scipy.constants``` библиотеки ```scipy``` ([документация со списком констант](https://docs.scipy.org/doc/scipy-1.16.0/reference/constants.html)).

````python
import matplotlib.animation as animation
from IPython.display import HTML

from scipy.constants import g

def create_animation(phi_0: float, omega_0: float, k: float):
    grid_t = np.linspace(0, 50, 1000) # сетка по времени
    sol = solve_Cauchy(phi_0=phi_0, omega_0=omega_0, k=k, t=grid_t)

    l = g / k # длина маятника
    lim = 1.25 * l # параметр для установки пределов по осям (чтобы маятник помещался)

    fig, ax = plt.subplots(figsize=(5, 5))
    ax.set_xlim( (-lim, lim) )
    ax.set_ylim( (-lim, lim) )

    plt.xticks([]) # убираем метки на осях
    plt.yticks([])

    x_p0 = g / k * np.sin(sol[0, 0]) # начальная точка расположения маятника
    y_p0 = - g / k * np.cos(sol[0, 0])

    pend = plt.Circle((x_p0, y_p0), l/20, color='k') # создаём объект маятника (круг)
    ax.add_artist(pend) # добавляем маятник на холст

    line, = ax.plot([0, x_p0], [0, y_p0], color='k') # рисуем нить с подвесом в начале координат

    def animate(i: int):
        x_p = g / k * np.sin(sol[i, 0]) # вычисляем координаты маятника для текущего кадра
        y_p = - g / k * np.cos(sol[i, 0])

        pend.center = (x_p, y_p) # отрисовываем
        line.set_data([0, x_p], [0, y_p])

        return line,

    ani = animation.FuncAnimation(fig, animate, interval=20, blit=True, frames=300)
    plt.close()

    return ani

HTML(create_animation(phi_0=np.pi/6, omega_0=0, k=1).to_jshtml()) # конвертируем в JavaScript
````

Нам не пришлось вникать в сущность этого алгоритма или знакомиться с теорией и принципами, на которых он построен. Мы воспользовались им как "чёрным ящиком", выдающим ответ по исходным данным.

Точно так же можно (и нужно!) поступать при решении любой другой прикладной задачи (если, конечно, разработка методов численного решения задачи не является основной целью работы). Подавляющее большинство численных методов, разработанных вычислительной математикой, уже имеет готовую реализацию или даже несколько реализаций в виде специализированных программных пакетов (по крайней мере, про методы, изучающиеся в нашем курсе, это можно утверждать совершенно точно). Современному физику-исследователю нет никакой необходимости самостоятельно писать программу, которая решала бы встретившуюся ему  математическую задачу.

В связи с этим возникает вопрос: **для чего нужно знать вычислительную математику**? Зачем знакомиться с принципами построения давно реализованных методов, если изучение этих методов не являются самоцелью?

Оказывается, что для грамотного применения готовых вычислительных программ (функций, модулей, библиотек, написанных на каком-либо языке) необходимо иметь хотя бы **общее представление** о принципах их работы. В противном случае затруднительно сделать осознанный выбор в пользу той или иной программной реализации из ряда внешне похожих. Кроме того, невозможно правильно истолковать отрицательный результат работы всей программы. Действительно, предположим, что некий готовый алгоритм является составной частью длинного кода, выдающего ошибку при попытке запуска (на практике подобные ситуации встречаются на каждом шагу!). Как понять, связана ли эта ошибка с неверно написанной программой или же дело в том, что **при правильно составленной программе** некорректно "работает" сам численный метод? Ситуация осложняется, если последовательно используются несколько разных вычислительных алгоритмов, так что результаты работы одного яляются входными данными для другого.

Ещё более запутанным является иной случай: программа запускается и выдаёт результат, отличный от ожидаемого. Как определить, является ли он ошибочным или правильным? Может быть, модель явления демонстрирует какое-то новое, ранее неизвестное свойство, т.е. никакой ошибки нет (совершено открытие)? Если же допущена ошибка, то в чём она состоит: в выборе неприменимого метода или в неверно написанной программе?

Очевидно, что анализировать описанные выше **нестандартные случаи** без понимания сути происходящего чрезвычайно трудно. А ведь смысл научной работы состоит именно в таком анализе! Знание же основных теоретических положений, связанных с конкретным численным методом, позволяет при необходимости разобраться в любой готовой программе, предлагающей реализацию этого метода. **Понимание** идей и принципов, стоящих за строками кода (с которыми и предлагается ознакомиться в рамках курса вычислительной математики во время учёбы), сэкономит время и силы при всей дальнейшей научной работе.

В подтверждение сказанного приведём два простых примера, в которых **использование готовых численных алгоритмов приводит к неверным результатам при правильно написанных программах**. Идеи примеров заимствованы из лекции А.В. Чикиткина: [cсылка](https://youtu.be/p2g3dY3nQx4?si=SaOzE7h2488VKKjM&t=659) (YouTube).

## Пример 1

Пусть требуется численно решить систему линейных алгебраических уравнений (СЛАУ)

$$\textbf{A}\textbf{x}=\textbf{f},$$

где $\textbf{A}$ — невырожденная матрица порядка $n$. Нас будет интересовать матрица специального вида — так называемая **матрица Гильберта** (на доказательстве того, что она является невырожденной, мы не останавливаемся):

$$\textbf{A}=(a_{ij}),\quad a_{ij}=\frac{1}{i+j-1},\quad 1\leq i, j\leq n.$$

Чтобы получить СЛАУ с выбранной матрицей и известным точным решением, зададим это решение вручную и вычислим соответствующую правую часть. Для простоты в качестве точного решения возьмём единичный вектор.

Напишем функцию, создающую матрицу Гильберта $\textbf{A}$ заданного порядка $n$ и определяющую вектор правой части $\textbf{f}$ так, чтобы точным решением был единичный вектор. В коде ниже при создании матрицы знак "плюс" в знаменателе дроби поставлен потому, что в языке Python индексирование элементов массива начинается с нуля, а не с единицы.

````python
import numpy as np

def create_SLE(n: int):
    A = np.fromfunction(lambda i, j : 1 / (i + j + 1), (n, n))
    f = A @ np.ones(n) # матричное умножение

    return A, f
````

Теперь мы можем легко задать СЛАУ из произвольного числа уравнений $n$ с матрицей выбранного нами вида и единичным вектором в качестве точного решения. Напишем аналогичную функцию для задания СЛАУ такой же размерности, но со случайной матрицей, элементы которой принадлежат отрезку $[0,1]$:

````python
def create_SLE_rand(n: int):
    A = np.random.rand(n, n) # матрица случайных чисел из отрезка [0, 1]
    f = A @ np.ones(n)

    return A, f
````

Будем численно решать обе системы при различных $n$ с помощью функции ```solve``` из модуля ```linalg``` библиотеки ```numpy``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html)). Так как нам известно точное решение (единичный вектор), мы можем контролировать норму ошибки (норму вектора разности численного и точного решений). Для расчета выберем кубическую норму, которая для вектора $\textbf{u}$ определяется как $||\textbf{u}||=\displaystyle\max_i |u_i|$ и принадлежит к классу $p$-норм векторов, $p=\infty$ (см. семинар 2 по прикладной линейной алгебре).

````python
from ipywidgets import interact, IntSlider

# ползунок для интерактивного графика (виджета)
n_w = IntSlider(min=1, max=100, step=1, value=3)

def compare_solutions(n: int):
    A, f = create_SLE(n)
    x = np.linalg.solve(A, f)
    eps1 = np.linalg.norm(x-np.ones(n), ord=np.inf) # кубическая норма разности решений

    A, f = create_SLE_rand(n)
    x = np.linalg.solve(A, f)
    eps2 = np.linalg.norm(x-np.ones(n), ord=np.inf)

    print("Специальная матрица. Ошибка: {:.3e}".format(eps1))
    print("Случайная матрица. Ошибка: {:.3e}".format(eps2))

interact(compare_solutions, n=n_w);
````

Как видим, уже при $n=3$ погрешность решения СЛАУ с матрицей Гильберта превосходит погрешность решения СЛАУ со случайной матрицей почти на два порядка. При дальнейшем увеличении $n$ первая ошибка стремительно нарастает и достигает величины $\sim 10^2$ при $n=100$ (т.е. превосходит норму точного решения на два порядка!). Величина же второй ошибки остаётся на уровне $\sim 10^{-14}$ (уровень ошибок округления, см. ниже) во всей области изменения $n$. Наблюдаемый эффект объясняется **плохой обусловленностью** матрицы Гильберта (см. семинар 2). Таким образом, один и тот же готовый алгоритм (одна и та же функция ```np.linalg.solve```) даёт разные результаты в разных случаях, причём причину различий следует искать не на уровне программной реализации, а глубже — в свойствах системы и метода её решения (в рассматриваемой функции реализованы **метод** последовательных исключений **Гаусса** и LU-разложение).

## Пример 2

Пусть дана таблица значений некоторой функции $y=f(x)$ в **конечном упорядоченном** наборе точек:

\begin{align}
  x_0,\quad  x_1,\quad ...,\quad  x_{n};\\
  y_0,\quad  y_1,\quad ...,\quad  y_{n}.\\
\end{align}

Требуется восстановить гладкую функцию $f(x)$ на отрезке $[x_0,x_{n}]$ так, чтобы её график прошёл через все заданные точки (т.е. решить задачу **точной интерполяции**).

Существует несколько способов интерполирования функций. Все они сводятся к построению **обобщённого многочлена**

$$P_n(x)=\sum\limits_{k=0}^nc_k\varphi_k(x)\quad \text{такого, что}\quad P_n(x_k)=y_k,\quad 0\leq k\leq n.$$

Здесь $\varphi_k(x)$ — некоторые фиксированные функции. Если $\varphi_k(x)=x^k$, то интерполяция называется **алгебраической**.

Оказывается, что если таблица из $n+1$ значения функции не содержит повторяющихся узлов, то существует один и только один полином степени не выше $n$, проходящий через все заданные точки.

Построим таблицу значений функции из $n+1$ узла. Узлы $x_k$ выберем некоторым специальным образом, а значения $y_k$ возьмём случайными из отрезка $[0,1]$.

````python
import numpy as np

def create_table(n: int):
    x = np.pi * np.linspace(-1.0, 0.0, n+1, endpoint=False) # задаём узлы
    x = np.cos(x)
    x = np.sort(x) # упорядочиваем узлы по возрастанию

    y = np.random.rand(n+1)

    return x, y
````

Построим интерполяционный полином двумя способами: с помощью функции ```polyfit``` библиотеки ```numpy``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.polyfit.html)) и с помощью класса ```BarycentricInterpolator``` модуля ```scipy.interpolate``` библиотеки ```scipy``` ([документация](https://docs.scipy.org/doc/scipy-1.16.0/reference/generated/scipy.interpolate.BarycentricInterpolator.html#scipy.interpolate.BarycentricInterpolator)).

````python
import matplotlib.pyplot as plt
from scipy.interpolate import BarycentricInterpolator

def poly_with_numpy(x: np.ndarray, y: np.ndarray):
    poly = np.polyfit(x, y, x.size)

    grid = np.linspace(np.min(x), np.max(x), 1000) # сетка по x для построения графика
    y_graph = np.polyval(poly, grid)

    plt.plot(x, y, 'ro', ms=5)
    plt.plot(grid, y_graph, 'k-')
    plt.title("Интерполяция с помощью numpy")
    plt.show()

def poly_with_scipy(x: np.ndarray, y: np.ndarray):
    poly = BarycentricInterpolator(x, y)

    grid = np.linspace(np.min(x), np.max(x), 1000)
    y_graph = poly(grid)

    plt.plot(x, y, 'ro', ms=5)
    plt.plot(grid, y_graph, 'b-')
    plt.title("Интерполяция с помощью scipy")
    plt.show()
````

Будем строить графики полученных полиномов (**интерполянтов**) при разных $n$:

````python
from ipywidgets import interact, IntSlider

# ползунок для интерактивного графика (виджета)
n_w = IntSlider(min=1, max=50, step=1, value=10)

def graph_poly(n: int):
    x, y = create_table(n)
    poly_with_numpy(x, y)
    poly_with_scipy(x, y)

interact(graph_poly, n=n_w);
````

Уже при сравнительно небольших $n$ (порядка 10) на графиках заметны локальные различия. При $n\sim 40$ результаты работы функции ```np.polyfit``` оказываются неудовлетворительными: появляются точки, через которые не проходит график интерполяционного многочлена, причём с увеличением $n$ количество таких точек растёт. В то же время полином, получаемый при помощи класса ```BarycentricInterpolator```, правильно решает поставленную задачу при всех рассматриваемых $n$.

Дело в том, что использованные библиотеки реализуют разные методы интерполяции. Функция ```np.polyfit``` строит интерполяционный многочлен по **методу наименьших квадратов**, а класс ```BarycentricInterpolator``` строит интерполяционный **полином Лагранжа** в модифицированом (барицентрическом) виде. Последний метод обеспечивает свойство интерполяционности по построению, а первый даёт лишь наилучшее приближение к заданным точкам (в смысле минимизации суммы квадратов отклонений от них).

## Погрешности вычислений

Пусть в качестве результата измерения или численного расчёта получена некоторая величина $u^*$. Обозначим **точное** (и заведомо неизвестное) значение этой же величины через $u$.

**Определение 1.** *Абсолютной погрешностью* величины $u^*$ называют *наименьшую* величину $\Delta(u^*)$ такую, что выполняется неравенство
$$|u-u^*|\leq \Delta(u^*).$$

**Определение 2.** *Относительной погрешностью* величины $u^*$ называют *наименьшую* величину $\delta(u^*)$ такую, что выполняется неравенство
$$\frac{|u-u^*|}{|u^*|}\leq \delta(u^*).$$

Иначе говоря, $\delta(u^*)=\Delta(u^*)/|u^*|$.

На практике воспользоваться этими определениями для вычисления $\Delta(u^*)$ или $\delta(u^*)$ невозможно, поскольку точное значение величины $u$ никогда не бывает известным. В связи с этим приходится прибегать к **оценкам погрешностей**.

## Оценки погрешностей

Пусть известно, что величина $u$ есть функция некоторого параметра $t$: $u=u(t)$  причём $u^*=u(t^*)$. Погрешность $\Delta(t^*)$ предполагается известной (либо оценённой ранее). Требуется дать оценку на $\Delta(u^*)$.

Обычно используют три оценки погрешностей:
* Предельная абсолютная погрешность:

  $$D(u^*)=\sup_{t\in\Omega} |u(t)-u^*|,\quad \Omega=\{t\in \mathbb{R}: |t-t^*|\leq \Delta (t^*)\}.$$

* Погрешность по точной верхней грани производной:

  $$D_1(u^*)=\sup_{\theta\in\Omega} |u'(\theta)|\Delta(t^*).$$

* Линейная оценка:

  $$D_2(u^*)= |u'(t^*)|\Delta(t^*).$$

Происхождение последних двух оценок легко понять, если вспомнить **теорему Лагранжа** для дифференцируемой функции $u(t)$, обычно изучаемую в курсе математического анализа 1 семестра.

**Теорема Лагранжа.** Пусть функция $u(t)$ дифференцируема на отрезке $[t^*, t]$. Тогда существует внутренняя точка $\theta$ этого отрезка, для которой
$$u'(\theta)=\frac{u(t^*)-u(t)}{t^*-t}.$$

Иначе говоря, $u(t)-u(t^*)=u'(\theta)(t-t^*)$. В таком виде теорему Лагранжа называют **формулой конечных приращений** или **теоремой о среднем значении**.

Отметим, что эту теорему можно рассматривать как частный случай формулы Тейлора с остаточным членом в форме Лагранжа:

$$u(t)=\sum\limits_{k=0}^{n}\frac{u^{(k)}(t^*)}{k!}(t-t^*)^k+\frac{u^{(n+1)}(\theta)}{(n+1)!}(t-t^*)^{n+1},\quad \theta\in[t^*, t].$$

Применим теорему Лагранжа дважды:

$$u(t^*)-u\big(t^*+\Delta (t^*)\big)=u'(\theta_1)\Delta (t^*),\quad \theta_1 \in[t^*, t^*+\Delta (t^*)],$$

$$u\big(t^*-\Delta (t^*)\big)-u(t^*)=u'(\theta_2)\Delta (t^*),\quad \theta_2 \in[t^*-\Delta (t^*), t^*].$$

Введём обозначения:

$$\Delta_1\equiv |u'(\theta_1)|\Delta (t^*),\quad\Delta_2\equiv |u'(\theta_2)|\Delta (t^*).$$


Тогда очевидно, что
$$D_1(u^*)=\max(\Delta_1,\Delta_1)=\sup_{\theta\in\Omega} |u'(\theta)|\Delta(t^*).$$

Оценку $D_2(u^*)$ можно рассматривать как упрощённую версию оценки $D_1(u^*)$, в которой вместо точной верхней грани производной берётся её значение при $t=t^*$.

## Задача 1

Найти предельную абсолютную погрешность, погрешность по производной и линейную оценку погрешности для функции $u=t^{10}$, если заданы точка приближения $t^*=1$, значение функции $u^*$ в этой точке и погрешность $\Delta (t^*)=0,1$.

**Решение** состоит в использовании приведённых выше формул для оценок погрешностей. Так как $u^*=u(t^*)=1$, а $u'(t)=10t^9$, последовательно находим:

$$D(u^*)=\sup_{|t-1|\leq 0,1}|t^{10}-1|=1,1^{10}-1=1,59,$$

$$D_1(u^*)=\sup_{|t-1|< 0,1}|10t^9|\cdot 0,1=10\cdot 1,1^9\cdot 0,1=2,36,$$

$$D_2(u^*)=|10t^9|_{t=1}\cdot 0,1=10\cdot 0,1=1.$$

## Погрешность неявной функции

Пусть функция $u=u(\textbf{t})$, $\textbf{t}\in \mathbb{R}^n$ задана неявно уравнением $F(u,\textbf{t})=0$. Требуется оценить погрешность $\Delta(u^*)$, если заданы точка приближения $\textbf{t}^*$ и погрешности $\Delta(t_i^*)$.

Воспользуемся обобщением линейной оценки на случай функции многих переменных:

$$D_2(u^*)=\sum\limits_{i=1}^n|u'_{t_i}(t^*_i)|\Delta(t_i^*).$$

Частные производные функции $u$ найдём, продифференцировав уравнение по $t_i$:

$$F'_u\cdot u'_{t_i}+F'_{t_i}=0.$$

Отсюда

$$u'_{t_i}(t^*_i)=-\frac{F'_{t_i}(u^*,\textbf{t}^*)}{F'_u(u^*,\textbf{t}^*)}.$$

Искомая оценка:

$$D_2(u^*)=\big|F'_u(u^*,\textbf{t}^*)\big|^{-1}\cdot\sum\limits_{i=1}^n\big|F'_{t_i}(u^*,\textbf{t}^*)\big|\Delta(t_i^*).$$

Величина $u^*$ может быть найдена как корень уравнения $F(u,\textbf{t}^*)=0$ в предположении выполнения условий теоремы о неявной функции.

## Задача 2

Оценить погрешность в определении действительного корня уравнения $ax^3+bx^2+cx+d=0$, коэффициенты которого заданы приближённо: $a^*=c^*=1,\quad b^*=d^*=-2,\quad \Delta(a^*)=\Delta(b^*)=\Delta(c^*)=\Delta(d^*)=10^{-4}\equiv \Delta$.

**Решение.** Найдём приближённое значение корня. Уравнение имеет вид:
$$x^3-2x^2+x-2=0\quad \text{или}\quad (x^2+1)(x-2)=0.$$
Действительный корень есть $x^*=2$.

Пусть $F=F(x,a,b,c,d)=ax^3+bx^2+cx+d$. Согласно полученной выше формуле,

$$D_2(x^*)=\big|F'_x(*)\big|^{-1}\Big[|F'_a(*)|\Delta(a^*)+|F'_b(*)|\Delta(b^*)+|F'_c(*)|\Delta(c^*)+|F'_d(*)|\Delta(d^*)\Big].$$

Здесь для краткости записей мы положили $F'_x(*)=F'_x(x^*,a^*,b^*,c^*,d^*)$ и т.д. В нашем случае абсолютные погрешности коэффициентов одинаковы и равны $\Delta$, поэтому

$$D_2(x^*)=\big|F'_x(*)\big|^{-1}\cdot\Delta\cdot\Big[|F'_a(*)|+|F'_b(*)|+|F'_c(*)|+|F'_d(*)|\Big].$$

Остаётся найти частные производные функции $F$ при значениях параметров "со звёздочкой".

$$F'_x=3ax^2+2bx+c,\quad F'_x(*)=3\cdot 1\cdot 2^2+2\cdot (-2)\cdot 2 + 1=5;$$

\begin{align}
  F'_a=x^3,&\quad F'_a(*)=8;\qquad F'_b=x^2,\quad F'_b(*)=4;\\
  F'_c=x,&\quad F'_c(*)=2;\qquad F'_d=1,\quad F'_d(*)=1.
\end{align}

Искомая оценка:
$$D_2(x^*)=5^{-1}\cdot 10^{-4}\cdot\Big[8+4+2+1\Big]=3\cdot 10^{-4}.$$

## Классификация погрешностей

Во введении была упомянута триада А.А. Самарского "модель — алгоритм — программа", отражающая суть численных методов решения практических задач. Понятно, что погрешности (ошибки) могут возникнуть на каждом из трёх этапов решения. Ответ, выдаваемый компьютером, может оказаться далёк от точного (заведомо неизвестного) ответа из-за неудачного выбора исходной модели, использования неверного (не применимого в рассматриваемом случае) алгоритма (метода решения) или ошибок в программе (опечаток, "багов" и пр.). Однако такой взгляд на источники погрешностей вычислений является слишком общим.

В вычислительной математике принята следующая классификация погрешностей вычислений. Погрешности делятся на
* **неустранимые**, к которым относятся ошибки входных данных и ошибки модели,
* **погрешности метода**,
* **погрешности округления**.

Пример **ошибки модели** мы, по сути, привели выше, когда обсуждали описание колебаний математического маятника с помощью двух задач Коши. Вторая модель

$$\ddot\varphi+k\varphi=0,\quad \varphi(0)=\phi_0, \quad \dot\varphi(0)=\omega_0$$

применима в случае малых колебаний. Ошибка, связанная с заменой $\sin\varphi$ на $\varphi$, будет тем больше, чем больше угол $\varphi$. С точки зрения вычислительной математики, это — **неустранимая** ошибка: её принципиально невозможно исключить, поскольку нельзя изменить модель. Модель строит и обосновывает не вычислительная математика, а предметная область. Вычислительная математика работает "с тем, что есть".

Приведём примеры других ошибок.

## Ошибка входных данных

Пусть решается система линейных алгебраических уравнений (СЛАУ)

\begin{equation}
  \begin{cases}
    0,99x_1 + x_2 =1,\\
    x_1 + 0,99x_2=1.
  \end{cases}
\end{equation}

Решение можно найти аналитически:

\begin{equation}
  x_1=x_2=\frac{100}{199}\simeq 0,5025.
\end{equation}

Для численного решения можно воспользоваться функцией ```solve``` из модуля ```linalg``` библиотеки ```numpy``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html)):



````python
import numpy as np

A = np.array( [ [0.99, 1.00], [1.00, 0.99] ] ) # матрица системы
b = np.array( [1.00, 1.00] ) # вектор правых частей

x = np.linalg.solve(A, b) # решение системы

np.set_printoptions(precision=4) # количество десятичных знаков при выводе
print(x)
````

Изменим правые части уравнений системы на малую величину $\varepsilon=10^{-2}$ (будем считать, что эта ошибка определяется, например, точностью некоторого эксперимента):

\begin{equation}
  \begin{cases}
    0,99x_1 + x_2 =1,01,\\
    x_1 + 0,99x_2=0,99.
  \end{cases}
\end{equation}

Снова находим решение (для примера — аналитически и численно):

\begin{equation}
  x_1^*=-\frac{99}{199}\simeq -0,4975,\quad x_2^*=\frac{299}{199}\simeq 1,5025.
\end{equation}

````python
eps = 10**(-2)
b = np.array( [1.00 + eps, 1.00 - eps] )

x_eps = np.linalg.solve(A, b)

print(x_eps)
````

Изменение вектора решения составило $\Delta\textbf{x}=(-1, 1)^\text{T}$:

````python
delta_x = x_eps - x
delta_x
print(delta_x)
````

Заметим, что знак первого корня изменился на противоположный, а модуль второго корня увеличился в три раза. Такие изменения по "физическим меркам" очень значительны.

Как видим, погрешность входных данных (в данном случае — вектора правых частей уравнений системы) привела к погрешности решения. Ошибка $\varepsilon$ является **неустранимой** в том смысле, что её влияние никак нельзя уменьшить или исключить: для этого потребовалось бы точнее задавать правые части, т.е. вмешиваться в эксперимент и предлагать способ произвести более точные измерения. Но подобное вмешательство выходит за рамки задач, решаемых вычислительной математикой. Занимаясь численными методами, можно только оценить погрешность результата, связанную с ошибкой входных данных. Как выполняются подобные оценки, мы рассматривали выше.

Итак, в данном случае малое по относительной величине изменение правых частей СЛАУ привело к существенному изменению решения. Другими словами, данная СЛАУ демонстрирует высокую чувствительность к погрешности входных данных. В подобных случаях говорят, что решаемая задача **плохо обусловлена**. К понятию обусловленности мы ещё вернёмся на семинаре 2.

Отметим, что правые части (как и коэффициенты матрицы СЛАУ) всегда заданы с точностью до ошибок округления. Анализ чувствительности решения к погрешностям округления и входных данных необходимо проводить при решении любых задач.

## Ошибка метода

Пусть требуется вычислить значение функции $y(x)=\sin x$ в некоторой точке $a$. Для этого рассмотрим её разложение в ряд Маклорена:

$$y(x)=\sum\limits_{k=0}^\infty\frac{(-1)^k}{(2k+1)!}x^{2k+1}.$$

Из курса анализа известно, что этот **знакопеременный** степенной ряд имеет бесконечный радиус сходимости. Это значит, что при любом $x$ он сходится абсолютно и, кроме того, для всякого отрезка числовой оси сходимость является равномерной. Напомним, что равномерная сходимость степенного ряда на некотором отрезке влечёт бесконечную дифференцируемость суммы ряда на этом отрезке. Однако при этом сумма ряда Маклорена может не совпасть с той функцией, по которой он построен. То, что в случае функции $y(x)=\sin{x}$ такое совпадение имеет место при любом $x$ (т.е. что функция $y(x)=\sin{x}$ является **регулярной** на $\mathbb{R}$), также доказывается в курсе анализа.

Практическое следствие приведённых математических фактов состоит в том, что точное значение величины $\sin a$ можно получить, если найти сумму бесконечного количества слагаемых ряда, вычисленных в точке $a$. Просуммировать бесконечное число слагаемых за конечное время невозможно. Однако можно предложить приближённый метод вычисления суммы ряда.

Оборвём ряд на $n$-м члене и будем считать, что $y(a)\simeq S_n(a)$, где $S_n(x)$ — сумма первых $n$ членов ряда:

$$S_n(x)=\sum\limits_{k=0}^{n-1}\frac{(-1)^k}{(2k+1)!}x^{2k+1}.$$

В силу знакопеременности ряда, для его остатка

$$r_n(x)=\sum\limits_{k=n}^{\infty}\frac{(-1)^k}{(2k+1)!}x^{2k+1}$$

справедлива оценка:

$$|r_n(x)|\leq\frac{|x|^{2n+1}}{(2n+1)!}\equiv\varepsilon_n.$$

*График функции $y=\sin x$ (чёрная кривая) и графики частичных сумм $S_n(x)$ ряда Маклорена (красные кривые)*

Фиксируя натуральное $n$, мы получаем метод вычисления величины $\sin a$ путём подсчёта конечной суммы. Ошибка метода определяется оценкой на остаток ряда, т.е. оценивается сверху величиной $\varepsilon_n$. Поскольку $\varepsilon_n\to 0$ при $n\to \infty$, ошибку метода можно сделать сколь угодно малой. В этом смысле она является **устранимой**.

Запрограммируем функцию ```sum_sin```, вычисляющую сумму первых $n$ членов ряда в точке $x$:

````python
import numpy as np
from math import factorial

def sum_sin(x: float, n: int):
    # для ускорения суммирования вместо цикла используем list comprehension и векторизацию
    s = np.array( [ (-1)**k * x**(2*k+1) / factorial(2*k+1) for k in range(n) ] )

    return s.sum()
````

Построим график зависимости $S_n(x)$ от числа слагаемых $1\leq n\leq N$. График будем строить при произвольном $x=a$. Для сравнения будем выводить на печать значения частичной суммы $S_N(a)$ и $\sin a$. Последнее получим с помощью функции ```np.sin```.

````python
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

from ipywidgets import interact, FloatSlider, IntSlider

# ползунки для интерактивного графика (виджета)
a_w = FloatSlider(min=-20.0, max=20.0, step=0.05, value=np.pi/6)
N_w = IntSlider(min=1, max=50, step=1, value=10) # верхняя граница диапазона для n

# функция для построения графика
def graph_sin(a: float, N: int):
    grid_n = [ i+1 for i in range(N) ] # сетка значений n
    sin_n = [ sum_sin(a, i) for i in range(N) ] # сетка значений S_n(x)

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.scatter(grid_n, sin_n) # точки
    ax.plot(grid_n, sin_n) # линия

    plt.title("Частичные суммы $S_n(x)$ в зависимости от $n$ при $x=a$")
    plt.xlabel("$n$")
    plt.ylabel("$S_n(a)$")

    plt.grid()
    ax.xaxis.set_major_locator(MaxNLocator(integer=True)) # подписи n - только целые

    plt.show()

    print("S_N(a) =", sin_n[N-1])
    print("sin(a) =", np.sin(a))

interact(graph_sin, a=a_w, N=N_w);
````

Мы видели, что в теории ошибка метода $\varepsilon_n$ может быть сделана сколь угодно малой выбором достаточно большого $n$. Однако при вычислении $S_n(a)$ на реальном компьютере из-за быстрого роста **ошибок округления** (см. ниже) достичь некоторой наперёд заданной точности $\varepsilon$ для $a\gg 1$ не удаётся.

Дело в том, что ошибки округления оказываются тем больше, чем больше $a$. Это связано с
различным поведением модулей членов ряда при $a< 1$ и $a>1$. При $a<1$ абсолютные величины членов ряда монотонно убывают, начиная с $n=0$. При $a>1$ модули членов ряда сначала растут (тем сильнее, чем больше $a$) и только потом, достигнув при некотором $k=m$ максимума, начинают убывать. Чтобы обеспечить при вычислении, например, максимального по модулю члена ряда $s_m\sim a^m$ абсолютную погрешность, не
превосходящую $\varepsilon$, необходимо вычислить его с относительной погрешностью, не превосходящей $\varepsilon/|s_m|\sim \varepsilon/|a^m|$. Иначе говоря, требуемая относительная точность вычисления членов конечной суммы $S_n(a)$ возрастает с увеличением $a$. Обеспечить же сколь угодно высокую точность можно только при неограниченной длине мантиссы (см. ниже).

В реальных программных пакетах значения элементарных функций **никогда не вычисляются** через частичные суммы степенных рядов.

## Ошибка округления

Рассмотрим квадратное уравнение

$$x^2-10^8x+1=0.$$

Его решения могут быть найдены по хорошо известной формуле:

$$x_{1,2}=\frac{10^8\pm\sqrt{10^{16}-4}}{2}=10^8\cdot\frac{1\pm\sqrt{1-4\cdot 10^{-16}}}{2}.$$

Обратим внимание на подкоренное выражение $q=1-4\cdot 10^{-16}$. В арифметике **бесконечной разрядности** (когда числа могут иметь сколь угодно много значащих разрядов) значение $q$ вычисляется точно. Однако в реальном компьютере число разрядов всегда ограничено, из-за чего неизбежны **ошибки округления**.

Ошибки округления можно охарактеризовать величиной $\varepsilon_\text{маш}$ (так называемое **машинное эпсилон** — подробнее о нём см. ниже), которая определяется как наибольшее число, для которого в машинной арифметике выполняется равенство $1+\varepsilon_\text{маш}=1$. Иначе говоря, числа, относительная разность которых не превосходит $\varepsilon_\text{маш}$, компьютером не различаются, т.е. $\varepsilon_\text{маш}$ — это максимальная относительная погрешность округления в представлении чисел с плавающей запятой.

Если предположить, что компьютер выполняет расчёты с $\varepsilon_\text{маш}=10^{-8}$, то число $q$ будет округлено до 1 (поскольку $4\cdot 10^{-16}<\varepsilon_\text{маш}=10^{-8})$. В результате получится, что $x_{1}=10^8$, $x_{2}=0$. Меньший корень уравнения будет найден неверно.

Один из возможных способов правильного вычисления меньшего корня сводится к использованию теоремы Виета, согласно которой

$$x_1+x_2=10^8,\quad x_1\cdot x_2 =1.$$

Из второго равенства следует, что $x_2=10^{-8}=\varepsilon_\text{маш}$. При этом первое равенство тоже окажется выполненным в силу определения $\varepsilon_\text{маш}$.

Можно также решить проблему, перейдя к расчётам с $\varepsilon_\text{маш}=10^{-16}$ (т.е. к представлению чисел с **двойной точностью**). Более подробная справка о вычислениях с одинарной и двойной точностью приведена ниже.


Программа для вычисления корней тремя указанными способами дает следующие результаты:

````python
import numpy as np

def quad_roots1(b: float):
    D = np.float32(b * b - 4) # дискриминант с одинарной точностью
    sqrt_D = np.sqrt(D)

    x_1 = ( b + sqrt_D ) / 2
    x_2 = ( b - sqrt_D ) / 2 # меньший корень найден через дискриминант

    return x_1, x_2

def quad_roots2(b: float):
    D = np.float32(b * b - 4)
    sqrt_D = np.sqrt(D)

    x_1 = ( b + sqrt_D ) / 2
    x_2 = 1 / x_1 # меньший корень найден по теореме Виета

    return x_1, x_2

def quad_roots3(b: float):
    D = np.float64(b * b - 4) # дискриминант с двойной точностью
    sqrt_D = np.sqrt(D)

    x_1 = ( b + sqrt_D ) / 2
    x_2 = ( b - sqrt_D ) / 2 # меньший корень найден через (более точный) дискриминант

    return x_1, x_2

x_11, x_12 = quad_roots1(10**8)
x_21, x_22 = quad_roots2(10**8)
x_31, x_32 = quad_roots3(10**8)

print( "Вычисление через дискриминант float 32: ", x_11, ", ", x_12, sep="")
print( "Поправка по теореме Виета: ", x_21, ", ", x_22, sep="")
print( "Вычисление через дискриминант float 64: ", x_31, ", ", x_32, sep="")
````

Точность $10^{-8}$, заданная в условии, хорошо соответствует $\varepsilon_\text{маш}= 6 \cdot 10^{-8}$ при вычислениях с одинарной точностью (см. ниже). В Python используется динамическая типизация, поэтому в программе выше пришлось использовать типы данных из библиотеки ```numpy``` ([документация по типам данных](https://numpy.org/doc/stable/user/basics.types.html)).

## Задача 3

Пусть $z=f(x,y)$, причём величины $x$ и $y$ измеряются приближенно. Известно, что (неустранимая) погрешность измеренного значения $x^*$ составляет $\Delta(x^*) = 10^{–3}$. Будем интересоваться **абсолютной** погрешностью $\Delta(z^*)$. С какой разумной точностью следует измерять $y^*$?

Рассмотреть случаи $f(x,y)=x+10y$ и $f(x,y)=xy+xy^2$.

**Решение в первом случае.** По свойству (вообще говоря, предельных) погрешностей

$$\Delta(z^*)=\Delta(x^*)+10\Delta(y^*).$$

Если какой-то из членов суммы отличается от другого хотя бы на порядок, то меньший член можно выбросить без большого ущерба для оценки величины $\Delta(z^*)$. Поэтому естественно требовать, чтобы вклад в $\Delta(z^*)$ от каждого из слагаемых был примерно одинаков:

$$\Delta(x^*)\sim 10\Delta(y^*).$$

Отсюда находим $\Delta(y^*)=10^{-1}\cdot\Delta(x^*)=10^{-4}$. Гнаться за большей точностью при измерении $y^*$ нет смысла: маленькую погрешность $y^*$ "перекроет" большая погрешность $x^*$.

**Решение во втором случае.** Будем ориентироваться на линейную оценку погрешности (см. выше). Тогда с величинами $\Delta(x^*)$, $\Delta(y^*)$, $\Delta(z^*)$ можно обращаться как с дифференциалами. Поэтому

$$\Delta(z^*)=y^*\Delta(x^*)+(y^*)^2\Delta(x^*)+x^*\Delta(y^*)+2x^*y^*\Delta(y^*).$$

Иначе:

$$\Delta(z^*)=\big(y^*+(y^*)^2\big)\Delta(x^*)+(x^*+2x^*y^*)\Delta(y^*).$$

Руководствуясь той же логикой, что и в первом случае, приравняем вклады от $\Delta(x^*)$ и $\Delta(y^*)$:

$$\big(y^*+(y^*)^2\big)\Delta(x^*)=(x^*+2x^*y^*)\Delta(y^*).$$

Как видим, выразить $\Delta(y^*)$ только через $\Delta(x^*)$ не удаётся: в равенство входят $x^*$ и $y^*$. Это значит, что дать оценку на абсолютную погрешность без дополнительной информации нельзя: надо знать измеренные значения $x^*$ и $y^*$. Однако можно оценить относительную погрешность (знаки модулей не пишем, считая $x^*$ и $y^*$ положительными):

$$\delta(y^*)\equiv\frac{\Delta(y^*)}{y^*}=\frac{1+y^*}{1+2y^*}\frac{\Delta(x^*)}{x^*}\equiv g(y^*)\cdot\delta(x^*).$$

Построим график функции $g(y^*)$:

````python
import matplotlib.pyplot as plt
import numpy as np

from ipywidgets import interact, FloatSlider

# ползунки для интерактивного графика (виджета)
l_w = FloatSlider(min=0.5, max=50.0, step=0.05, value=4)

def graph_g(l: float):
    # выкалываем точку y=-0.5, чтобы не было деления на 0 и ветки гиперболы не соединялись
    eps = 10**(-10)
    y_left = np.linspace(-l, -0.5-eps, 1000)
    y_right = np.linspace(-0.5+eps, l, 1000)
    g_left  = (1 + y_left) / (1 + 2*y_left)
    g_right = (1 + y_right) / (1 + 2*y_right)

    fig, ax = plt.subplots(figsize=(5, 5))
    ax.plot(y_left, g_left, color='k')
    ax.plot(y_right, g_right, color='k')

    # выделяем область y>0
    highlighting = plt.Rectangle( (0, -l-1),  2*l+2, 2*l+2, color='yellow', alpha=0.5 )
    ax.add_artist(highlighting)

    ax.set_title("График функции $g=g(y^*)$")

    ax.set_xlabel("$y^*$")
    ax.set_ylabel("$g(y)$")

    ax.set_ylim( (-l, l) )
    ax.grid()

    plt.show()

interact(graph_g, l=l_w);
````

Из графика видно, что в интересующей нас области $y^*>0$ выполняется условие $1/2\leq g(y^*)\leq 1$. Поэтому окончательно

$$\frac{1}{2}\delta(x^*)\leq\delta(y^*)\leq \delta(x^*).$$

Уточнение измерения $y^*$ до такой степени, что $\delta(y^*)<\delta(x^*)/2$, не имеет смысла.

## Представление чисел в IEEE-арифметике

IEEE-арифметика, в частности стандарт IEEE754, описывает способ представления чисел с **плавающей точкой** в компьютерах. Все числа в такой арифметике заданы с одинаковой **относительной** точностью. Наоборот, в представлении чисел с **фиксированной точкой** все числа заданы с одинаковой **абсолютной** погрешностью.

В системе счисления с основанием $k$ любое число $a$ можно представить в **экспоненциальном виде**:

$$a_k=(-1)^s\cdot M_k\cdot k^p,$$

где $s\in\{0,1\}$ — знак, $M_k\in\mathbb{R}_+$ — **мантисса**, $p\in \mathbb{Z}$ — порядок.

**Определение 3.** Говорят, что число $a_k$ представлено в *нормализованном виде*, если его мантисса $M_k$ удовлетворяет условию $1\leq M_k<k$.

Очевидно, что при вычислениях с мантиссой и порядком **бесконечно возможной разрядности** к нормализованному виду можно привести **любое** число.

Как известно, все компьютерные вычисления проводятся в двоичной арифметике, поскольку с технической точки зрения она наиболее удобна для реализации вычислительного устройства. Поэтому нас будет интересовать вид нормализованного числа в двоичной системе счисления. По определению имеем:

\begin{equation}
  a_2^\text{norm}=(-1)^s\cdot M_2\cdot 2^p,\quad 1\leq M_2<2.
\end{equation}

Так как $1\leq M_2<2$, старший разряд мантиссы всегда равен 1: $M_2=1,..._2$ (**неявная единица** — implicit one), поэтому в памяти компьютера достаточно хранить только дробную часть мантиссы $f_2$: $M_2=1+f_2$, $0\leq f_2<1$.

В силу ограниченности памяти мантисса не может быть сколь угодно длинной. Предположим, что на хранение $f_2$ выделено $m$ двоичных разрядов (бит). Тогда $f_2=f\cdot 2^{-m}$, где $f$ — $m$-битное целое неотрицательное число.

Таким образом, $a_2^\text{norm}=(-1)^s\cdot (1+f\cdot 2^{-m})\cdot 2^p$, причём $f\in[0,2^{m}-1]$.

Пусть на хранение порядка $p$ выделено $n$ бит. С помощью $n$ бит можно задать $2^n$ целых неотрицательных чисел из отрезка $[0,2^n-1]$. Однако $p$ может быть и отрицательным. Чтобы не хранить отдельно знак порядка, сдвинем $p$ на $x$ единиц: $e=p+x$. В памяти будем хранить смещённый порядок — $n$-битное целое число $e\in[0,2^n-y]$, $y\in\mathbb{N}$. Тогда $p\in [-x, 2^n-y-x]$.

Величину смещения $x<2^n-y$ выбираем из условия равенства модулей границ диапазона для $p$: $x=2^n-y-x$. Тогда $x=2^{n-1}-y/2$. При этом $p=e-2^{n-1}+y/2$, $p\in[-2^{n-1}+y/2,2^{n-1}-y/2]$. Параметр $y$ выбираем наименьшим из возможных (он должен быть чётным): $y=2$.

Окончательно: $p=e-2^{n-1}+1$, $e\in[0, 2^n-2]$, т.е. $p$ может принимать $2^n-1$ различных целых значений. При отдельном хранении знака их было бы $2^n$ за счёт двух нулей: $+0$ и $-0$.

Итак, нормализованное двоичное число $a_2^\text{norm}$ в IEEE-арифметике задаётся тремя величинами:
* знаковым битом $s\in\{0,1\}$,
* $m$-битным целым числом (дробной частью мантиссы) $f\in[0,2^m-1]$,
* $n$-битным целым числом (смещённым порядком): $e\in[0,2^n-2]$.

Отметим, что в действительности в качестве нижней границы диапазона для $e$ выбирается не $0$, а $1$. Это делается для того, чтобы отделить множество нормализованных чисел от нуля "щелью" из так называемых **субнормальных чисел** (как раз у них $e=0$). Субнормальные числа необходимы для реализации  округления. Кроме того, они предотвращают возникновение ряда ошибок при арифметических операциях на границах диапазона нормализованных чисел. За подробностями отсылаем читателя к курсам информатики и вычислительной техники.

По трём параметрам $s$, $f$ и $e$ число восстанавливается в соответствии с формулой:
  \begin{equation}
    a_2^\text{norm}=(-1)^s\cdot (1+f\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1}.
  \end{equation}

Заметим, что нуль нельзя представить в нормализованном виде из-за использования неявной единицы. Поэтому его вводят в IEEE-арифметику искусственно: если $e=0$ (все $n$ бит нулевые) и $f=0$ (все $m$ бит нулевые), то число полагается равным $\pm 0$ (**машинный нуль**), в зависимости от значения знакового бита.

Если $e=2^n-1$ (все $n$ бит единичные), а $f=0$ (все $m$ бит нулевые), то число полагается равным $\pm\infty$ (**машинная бесконечность**), в зависимости от значения знакового бита.

Если же $e=2^n-1$ и $f\neq 0$ (среди $m$ бит есть ненулевые), то число заменяется объектом "нечисло" (NaN — Not-a-Number). Иначе говоря, из арифметики исключаются числа, большие $+\infty$ или меньшие $-\infty$.

С представлением чисел в IEEE-арифметике связаны **три машинные константы**:

* Over Flow Limit (OFL) — **порог переполнения**, т.е. максимальное число, представимое в машинной арифметике.

* Under Flow Limit (UFL) — **порог машинного нуля**, т.е. наименьшее (по модулю) **нормализованное** число, представимое в машинной арифметике.

* Относительная машинная точность ("**машинное эпсилон**") — максимальное число, обеспечивающее справедливость равенства $1+\varepsilon_\text{маш}=1$ в машинной арифметике.

Пользуясь выведенной формулой для $a_2^\text{norm}$, находим:

\begin{equation}
  \text{OFL}=\max (a_2^\text{norm})=[1+(2^m-1)\cdot 2^{-m}]\cdot 2^{2^n-2-2^{n-1}+1}=[1-2^{-(m+1)}]\cdot 2^{2^{n-1}}\underset{\strut{m\,\gtrsim\, 10}}{\simeq} 2^{2^{n-1}},
\end{equation}

\begin{equation}
  \text{UFL}=\min (a_2^\text{norm})=[1+0\cdot 2^{-m}]\cdot 2^{1-2^{n-1}+1}=2^{2-2^{n-1}}.
\end{equation}

Из сказанного следует, что множество нормализованных чисел конечно. Его можно представлять себе как некоторую одномерную сетку с очень большим, но конечным набором узлов. Эта сетка заменяет собой непрерывную числовую прямую. Если число, задаваемое извне или полученное как результат расчёта, не попадает в узел сетки, то его приходится округлять до ближайшего числа, попадающего в узел. Возникает **ошибка округления** — специфическая погрешность, характерная для любых компьютерных расчётов.

Чтобы оценить ошибку округления, найдём расстояние (шаг) между двумя соседними нормализованными числами **одного порядка**. Легко видеть, что это расстояние определяется как

$$h(e)=(1+(f+1)\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1}-(1+f\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1}=2^{e-2^{n-1}-m+1}.$$

Здесь $0\leq f< 2^m-1$. Видно, что при фиксированных $n$ и $m$ шаг $h$ оказывается тем больше, чем выше порядок числа. Эту закономерность нетрудно понять: так получается потому, что числа, представимые в IEEE-арифметике, **распределелены на числовой оси неравномерно** — между соседними степенями двойки заключено одинаковое количество чисел.

Очевидно, что погрешность округления будет наибольшей, если число попало строго в центр шага между узлами сетки нормализованных чисел. Поэтому для погрешности округления числа со смещённым порядком $e$ можно дать следующую оценку:

$$\Delta^\text{r}_e\leq h(e)/2=2^{e-2^{n-1}-m}.$$

Наибольшей эта оценка окажется для максимально возможного порядка:

$$\Delta^\text{r}_e\leq \Delta^\text{r}=\max_e\big[h(e)/2\big]=2^{2^n-2-2^{n-1}-m}=2^{2^{n-1}-m-2}.$$

Можно перейти к относительным величинам:

$$\delta^\text{r}_e\leq\frac{\Delta^\text{r}_e}{(1+f\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1}}=\frac{2^{-m-1}}{1+f\cdot 2^{-m}},$$

$$\delta^\text{r}_e\leq\delta^\text{r}=\max_f\big[\delta_e^\text{r}\big]=\frac{1}{2}\cdot 2^{-m}.$$


Видно, что оценка на относительную ошибку не зависит от порядка числа: в каждом порядке она уменьшается от максимального значения $\delta^\text{r}$ для наименьшего числа с $f=0$ до значения

$$\frac{\delta^\text{r}}{2-2^{-m}}<\delta^\text{r}$$

для наибольшего числа с $f=2^m-1$.

Выясним смысл величины $\delta^\text{r}$. Пусть даны два числа $a$ и $b$ такие, что $b>a$. Их относительная разность есть

$$\frac{b-a}{a}=\frac{b}{a}-1.$$

Если эта величина окажется меньше (или равной) $\delta^\text{r}$, то в машинной арифметике числа $a$ и $b$ невозможно будет различить из-за погрешностей округления. Таким образом, **условие различимости** чисел формулируется в виде неравенства

$$\frac{b}{a}-1> \delta^\text{r}\quad \text{или}\quad b>a(1+\delta^\text{r}).$$

Поскольку при $b=a$ это неравенство не удовлетворяется ($\delta^\text{r}>0$), в машинной арифметике числа $a$ и $a(1+\delta^\text{r})$ неразличимы, т.е. $a=a(1+\delta^\text{r})$, или $1+\delta^\text{r}=1$. Отсюда следует, что $\delta^\text{r}=\varepsilon_\text{маш}$. Таким образом, по смыслу $\varepsilon_\text{маш}$ — это **максимальная относительная ошибка, которая может быть допущена при округлении числа в конечно-разрядной арифметике**.

## Числа одинарной и двойной точности

Стандарт IEEE754 (последнее обновление — 2019 г.) определяет два основных представления чисел в IEEE-арифметике: представление с **одинарной точностью** (float) и представление с **двойной точностью** (double). Они отличаются количеством бит, выделенных на хранение мантиссы и порядка числа.

При храненении чисел с одинарной точностью на одно число выделяется $32$ бита (4 байта), причём на мантиссу отведено $m=23$ бита, а на порядок — $n=8$ бит.

При храненении чисел с двойной точностью на одно число выделяется $64$ бита (8 байт), причём на мантиссу отведено $m=52$ бита, а на порядок — $n=11$ бит.

В таблице ниже приведены основные характеристики этих двух представлений. Все они получаются по общим формулам, выведенным выше. Для удобства приведём сводку этих формул:

$$a_2^\text{norm}=(-1)^s\cdot (1+f\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1},\quad s\in\{0,1\},\quad f\in[0,2^m-1],\quad e\in[1,2^n-2];$$

$$\text{OFL}=[1-2^{-(m+1)}]\cdot 2^{2^{n-1}}\underset{\strut{m\,\gtrsim\, 10}}{\simeq} 2^{2^{n-1}},\quad\text{UFL}=2^{2-2^{n-1}},\quad\varepsilon_\text{маш}=\frac{1}{2}\cdot 2^{-m}.$$

Для оценок полезно помнить, что $2^{10}\simeq 10^3$.

**Таблица.** *Сравнение характеристик представлений чисел с одинарной и двойной точностью*

|  | Одинарная точность (float) |  Двойная точность (double) |
|-------|:-------:|:-------:|
| Число бит | $$32=1+23+8$$  | $$64=1+52+11$$ |
| Формула для $a_2^\text{norm}$ | $$(-1)^s\cdot (1+f\cdot 2^{-23})\cdot 2^{e-127}$$ | $$(-1)^s\cdot (1+f\cdot 2^{-52})\cdot 2^{e-1023}$$ |
| OFL   | $$\simeq 3,4\cdot 10^{38}$$ | $$\simeq 1,8\cdot 10^{308}$$ |
| UFL   | $$\simeq 1,2\cdot 10^{-38}$$  | $$\simeq 2,2\cdot 10^{-308}$$ |
| $\varepsilon_\text{маш}$   | $$\simeq 6,0\cdot 10^{-8}$$  | $$\simeq 1,1\cdot 10^{-16}$$   |

Отметим, что при работе с числами на границе диапазона машинной арифметики (или числами, выходящими за эти границы) может возникнуть потребность в использовании специальных библиотек. Например, попытка вычислить значение OFL для двойной точности вызывает ошибку переполнения:

````python
try:
    (1 - 2**(-53)) * 2**(1024) # OFL для представления с двойной точностью
except Exception as e:
    print(e)
````

Любопытно, что **по отдельности** числа $2^{-53}$ и $2^{1024}$ обрабатываются без затруднений:

````python
2**(-53)
````

````python
2**1024
````

Проблема при вычислении OFL возникает из-за необходимости конвертации целого числа $2^{1024}$ в число с плавающей точкой:



````python
try:
    float(2**1024)
except Exception as e:
    print(e)
````

Дело в том, что в Python для величин типа ```int``` реализовано **динамическое управление памятью**, благодаря которому целые числа могут быть сколь угодно длинными (верхняя граница дипазона целых чисел не фиксирована на программном уровне и определяется только доступным объёмом памяти). С другой стороны, диапазон чисел типа ```float``` ограничен программно, поэтому очень большие целые числа не могут быть приведены к этому типу напрямую ([документация по встроенным типам Python](https://docs.python.org/3/library/stdtypes.html)). Отметим, что сведения о диапазоне чисел типа ```float``` (и другие параметры типа) можно получить с помощью функции ```float_info``` библиотеки ```sys``` ([документация](https://docs.python.org/3/library/sys.html)):

````python
import sys

sys.float_info
````

Справку по обоим типам (в том числе три машинные константы) можно получить с помощью функции ```np.finfo``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.finfo.html#numpy.finfo)). Обратим внимание на то, что при попытке вызвать её для встроенного типа ```int``` информация о динамическом управлении памятью не отображается: указываются границы диапазона для 64-битного целого числа (т.е., строго говоря, не для типа ```int```, а для типа ```np.int64```):

````python
print(np.iinfo(int))
````

Возвращаясь к примеру с подсчётом OFL, заметим, что указанную выше проблему можно решить элементарно, раскрыв скобки в вычисляемом выражении:

````python
OFL_int = 2**(1024) - 2**(1024-53)
OFL_int
````

Вместо упрощения исходного выражения путём раскрытия скобок (которое, конечно, возможно не всегда) можно было бы воспользоваться типом данных ```Decimal```, предоставляемым одноимённым модулем ```decimal``` ([документация](http://grep.cs.msu.ru/python3.8_RU/digitology.tech/docs/python_3/library/decimal.html)):

````python
getcontext()
````

По умолчанию арифметические операции с числами типа ```Decimal``` выполняются с **фиксированной точностью** в 28 знаков. Эту точность можно менять:

````python
getcontext().prec = 4

OFL_4 = Decimal(1 - 2**(-53)) * Decimal(2**(1024))
OFL_4
````

Настраиваемая фиксированная точность — основная отличительная черта типа ```Decimal```. Она позволяет обрабатывать произвольно большие числа (ограничение связано только с объёмом доступной памяти), поскольку количество точно вычисляемых цифр числа задано наперёд и не увеличивается по мере роста числа. Кроме того, фиксация точности во многих случаях позволяет избежать "артефактов" машинной арифметики, т.к. модуль работает в десятичной системе счисления. Пример:

````python
0.3 - 0.2 - 0.1 # вроде бы нуль?
````

````python
getcontext().prec = 2
Decimal('0.3') - Decimal('0.2') - Decimal('0.1')
````

Нужно, однако, следить за правильностью создания чисел типа ```Decimal```. В ячейке выше числа заданы конвертацией из строк. При попытке заменить строки десятичными дробями (```float```) результат будет другой:

````python
Decimal(0.3) - Decimal(0.2) - Decimal(0.1)
````

С этими и другими тонкостями предлагается ознакомиться самостоятельно по документации, на которую мы ссылались выше.

## Задача 4

Рассмотрим модель представления чисел в IEEE-арифметике следующего вида:

$$S=\{\pm b_0,b_1b_2\cdot 2^{\pm a}\}.$$

Здесь $a,b_1,b_2\in\{0,1\}$, а $b_0=1$ всегда, за исключением случая, когда $a=b_1=b_2=0$; в этом случае $b_0\in\{0,1\}$.

Нарисовать множество $S$ на действительной оси. Определить количество чисел в рассматриваемой арифметике. Вычислить машинные константы OFL, UFL, $\varepsilon_\text{маш}$.

*Примечание.* В рассматриваемой арифметике нет субнормальных чисел, в отличие от "настоящей" IEEE-арифметики.

**Решение.** Пусть сначала $b_0=1$. Составим таблицу перебора всех возможных мантисс:

\begin{gather}
  & b_0 &\quad& b_1 &\quad& b_2 &\quad& \text{В десятичном виде}\\
  M_1& 1   &\quad& 0   &\quad& 0   &\quad&   1   \\
  M_2& 1   &\quad& 0   &\quad& 1   &\quad&   5/4 \\
  M_3& 1   &\quad& 1   &\quad& 0   &\quad&   3/2 \\
  M_4& 1   &\quad& 1   &\quad& 1   &\quad&   7/4 \\
\end{gather}

Для каждого из этих четырёх чисел существует три возможных порядка: $a=\pm 1$ или $a=0$. Таким образом, имеем 12 чисел:

\begin{gather}
  M_i\cdot 2^0 &\quad& M_i\cdot 2^{-1} &\quad& M_i\cdot 2^{+1} \\
   1     &\quad& 1/2   &\quad& 2      \\
   5/4   &\quad& 5/8   &\quad& 5/2    \\
   3/2   &\quad& 3/4   &\quad& 3    \\
   7/4   &\quad& 7/8   &\quad& 7/2    \\
\end{gather}

Каждое из них может иметь один из двух знаков. Кроме того, при $a=b_1=b_2=0$ имеем ещё два числа: нуль ($b_0=0$) и один ($b_1=1$). Но единица уже учтена в таблице, т.е. в рассматриваемой арифметике она дублируется. Таким образом, общее количество чисел (мощность конечного множества $S$):

$$2\cdot 12 + 2 - 1 = 25.$$

Из второй таблицы видно, что $\text{OFL}=7/2$, $\text{UFL}=1/2$. **Машинное эпсилон в силу своего смысла совпадает с половиной последнего разряда**, т.е. в данном случае

$$\varepsilon_\text{маш}=\frac{1}{2}\cdot 2^{-2}=\frac{1}{8}.$$

Его можно было бы найти и по выведенной выше общей формуле, заметив, что $m=2$ (число бит, выделенных на хранение **дробной части** мантиссы).

Остаётся отметить числа из множества $S$ на числовой прямой. Это можно сделать с помощью программы. Тривиальный вариант заключается в ручном задании чисел из таблицы:

````python
import matplotlib.pyplot as plt
import numpy as np

def graph_triv():
    x = np.array([1, 5/4, 3/2, 7/4, 1/2, 5/8, 3/4, 7/8, 2, 5/2, 3, 7/2])
    x = np.concatenate((-x, x, [0])) # добавляем отрицательные и нуль
    y = [0]*25

    fig, ax = plt.subplots(figsize=(10, 2))
    ax.scatter(x, y, color='r', s=20)

    # рисуем числовую ось
    ax.plot([np.min(x)-1/2, np.max(x)+1/2], [0, 0], color='k', zorder=0)

    ax.set_title("Числа из множества $S$")

    plt.yticks([])

    plt.show()

graph_triv()
````

Видно, что распределение чисел неравномерно.

## Постановка задачи

Одной из простейших задач вычислительной математики является **задача численного дифференцирования**. Эта задача обычно является составляющей более сложных задач. Понимать основы численного диференцирования важно в первую очередь для численного решения дифференциальных уравнений (обыкновенных и в частных производных).

С точки зрения математического анализа задача дифференцирования функции $\varphi=\varphi(x)$, определённой на отрезке $[a,b]$, в точке $x=\xi\in[a,b]$ сводится к вычислению предела дифференциального отношения:

$$\varphi'(\xi)=\lim_{x\to \xi}\frac{\varphi(x)-\varphi(\xi)}{x-\xi}.$$

Полагая $x=\xi+\Delta x$, приходим к эквивалентной записи:

$$\varphi'(\xi)=\lim_{\Delta x\to 0}\frac{\varphi(\xi+\Delta x)-\varphi(\xi)}{\Delta x}.$$

В курсах анализа такие пределы вычисляют для некоторого набора функций (составляют **таблицу производных**). Затем доказывают ряд **правил дифференцирования**. После этого задача дифференцирования произвольной функции (в области её определения) сводится к последовательному применению правил: ими пользуются для упрощения дифференцируемого выражения до одной или нескольких табличных функций.

Как уже отмечалось выше, в вычислительной математике любая (непрерывная) функция $\varphi(x)$ заменяется совокупностью своих значений в узлах некоторой **сетки**. В результате получается новый объект — **сеточная функция**.

**Определение 1.** Будем говорить, что на отрезке $[a,b]$ введена *сетка*, если задано *конечное* множество

$$G=\{x_i\in\mathbb{R}, 0\leq i\leq n: a\leq x_0<x_1<...<x_n\leq b\}.$$

Точки $x_i$ будем называть *узлами сетки*, а величины $h_i=x_i-x_{i-1}$, $1\leq i\leq n$ — её *шагами*. Если $h_1=h_2=...=h_n\equiv h$, то сетка называется *равномерной*.

**Определение 2.** Пусть на отрезке $[a,b]$ введена сетка $G$. *Сеточной функцией* называется *конечное* множество пар

$$F=\big\{(f,x): f\in\mathbb{R}, x\in G\big\},$$

в котором каждое значение $x$ принадлежит лишь одной паре. Величины $f$ называются *значениями сеточной функции*.

**Определение 3.** Пусть на отрезке $[a,b]$ определена некоторая гладкая функция $\varphi=\varphi(x)$. *Сеточной проекцией* функции $\varphi$ называется сеточная функция

$$F_\varphi=\big\{(\varphi(x),x): x\in G\big\},$$

где $G$ — некоторая сетка, введённая на отрезке $[a,b]$.

Операцию получения сеточной функции $F_\varphi$ из гладкой (нужное число раз дифференцируемой) функции $\varphi$ называют **проектированием функции на сетку**. Сформулировав третье определение, мы зафиксировали конкретный способ проектирования: он сводится к вычислению конечного набора значений

$$f_i=\varphi(x_i)$$

в узлах $x_i$ некоторой сетки. Вообще говоря, можно рассматривать и другие (более сложные) способы построения сеточных проекций. В нашем курсе мы ограничимся наиболее простым — тем, который следует из третьего определения.

В терминах приведённых выше определений задача численного дифференцирования формулируется следующим образом.

Пусть на отрезке $[a,b]$ определена некоторая гладкая (нужное число раз дифференцируемая) функция $\varphi=\varphi(x)$, а $F_\varphi$ — её сеточная проекция. Требуется определить по $F_\varphi$ сеточную функцию $F'_\varphi$ такую, что $F'_\varphi\simeq F_{\varphi'}$. Точный смысл знака приближённого равенства в последней формуле станет ясен из дальнейшего.

## Простейшие формулы численного дифференцирования

Простейший способ приближённого вычисления производной подсказывается её определением. Составим дифференциальное отношение

$$\frac{\varphi(\xi+\Delta x)-\varphi(\xi)}{\Delta x}.$$

Если величина $\Delta x$ достаточно мала (но конечна!), то это отношение близко к $\varphi'(\xi)$. Формализуем указанный способ вычисления производной.

Зададим на отрезке $[a,b]$ сетку $G$. Здесь и далее мы для простоты будем считать её равномерной — это предположение ничего не меняет с принципиальной точки зрения, но упрощает формулы. Построим сеточную проекцию $F_\varphi$ функции $\varphi$:

$$F_\varphi=\big\{(f_m,x_m):f_m=\varphi(x_m), x_m\in G\big\}.$$

Обозначения $x_m$ и $f_m$ введены для того, чтобы подчеркнуть конечность множества $F_\varphi$. Искомую сеточную функцию $F'_\varphi$ определим так:

$$F'_\varphi=\bigg\{(f_m',x_m):f_m'=\frac{f_{m+1}-f_m}{h},x_m\in G' \bigg\}.$$

Через $h$ здесь обозначен шаг сетки $G$. Обратим внимание на то, что значение $f'_m$ не определено для последнего узла сетки $G$, поскольку при $m=n$ невозможно вычислить $f_{m+1}$, так как достигнута правая граница сетки. В связи с этим в формуле для $F'_\varphi$ вместо $G$ написано $G'$ — имеется в виду сетка из $n$ узлов, совпадающих с первыми $n$ узлами сетки $G$.

Выясним, насколько точен предложенный метод дифференцирования, т.е. оценим **ошибку метода**. Для этого заметим, что

$$f_m'=\frac{f_{m+1}-f_m}{h}=\frac{\varphi(x_{m+1})-\varphi(x_{m})}{h},$$

$$\varphi(x_{m+1})=\varphi(x_{m}+h)=\varphi(x_{m})+\varphi'(x_{m})h+\varphi''(x_{m})\frac{h^2}{2}+O(h^3),$$

$$f_m'=\varphi'(x_{m})+\varphi''(x_{m})\frac{h}{2}+O(h^2).$$

Из последнего равенства следует, что отличие каждого значения $f'_m$ сеточной функции $F'_\varphi$ от соответствующего значения $\varphi(x_m)$ сеточной проекции $F_{\varphi'}$ оказывается тем меньше, чем меньше шаг сетки $h$, т.е. в пределе $h\to 0$ сеточные функции $F'_\varphi$ и $F_{\varphi'}$ совпадают (ошибки округления мы пока не рассматриваем). Именно эту связь между сеточными функциями мы имели в виду, когда писали равенство $F'_\varphi\simeq F_{\varphi'}$ при формулировке задачи численного дифференцирования.

В реальных вычислениях шаг сетки всегда выбирается малым, но конечным, и поэтому разность $f'_m-\varphi'(x_m)$ оказывается ненулевой. С точностью до членов первого порядка малости по $h$ можем написать:

$$|f'_m-\varphi'(x_m)|=|\varphi''(x_{m})|\frac{h}{2}.$$

Положим $M_2=\displaystyle\max_{x\in[a,b]}|\varphi''(x)|\geq\displaystyle\max_m|\varphi''(x_m)|$ (индекс "2" указывает на второй порядок производной). Тогда

$$|f'_m-\varphi'(x_m)|\leq\frac{M_2 h}{2}\equiv \Delta_\text{forw}^{(\text{m})},\quad 0\leq m\leq n-1.$$

Величина $\Delta_\text{forw}^{(\text{m})}$ и является искомой ошибкой метода.

В дальнейшем мы для краткости будем опускать обозначения сеточных функций и сеток, характеризуя метод дифференцирования только формулой для вычисления $f_m'$. Рассмотренная выше формула

$$f_m'=\frac{f_{m+1}-f_m}{h}$$

носит название "**производная вперёд**". Наряду с этой формулой можно рассмотреть также **производную назад**

$$f_m'=\frac{f_{m}-f_{m-1}}{h}$$

и формулу **центральной разности**

$$f_m'=\frac{f_{m+1}-f_{m-1}}{2h}.$$

Читателю предлагается самостоятельно убедиться в том, что для производной назад (как и для производной вперёд)

$$\Delta_\text{back}^{(\text{m})}=\frac{M_2 h}{2},$$

а для формулы центральной разности

$$\Delta_\text{cent}^{(\text{m})}=\frac{M_3 h^2}{6},$$

где $M_3=\displaystyle\max_{x\in[a,b]}|\varphi'''(x)|$.



## Порядки точности, аппроксимации и сходимости

Выше мы оценили погрешность метода для трёх формул численного дифференцирования. Полученные оценки включают константы $M_2$ и $M_3$ — максимумы второй и третьей производных соответственно. Как мы увидим ниже, подобную оценку можно дать для произвольной формулы численного дифференцирования, причём в неё войдёт константа $M_k=\displaystyle\max_{x\in[a,b]}|\varphi^{(k)}(x)|$. Очевидно, что если в качестве функции $\varphi(x)$ взят полином

$$P_n(x)=\sum\limits_{i=0}^{n}a_ix^i,$$

то $M_k=0$ при $n<k$, т.е. рассматриваемый метод численного дифференцирования точен на полиномах со степенью, меньшей $k$. Это даёт основание ввести следующее определение.

**Определение 4.** *Порядком точности формулы численного дифференцирования* называется наибольшая из степеней полиномов, для которых формула даёт точный результат дифференцирования в предположении отсутствия ошибок округления.

Легко видеть, что в приведенных выше примерах **порядок точности совпадает со степенью** $h$ в оценке для погрешности метода, т.е. производные вперёд и назад имеют первый порядок точности, а формула центральной разности — второй.

Хотя определение 4 часто встречается в литературе, мы в дальнейшем редко будем им пользоваться. Дело в том, что для дифференциальных задач вычислительной математики принято вводить более фундаментальные понятия **аппроксимации** и **порядка аппроксимации**. Мы дадим соответствующие общие определения, когда перейдём к изучению теории разностных схем для обыкновенных дифференциальных уравнений. Пока же ограничимся следующим алгоритмическим подходом.

**Алгоритм исследования порядка аппроксимации формулы численного дифференцирования**

**Шаг 1.** Подставить в рассматриваемую формулу вместо $f_m$ сеточную проекцию $\varphi(x_m)$ функции $\varphi$. Иллюстрация из приведённого выше примера:

$$f_m'=\frac{f_{m+1}-f_m}{h}=\frac{\varphi(x_{m+1})-\varphi(x_{m})}{h}=\frac{\varphi_{m+1}-\varphi_{m}}{h}.$$

Замена $f'_m$ на $\varphi(x_m)$ нужна для следующего шага алгоритма, так как **сеточную функцию нельзя разложить в ряд Тейлора**, а достаточно гладкую функцию $\varphi(x)$ — можно.

**Шаг 2.** Воспользоваться разложением в ряд Тейлора вблизи одной выбранной точки сетки (например, $x_m$ или другой удобной) для вычисления значений функции в остальных точках сетки, входящих в формулу. Иллюстрация из приведённого выше примера:

$$\varphi(x_{m+1})=\varphi(x_{m}+h)=\varphi(x_{m})+\varphi'(x_{m})h+\varphi''(x_{m})\frac{h^2}{2}+O(h^3).$$

**Шаг 3.** Подставить полученные разложения в формулу численного дифференцирования и получить оценку вида

$$|f'_m-\varphi'(x_m)|\leq Ch^p,\quad C\in\mathbb{R}_+,\quad p\in \mathbb{N}$$

с не зависящей от параметров сетки константой $C$. Число $p$ называется **порядком аппроксимации**.


Проще говоря, в случае задачи численного дифференцирования **порядок аппроксимации — это степень шага сетки** $h$ в лидирующем члене погрешности метода. Подчеркнём, что порядок аппроксимации метода может как совпадать с порядком точности, так и отличаться от него.

Ниже мы рассмотрим **метод неопределённых коэффициентов**, позволяющий легко построить формулу численного дифференцирования любого заданного порядка аппроксимации. Здесь же остановимся на следующем вопросе: **есть ли смысл строить и использовать формулы высоких порядков аппроксимации?**

Пусть перед нами стоит задача численного дифференцирования некоторой функции. Предположим, что мы вычисляем её производную по формуле первого порядка аппроксимации. Пусть выбран достаточно крупный шаг сетки $h$. При этом ошибка метода составляет $Ch$. Изменим шаг $h$ — например, уменьшим его вдвое. Тогда и ошибка метода уменьшится вдвое.

Если при тех же условиях мы будем использовать метод второго порядка аппроксимации, то в результате уменьшения шага сетки в два раза ошибка метода уменьшится уже в $2^2=4$ раза. В случае же метода произвольного порядка аппроксимации $p$ при уменьшении шага сетки вдвое ошибка уменьшается в $2^p$ раз.

Как видим, **методы высокого порядка аппроксимации представляют интерес**, так как с их помощью даже на неподробных сетках можно получить результат, сопоставимый по точности с результатом, полученным методом низкого порядка на очень подробной сетке.

С другой стороны, **существует некоторая граница, до которой разумно увеличивать порядок аппроксимации формул численного дифференцирования**. Это видно хотя бы из того, что при классическом способе построения формул порядок аппроксимации оказывается связанным с количеством используемых точек сетки. Ниже мы увидим, что чем выше требуемый порядок аппроксимации, тем больше точек приходится включать в сеточный шаблон. Но формулу с широким шаблоном невозможно использовать для вычисления производной вблизи границ расчётной области. При слишком широком шаблоне и сложной геометрии области количество "недоступных" точек может оказаться велико.

Кроме того, следует помнить, что с увеличением порядка аппроксимации формулы численного дифференцирования повышаются требования на гладкость функции. Например, в оценку погрешности метода сотого порядка аппроксимации вошел бы максимум модуля 101-й производной функции. На практике о степени гладкости функции обычно известно немногое, и часто приходится решать задачи, в которых функции вообще терпят разрывы. Наконец, вычисления не всегда проводятся на равномерных сетках, а строить формулу высокого порядка аппроксимации на неравномерной сетке сложнее.

Говоря о порядке аппроксимации, нельзя не упомянуть также о **порядке сходимости**. Эти порядки, как мы увидим в общей теории разностных схем, связываются **основной теоремой вычислительной математики**, в которой фигурируют три центральных понятия: **аппроксимация**, **устойчивость** и **сходимость**. Их строгие определения будут даны позже, но на семинарах понятиями аппроксимации и сходимости мы будем пользоваться ещё до введения определений. На этом этапе понятия аппроксимации и сходимости (а также соответствующих порядков) кажутся похожими. Важно понимать, в чём состоит принципиальная разница.

**Порядок аппроксимации** определяется по оценке ошибки метода, полученной **аналитически** при помощи разложений в ряд Тейлора в предположении нужной гладкости функции. **Порядок сходимости** определяется по ошибке метода, полученной при помощи **вычислений на компьютере** для конкретной задачи. На этапе тестирования программы обычно используют задачи с известным решением, поэтому ошибку можно легко определить по разности численного и аналитического решений. Чтобы определить порядок сходимости, обычно последовательно запускают программу, реализующую рассматриваемый численный метод, **на совокупности сгущающихся сеток** (например, на сетках с шагами $h$, $h/2$, $h/4$ и т.д.) и исследуют, во сколько раз уменьшается ошибка метода при измельчении сетки. Существует также способ оценки порядка сходимости, применимый в случае задач, не имеющих аналитического решения (именно для таких задач и создаются численные методы). Этот способ мы рассмотрим позже — на семинаре, посвящённом численному интегрированию.

Подчеркнём существенные отличия порядков аппроксимации и сходимости.

* При исследовании порядка аппроскимации оценки сугубо теоретические. При исследовании порядка сходимости оценки получают в реальном расчёте.

* Если функция, с которой приходится иметь дело, достаточно гладкая, то порядок сходимости совпадёт с порядком аппроксимации. Если степень гладкости функции ниже, чем предполагалось в исследовании аппроксимации, то порядок сходимости будет меньше порядка аппроксимации.

* Порядок сходимости в реальной задаче может быть и не целым числом.

## Погрешности округления в задаче численного дифференцирования

Итак, если некоторая формула численного дифференцирования имеет порядок аппроксимации $p$, то ошибка метода $\Delta^{(\text{m})}\sim h^p$. Отсюда как будто бы следует, что нужно выбирать сетку с наименьшим возможным $h$ — тогда погрешность в вычислении производной тоже будет наименьшей. Однако в расчётах на реальном компьютере неизбежны ошибки округления, которые при слишком малом шаге вносят значительную погрешность в результат дифференцирования. Убедимся в этом на конкретных примерах.

### Численные примеры

Проверим, прежде всего, что результат численного дифференцирования существенно зависит от величины шага. Для этого рассмотрим функцию $\varphi(x)=\exp(x)$ на отрезке $[0,l]$. Её производная, как известно, совпадает с самой функцией: $\varphi'(x)=\varphi(x)=\exp(x)$. Зададим равномерную сетку с некоторым шагом $h$ и построим на этой сетке разностный аналог функции $\varphi'(x)$ — производную вперёд. Результаты отобразим на графике (для начала возьмём $l=1$):

````python
import numpy as np

def exp_deriv(h: float, l: float):
    x = np.arange(0, l, h)
    y = np.exp(x)

    f_n1 = y[1:]
    f_n = y[:-1]

    f_prime = (f_n1 - f_n) / h

    return f_prime, x[:-1]
````

````python
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def graph_exp_deriv(h: float, l: float):
    f_prime, x = exp_deriv(h, l)

    fig, ax = plt.subplots()
    ax.scatter(x, f_prime, s=5, c='b')

    grid_exp = np.arange(0, l, h)
    ax.plot(grid_exp, np.exp(grid_exp), c='r')

    plt.title("Производная функции $\\varphi(x)=\\exp(x)$")
    plt.xlabel("$x$")
    plt.ylabel("$\\varphi'(x)$")

    plt.grid()
    plt.show()

# ползунок для интерактивного графика (виджета)
h_w = FloatSlider(min=10**(-6), max=0.1, step=10**(-3), value=0.1)

interact(lambda h: graph_exp_deriv(h, 1), h=h_w);
````

Попытка запустить отрисовку графика при $h<10^{-6}$ приводит к чрезмерному расходованию памяти. Поэтому для $h<10^{-6}$ возьмём меньший отрезок: $l=10^{-6}$.

````python
# создаём ползунок заново, чтобы не менять график из предыдущей ячейки
h_again = FloatSlider(min=10**(-6), max=0.1, step=10**(-3), value=0.1)

interact(lambda h: graph_exp_deriv(h*10**(-6), 10**(-6)), h=h_again);
````

Как видим, уменьшение шага, сопровождающееся увеличением количества узлов в сетке, **сначала уточняет результат** дифференцирования, **а потом только ухудшает** его: при малых шагах порядка $10^{-12}$ точки группируются в отрезки, совершенно не соответствующие истинному ходу производной.

Приведём другой пример. Будем вычислять дифференциальное отношение

$$\frac{\varphi(\xi+h)-\varphi(\xi)}{h}$$

для функции $\varphi(x)=\sin x$. С уменьшением шага $h$ оно должно приближаться к величине $\varphi '(\xi)=\cos\xi$. По сути, мы пользуемся формулой "производная вперёд" на сетке, состоящей всего из двух точек: $\xi$ и $\xi+h$. При этом ошибка численного дифференцирования есть

$$\Delta(\xi,h)=\Bigg|\frac{\sin(\xi+h)-\sin\xi}{h}-\cos\xi\Bigg|.$$

Условимся рассматривать шаги в диапазоне от $1$ до $10^{-16}$ ($\varepsilon_{маш}$ для двойной точности).

Функция для вычисления массива значений $\Delta=\Delta(\xi, h)$ при заданном $\xi$ и $h$, меняющемся от $1$ до $10^{-16}$:

````python
def deriv_sin(xi: float):
    h = np.logspace(-16, 0, 1000, base=10) # сетка шагов по степеням десятки (от 10^{-16} до 1)
    deriv = (np.sin(xi+h) - np.sin(xi)) / h

    return abs(deriv - np.cos(xi)), h
````

Построим график зависимости ошибки численного дифференцирования от величины шага. Кроме того, выведем на печать минимальную ошибку, значение шага, при котором она достигается, и величину $\varepsilon_\text{маш}$, восстановленную по минимальной ошибке (см. формулу ниже).

````python
# ползунок для интерактивного графика (виджета)
xi_w = FloatSlider(min=0, max=10000, step=0.05, value=np.pi/3)

# функция для построения графика
def graph_sin_deriv(xi: float):
    delta, grid_h = deriv_sin(xi)

    fig, ax = plt.subplots()
    ax.plot(grid_h, delta)

    plt.title("Погрешность производной вперёд $\\Delta(x,h)$ в зависимости от $h$ при $x=\\xi$")
    plt.xlabel("$h$")
    plt.ylabel("$\\Delta(x,h)$")

    plt.xscale('log') # логарифмические (по основанию 10) шкалы по x и y
    plt.yscale('log')

    plt.grid()
    plt.show()

    delta_min = np.min(delta)
    h_opt = grid_h[np.argmin(delta)]
    epsilon_mach = delta_min**2 / 4

    print("Минимальная ошибка: {:.3e}".format(delta_min))
    print("Шаг, при котором достигается минимальная ошибка: {:.3e}".format(h_opt))
    print("Машинное эпсилон: {:.3e}".format(epsilon_mach))

interact(graph_sin_deriv, xi=xi_w);
````

Из графика видно, что с уменьшением шага величина ошибки сначала падает, а потом возрастает, достигая минимума при некотором **оптимальном шаге** $h_\text{opt}$. Стохастический характер роста ("гребёнка" в левой части графика) указывает на то, что при $h<h_\text{opt}$ основной вклад в ошибку дают погрешности округления.

### Теоретическая оценка погрешности округления

Теоретически оценить погрешность округления можно следующим образом. Поскольку $\varepsilon_\text{маш}$ есть **максимальная относительная погрешность округления**, абсолютная погрешность округления $\Delta^{(\text{r})}(f_{m+1})$ величины $f_{m+1}$, входящей в формулу производной вперёд, не превосходит $\varepsilon_\text{маш} f_{m+1}\leq M_0\varepsilon_\text{маш}$, где $M_0=\max\limits_{x\in[a,b]}|\varphi(x)|$. Аналогичная оценка справедлива для $\Delta^{(\text{r})}(f_{m})$. По свойствам (вообще говоря, предельных) погрешностей имеем:

$$\Delta_\text{forw}^{(\text{r})}=\frac{\Delta^{(\text{r})}(f_{m+1})+\Delta^{(\text{r})}(f_m)}{h}=\frac{2M_0\varepsilon_\text{маш}}{h}.$$

Таким образом, полная погрешность численного дифференцирования по формуле производной вперёд есть

$$\Delta_\text{forw}(h)=\Delta_\text{forw}^{(\text{m})}+\Delta_\text{forw}^{(\text{r})}=\frac{M_2 h}{2}+\frac{2M_0\varepsilon_\text{маш}}{h}.$$

При больших $h$ доминирует ошибка метода, растущая линейно с увеличением шага. При малых $h$ существенной становится ошибка округления, нарастающая при $h\to 0$ как $h^{-1}$. Значит, в полном соответствии с численной демонстрацией, существует некое **оптимальное значение шага** $h_\text{opt}$, минимизирующее $\Delta_\text{forw}$. Найдём его:

$$\Delta_\text{forw}'(h)=\frac{M_2}{2}-\frac{2M_0\varepsilon_\text{маш}}{h^2}=0,\quad h_\text{opt}=2\sqrt{\frac{M_0\varepsilon_\text{маш}}{M_2}}.$$

Наименьшая величина ошибки оценивается как

$$\Delta_\text{forw}^{\text{min}}=\Delta_\text{forw}(h_\text{opt})=2\sqrt{M_0M_2\varepsilon_\text{маш}}.$$

Если $\varphi(x)=\sin x$, то $M_0=M_2=1$ и $\Delta_\text{forw}^\text{min}=2\sqrt{\varepsilon_\text{маш}}$. Это даёт возможность оценить величину $\varepsilon_\text{маш}$: $\varepsilon_\text{маш}=(\Delta_\text{forw}^\text{min})^2/4$, что мы и сделали в приведённой выше программе. Заметим, что при $\xi=1$ оценка на $\varepsilon_\text{маш}$ получается явно заниженной, а с увеличением $\xi$ она увеличивается (поскольку растёт минимальная ошибка).

**Упражнение 2.** Объяснить, почему полученное в примере значение $\varepsilon_\text{маш}$ отличается от правильного.

Аналогичный анализ ошибок округления с вычислением оптимального шага можно провести для производной назад и центральной разности. Приведём только ответы: для производной назад формулы те же, что и для производной вперёд, а для формулы центральной разности

$$\Delta_\text{cent}(h)=\Delta_\text{cent}^{(\text{m})}+\Delta_\text{cent}^{(\text{r})}=\frac{M_3 h^2}{6}+\frac{M_0\varepsilon_\text{маш}}{h},\quad h_\text{opt}=\bigg(\frac{3M_0\varepsilon_\text{маш}}{M_3}\bigg)^{1/3}.$$

## Метод неопределённых коэффициентов построения формул численного дифференцирования

Мы видели, что формула производной вперёд

$$f_m=\frac{f_{m+1}-f_m}{h}$$

имеет **первый порядок аппроксимации**, а формула центральной разности

$$f_m=\frac{f_{m+1}-f_{m-1}}{2h}$$

имеет **второй порядок аппроксимации**, т.е. является потенциально более точной. Возникает вопрос: чем вторая формула принципиально отличается от первой? Какое существенное отличие объясняет разный порядок точности формул?

Нетрудно заметить, что в первом случае для вычисления значения производной в узле $x_m$ сетки требуется знать значения функции в узлах $x_m$ и $x_{m+1}$, т.е. в вычислении задействованы "текущий" узел $x_m$ и **один** соседний справа узел $x_{m+1}$. С другой стороны, в формуле центральной разности задействованы **два** соседних узла: узел слева $x_{m-1}$ и узел справа $x_{m+1}$, — а общее количество узлов, рассматриваемых при вычислении производной в данной точке сетки, равно трём ($x_m$, $x_{m-1}$ и $x_{m+1}$). Говорят, что эти формулы отличаются **сеточным шаблоном**: у производной вперёд шаблон двухточечный (точки $x_m$ и $x_{m+1}$), а у центральной разности — трёхточечный (точки $x_m$, $x_{m-1}$ и $x_{m+1}$).

После сказанного естественно выдвинуть следующую гипотезу: **чем больше точек** включает сеточный шаблон, **тем более высокий порядок аппроксимации** имеет построенная на нём формула численного дифференцирования. Далее мы подтвердим эту гипотезу, доказав, что **на шаблоне из $N+1$ точки всегда можно построить единственную формулу численного дифференцирования с порядком аппроксимации, равным $N$**.

Пусть сеточный шаблон включает $i$ точек слева от "текущего" узла $x_m$ и $j$ точек справа от него (т.е. всего в шаблоне $i+j+1\equiv N+1$ точка). Общий вид построенной на этом шаблоне формулы численного дифференцирования:

$$f'_m=\frac{\alpha_{-i}f_{m-i}\,+\, ...\,+\,\alpha_{-1}f_{m-1}+\alpha_0f_m+\alpha_1f_{m+1}\,+\,...\,+\,\alpha_jf_{m+j}}{h}=\frac{1}{h}\sum\limits_{k=-i}^{j}\alpha_k f_k.$$

Множитель $1/h$ включён в формулу из размерных соображений. Среди коэффициентов $\alpha_k$ могут быть нулевые. Например, для формулы производной вперёд $i=0$, $j=1$, $\alpha_0=-1$, $\alpha_1=1$; для формулы центральной разности $i=j=1$, $\alpha_0=0$, $\alpha_1=1$, $\alpha_{-1}=-1$. Значения сеточной функции $f_k$ получены в результате проектирования гладкой функции $\varphi$ на сетку, поэтому $f_k=\varphi(x_k)$, т.е.

$$f'_m=\frac{1}{h}\sum\limits_{k=-i}^{j}\alpha_k \varphi(x_m+kh).$$

Разложим функцию $\varphi$ в ряд Тейлора в окрестности точки $x_m$:

$$\varphi(x_m+kh)=\sum\limits_{n=0}^q\frac{\varphi^{(n)}(x_m)}{n!}(kh)^n+O(h^{q+1}).$$

Подставим это разложение в формулу для $f'_m$:

$$f'_m=\sum\limits_{n=0}^q\varphi^{(n)}(x_m)h^{n-1}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).$$

Выпишем явно первые два члена суммы по $n$:

$$f'_m=\frac{\varphi(x_m)}{h}\sum\limits_{k=-i}^j\alpha_k+\varphi'(x_m)\sum\limits_{k=-i}^j\alpha_k k +\sum\limits_{n=2}^q\varphi^{(n)}(x_m)h^{n-1}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).$$

**Потребуем**, чтобы

$$\sum\limits_{k=-i}^j\alpha_k=0,\quad \sum\limits_{k=-i}^j\alpha_k k=1.$$

Первое условие необходимо для того, чтобы исключить из правой части слагаемое, содержащее значение функции $\varphi(x_m)$ и не убывающее при измельчении сетки. Второе условие обеспечивает аппроксимацию первой производной функции. Тогда

$$f'_m=\varphi'(x_m)+\sum\limits_{n=2}^q\varphi^{(n)}(x_m)h^{n-1}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).$$

Пусть требуется получить формулу численного дифференцирования с порядком аппроксимации $p<q$. Это значит, что в сумме по $n$ первый ненулевой член должен включать $h^p$, т.е. все члены с $2\leq n<p+1$ должны обратиться в нуль. В силу произвольности функции $\varphi$ и шага $h$ это требование даёт $p-1$ условие на коэффициенты $\alpha_k$ **в дополнение к двум наложенным выше условиям** (множитель $1/n!$ опускаем, т.к. он отличен от нуля):

$$\sum\limits_{k=-i}^j\alpha_k k^n=0\quad\text{для }n=2,3,...,p.$$

Таким образом, всего имеем $p+1$ условие на $N+1$ коэффициентов $\alpha_k$. Каждое условие представляет собой линейное уравнение на $\alpha_k$, т.е. мы имеем **систему из $p+1$ линейного алгебраического уравнения с $N+1$ неизвестной** (СЛАУ). Как известно, существование единственного решения можно гарантировать, если $p+1=N+1$ и матрица системы не является вырожденной. Если $p+1>N+1$ (переопределённая система), то выбранный **шаблон недостаточно широк**. Если же $p+1<N+1$ (недоопределённая система), то выбранный **шаблон избыточен**. Поэтому далее мы считаем, что $p+1=N+1$, т.е. $p=N$.

Полученная СЛАУ может быть записана в векторно-матричном виде следующим образом:

\begin{equation}
  \begin{pmatrix}
    1  & 1 & \dots & 1 & 1\\
    -i & -i+1 & \dots & j-1 & j\\
    (-i)^2 & (-i+1)^2 & \dots & (j-1)^2 & j^2 \\
    \vdots & \vdots & \ddots & \vdots & \vdots\\
    (-i)^p & (-i+1)^p & \dots & (j-1)^p & j^p
  \end{pmatrix}\,
  \begin{pmatrix}
    \alpha_{-i} \\
    \alpha_{-i+1} \\
    \vdots\\
    \alpha_{j-1}\\
    \alpha_{j}
  \end{pmatrix}=
  \begin{pmatrix}
    0 \\
    1 \\
    \vdots\\
    0\\
    0
  \end{pmatrix}.
\end{equation}

Детерминант матрицы системы имеет вид **определителя Вандермонда** порядка $p$, для которого в курсе линейной алгебры доказывается (по индукции) формула:

\begin{equation}
  \begin{vmatrix}
    1 & x_1 & \dots & x_1^{p-1}\\
    1 & x_2 & \dots & x_2^{p-1}\\
    \vdots & \vdots & \ddots & \vdots\\
    1 & x_p & \dots & x_p^{p-1}
  \end{vmatrix}=\prod\limits_{1\leq b<a\leq p}(x_a-x_b).
\end{equation}

В нашем случае в роли $x_a$ и $x_b$ выступают целые числа от $-i$ до $j$ включительно. Поскольку среди них нет повторяющихся (каждый узел входит в шаблон единожды), соответствующий определитель Вандермонда отличен от нуля. Таким образом, рассматриваемая СЛАУ имеет единственное **нетривиальное** (поскольку правая часть отлична от нуля) решение.

Полученный результат можно легко обобщить на неравномерную сетку. Сформулируем его в виде теоремы.


**Теорема 1.** Пусть на отрезке $[a,b]$ определена гладкая функция $\varphi$ и введена (вообще говоря, неравномерная) сетка $G$. Тогда на сеточном шаблоне из $N+1$ точки (с фиксированной "текущей" точкой) можно построить *единственную* формулу численного дифференцирования, аппроксимирующую первую производную $\varphi'$ с порядком аппроксимации $N$.

Отметим, что при одинаковом общем количестве точек шаблоны могут отличаться расположением "текущей" точки. Например, и производная вперёд, и производная назад строятся по двум точкам, но в первом случае $i=0$, $j=1$, а во втором случае $i=1$, $j=0$, т.е. у этих формул **разные** двухточечные шаблоны.

Из доказательства теоремы также следует, что **минимальное число точек в шаблоне равно двум**, поскольку на коэффициенты $\alpha_k$ приходится накладывать как минимум два условия: $\sum\alpha_k=0$ и $\sum\alpha_k k=1$.

Напишем программу "**Калькулятор формул численного дифференцирования**", выдающую формулу заданного порядка аппроксимации $p$ на минимальном шаблоне. Шаблон будем характеризовать числами $i$ и $j$, как и выше.

Функция для вычисления коэффициентов:

````python
import numpy as np
from numpy.polynomial.polynomial import polyvander

def coeffs(p: int, i: int, j: int):
    template = [k for k in range(-i, j+1, 1)] # задаём шаблон

    A = polyvander(template, p) # строим матрицу Вандермонда
    A = A.transpose() # транспонируем матрицу

    f = np.zeros(p+1) # задаём правую часть
    f[1] = 1

    return np.linalg.solve(A, f)
````

Можно проверить работоспособность написанной функции на трёх известных нам примерах:

````python
coeffs(1, 0, 1) # производная вперёд
````

````python
coeffs(1, 1, 0) # производная назад
````

````python
coeffs(2, 1, 1) # центральная разность
````

Программа выдаёт правильные коэффициенты, начиная с самого левого узла ($k=-i$). Попробуем получить, например, формулу четвёртого порядка аппроксимации на шаблоне с $i=j=2$:

````python
coeffs(4, 2, 2) # формула O(h^4)
````

Теоретическое рассмотрение (см. задачу 1 ниже) показывает, что соответствующая формула имеет вид:

$$f'_m=\frac{f_{m-2}-8f_{m-1}+8f_{m+1}-f_{m+2}}{12h}.$$

Было бы удобнее и в программе получать коэффициенты формул в рациональном виде, а не в виде округлённых десятичных дробей. Кроме того, желательно добавить в начало программы проверку корректности задачи: формула может быть построена при выполнении условия $i+j=p$.

Усовершенствованная версия калькулятора:

````python
from fractions import Fraction

class TemplateError(Exception):
    pass

def FDC_calculator(p: int, i: int, j: int):
    if p != i + j:
        raise TemplateError('Заданный шаблон не соответствует требуемому порядку аппроксимации')

    # функция, преобразующая элементы массива в обыкновенные дроби
    func = np.vectorize(lambda x: str(Fraction(x).limit_denominator()))
    coeffs_frac = func(coeffs(p, i, j))

    return coeffs_frac

try:
    print(FDC_calculator(4, 2, 2)) # формула O(h^4)
except TemplateError as e:
    print(e)
````

Советуем обратить внимание на модуль ```fractions``` ([документация](https://docs.python.org/3/library/fractions.html)), который, как и модуль ```decimal```, рассмотренный выше, предоставляет некоторые методы точной обработки чисел ("борется" с "артефактами" машинной арифметики). В частности, модуль ```fractions``` делает доступными точные операции с рациональными числами.

## Задача 5

Методом неопределённых коэффициентов построить формулу для численного нахождения **первой** производной гладкой функции $\varphi$ в точке $x_m$ равномерной сетки по значениям функции в точках

$$x_{m-2},\quad x_{m-1},\quad x_{m},\quad x_{m+1},\quad x_{m+2}.$$

Оценить погрешность численного дифференцирования и оптимальный шаг сетки.

**Решение.** Вводим пять неопределённых коэффициентов (по числу точек) и составляем выражение для $f'_m$:

$$f'_m=\frac{Af_{m-2}+Bf_{m-1}+Cf_m +Df_{m+1}+Ef_{m+2}}{h}.$$

Раскладываем каждый член числителя по формуле Тейлора. Поскольку задействовано пять точек, порядок аппроксимации не может быть больше четырёх, т.е. **для определения коэффициентов** в формуле для $f'_m$ достаточно получить члены вплоть до $h^3$ (члены с $h^4$ точно не уйдут, каковы бы ни были коэффициенты). Значит, числитель достаточно разложить до $h^4$ (с учётом последующего деления на $h$). Однако по условию **нужно также оценить погрешность**, поэтому раскладывать будем до пятого порядка по $h$: тогда после нахождения коэффициентов мы сможем сразу определить константу, стоящую при $h^4$ (она необходима для оценки погрешности метода). Имеем (остаточные члены для краткости опускаем):

\begin{align}
  f_{m-2}&=\varphi(x_m-2h)=f_m-2hf'_m+\frac{(2h)^2}{2}f''_m-\frac{(2h)^3}{6}f'''_m+\frac{(2h)^4}{24}f^{(4)}_m-\frac{(2h)^5}{120}f^{(5)}_m;\\[0.2cm]
  f_{m-1}&=\varphi(x_m-h)=f_m-hf'_m+\frac{h^2}{2}f''_m-\frac{h^3}{6}f'''_m+\frac{h^4}{24}f^{(4)}_m-\frac{h^5}{120}f^{(5)}_m;\\[0.2cm]
  f_{m+1}&=\varphi(x_m+h)=f_m+hf'_m+\frac{h^2}{2}f''_m+\frac{h^3}{6}f'''_m+\frac{h^4}{24}f^{(4)}_m+\frac{h^5}{120}f^{(5)}_m;\\[0.2cm]
  f_{m+2}&=\varphi(x_m+2h)=f_m+2hf'_m+\frac{(2h)^2}{2}f''_m+\frac{(2h)^3}{6}f'''_m+\frac{(2h)^4}{24}f^{(4)}_m+\frac{(2h)^5}{120}f^{(5)}_m.
\end{align}

Подставляем эти разложения в формулу для $f'_m$. Группируем члены с одинаковыми степенями $h$:

\begin{align}
  f'_m&=\frac{1}{h}(A+B+C+D+E)f_m+(-2A-B+D+2E)f'_m\,+\\[0.2cm]
  &+\frac{h}{2}(4A+B+D+4E)f''_m+\frac{h^2}{6}(-8A-B+D+8E)f'''_m\,+\\[0.2cm]
  &+\frac{h^3}{24}(16A+B+D+16E)f^{(4)}_m+\frac{h^4}{120}(-32A-B+D+32E)f^{(5)}_m.
\end{align}

Сравнивая левую и правую части, составляем систему уравнений:

\begin{equation}
  \begin{cases}
    A+B+C+D+E=0,\\
    -2A-B+D+2E=1,\\
    4A+B+D+4E=0,\\
    -8A-B+D+8E=0,\\
    16A+B+D+16E=0.
  \end{cases}
\end{equation}

Вычитая из пятого уравнения третье, находим:

$$12A+12E=0,\quad A=-E.$$

Подставляя это в третье уравнение, выясняем, что

$$B+D=0,\quad B=-D.$$

Тогда из первого уравнения видно, что $C=0$, а четвёртое и второе уравнения принимают вид:

$$-16A-2B=0,\quad -4A-2B=1.$$

Отсюда

$$A=\frac{1}{12},\quad B=-\frac{8}{12}.$$

Соответственно,

$$E=-\frac{1}{12},\quad D=\frac{8}{12}.$$

Итак, получаем следующую формулу:

$$f'_m=\frac{f_{m-2}-8f_{m-1}+8f_{m+1}-f_{m+2}}{12h}.$$

При этом первый нетривиальный член погрешности метода имеет вид:

$$\frac{h^4}{120}(-32A-B+D+32E)f^{(5)}_m=\frac{h^4}{30}f^{(5)}_m\leq \frac{M_5h^4}{30}\equiv\Delta^{(\text{m})}.$$

Оценка на погрешность округления следует из вида полученной формулы:

$$\Delta^{(\text{r})}=\frac{1+8+8+1}{12h}M_0\varepsilon_\text{маш}=\frac{3}{2h}M_0\varepsilon_\text{маш}.$$

Полная погрешность:

$$\Delta(h)=\Delta^{(\text{m})}+\Delta^{(\text{r})}=\frac{M_5h^4}{30}+\frac{3}{2h}M_0\varepsilon_\text{маш}.$$

Минимизируя $\Delta(h)$, находим оптимальный шаг:

$$\Delta'(h)=\frac{2M_5h^3}{15}-\frac{3}{2h^2}M_0\varepsilon_\text{маш}=0,\quad h_\text{opt}=\bigg(\frac{45M_0\varepsilon_\text{маш}}{4M_5}\bigg)^{1/5}.$$

Минимальная погрешность:

$$\Delta^\text{min}=\Delta(h_\text{opt})=\frac{M_0\varepsilon_\text{маш}}{h_\text{opt}}\bigg(\frac{M_5h^5_\text{opt}}{30M_0\varepsilon_\text{маш}}+\frac{3}{2}\bigg)=\frac{15M_0\varepsilon_\text{маш}}{8h_\text{opt}}.$$

## Задача 6

Оценить оптимальный шаг (равномерной) сетки и минимальную погрешность при численном дифференцировании функций $y(x)=\sin x$ и $y(x)=\sin 100x$ по формуле производной вперёд и по формуле четвёртого порядка аппроксимации. Рассмотреть случаи вычислений с одинарной и двойной точностью.

**Решение.** Погрешности обеих формул оценивались выше. Приведём полученные оценки ещё раз.

*   Производная вперёд:
$$f_m'=\frac{f_{m+1}-f_m}{h},\quad h_\text{opt}=2\sqrt{\frac{M_0\varepsilon_\text{маш}}{M_2}},\quad \Delta^{\text{min}}=2\sqrt{M_0M_2\varepsilon_\text{маш}}.$$

*   Формула четвёртого порядка аппроксимации:

$$f'_m=\frac{f_{m-2}-8f_{m-1}+8f_{m+1}-f_{m+2}}{12h},\quad h_\text{opt}=\bigg(\frac{45M_0\varepsilon_\text{маш}}{4M_5}\bigg)^{1/5},\quad \Delta^{\text{min}}=\frac{15M_0\varepsilon_\text{маш}}{8h_\text{opt}}.$$

Константы $M_0$, $M_2$ и $M_5$ зависят от рассматриваемой функции. В случае функции $y(x)=\sin x$ имеем:

$$M_0=M_2=M_5=1.$$

Если же $y(x)=\sin 100x$, то

$$M_0=1,\quad M_2=10^4,\quad M_5=10^{10}.$$

При вычислениях с одинарной точностью $\varepsilon_\text{маш}=6\cdot 10^{-8}$. Для двойной точности $\varepsilon_\text{маш}=10^{-16}$.

Подстановку констант в формулы осуществим с помощью программы.




````python
import numpy as np

def forward(M0: float, M2: float, eps: float):
    h_opt = 2 * np.sqrt(M0 * eps / M2)
    delta_min = 2 * np.sqrt(M0 * M2 * eps)

    return h_opt, delta_min

def fourth(M0: float, M5: float, eps: float):
    h_opt = (45 * M0 * eps / (4 * M5) ) ** (1/5)
    delta_min = 15 * M0 * eps / (8 * h_opt)

    return h_opt, delta_min

eps1 = 6 * 10 ** (-8) # одинарная точность
eps2 = 10 ** (-16) # двойная точность
````

````python
def calc(M0: float, M2: float, M5: float):
    h_opt, delta_min = forward(M0, M2, eps1)
    print("Производная вперёд, float: h_opt={:.3e}, delta_min={:.3e}".format(h_opt, delta_min))

    h_opt, delta_min = forward(M0, M2, eps2)
    print("Производная вперёд, double: h_opt={:.3e}, delta_min={:.3e}".format(h_opt, delta_min))

    h_opt, delta_min = fourth(M0, M5, eps1)
    print("Формула IV порядка, float: h_opt={:.3e}, delta_min={:.3e}".format(h_opt, delta_min))

    h_opt, delta_min = fourth(M0, M5, eps2)
    print("Формула IV порядка, double: h_opt={:.3e}, delta_min={:.3e}".format(h_opt, delta_min))
````

````python
print("y(x) = sin(x)")
calc(1, 1, 1)
````

````python
print("y(x) = sin(100x)")
calc(1, 10**4, 10**10)
````

Результаты, полученные в задаче, подтверждают сказанное выше: при использовании формул высокого порядка аппроксимации достаточно точный результат можно получить на относительно неподробных сетках. Оптимальный шаг численного дифференцирования по формуле четвёртого порядка аппроксимации даже в арифметике двойной точности оказался больше оптимального шага численного дифференцирования по формуле первого порядка как при использовании арифметики одинарной точности, так и при использовании арифметики двойной точности (т.е. для расчёта по формуле высокого порядка аппроксимации нужна менее подробная сетка). Минимальная ошибка численного дифференцирования тем меньше, чем больше порядок аппроксимации используемой формулы численного дифференцирования и чем больше памяти выделяется под хранение числа в арифметике (минимальная ошибка в арифметике двойной точности меньше, чем в арифметике одинарной точности).

## Вычисление производных высших порядков

Изложенный выше метод неопределённых коэффициентов очевидным образом обобщается на случай производных порядка $s$. Пусть, как и прежде, сеточный шаблон включает $i$ точек слева от $x_m$ и $j$ точек справа. Общий вид формулы численного дифференцирования:

$$f_m^{(s)}=\frac{1}{h^s}\sum\limits_{k=-i}^{j}\alpha_k \varphi(x_m+kh).$$

После подстановки разложения функции $\varphi$ по формуле Тейлора имеем:

\begin{align}
f_m^{(s)}=\frac{\varphi(x_m)}{h^s}\sum\limits_{k=-i}^j\alpha_k+\frac{\varphi'(x_m)}{h^{s-1}}\sum\limits_{k=-i}^j\alpha_kk\,&+\,...\,+\,\varphi^{(s)}(x_m)\sum\limits_{k=-i}^j\frac{\alpha_kk^s}{s!}+\\
&+\sum\limits_{n=s+1}^q\varphi^{(n)}(x_m)h^{n-s}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).
\end{align}

Налагаем $s+1$ условие:

$$\sum\limits_{k=-i}^j\alpha_k=0,\quad \sum\limits_{k=-i}^j\alpha_kk=0,\quad ... \quad \sum\limits_{k=-i}^j\alpha_kk^{s-1}=0,\quad \sum\limits_{k=-i}^j\frac{\alpha_kk^s}{s!}=1.$$

Получаем:

$$f_m^{(s)}=\varphi^{(s)}(x_m)+\sum\limits_{n=s+1}^q\varphi^{(n)}(x_m)h^{n-s}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).$$

Если требуется построить формулу $p$-го порядка аппроксимации, то нужно занулить ещё $p-1$ член в сумме по $n$ добавлением условий вида

$$\sum\limits_{k=-i}^j\alpha_k k^n=0\quad\text{для }n=s+1,...,s+p-1.$$

Таким образом, полное число условий на коэффициенты $\alpha_k$ оказывается равным $p+s$. Ясно, что матрица получившейся СЛАУ снова имеет вид матрицы Вандермонда (порядка $p+s$). Изменяется только вектор правой части: это вектор длины $p+s$, все компоненты которого, за исключением одной, нулевые; на $(s+1)$-м месте в этом векторе стоит $s!$.

Итак, мы приходим к результату, который, на первый взгляд, совершенно аналогичен теореме 1: если количество точек в шаблоне $N+1$ совпадает с комбинацией $p+s$, то на этом шаблоне можно построить единственную формулу численного дифференцирования $p$-ого порядка аппроксимации для $s$-й производной. Отсюда получаем связь: $p=N+1-s$. Однако нужно обратить внимание на один частный случай, в котором порядок аппроксимации построенной формулы оказывается на единицу выше, чем $N+1-s$. Речь идёт о формулах для производных **чётного порядка**, построенных на **симметричных** шаблонах.

Чтобы понять, почему в этом специальном случае порядок аппроксимации оказывается на единицу выше ожидаемого, необходимо заметить, что **при чётном порядке производной и симметричном шаблоне неопределённые коэффициенты тоже симметричны** (попарно равны), т.е. $\alpha_{k}=\alpha_{-k}$.

Действительно, пусть рассматриваемый шаблон симметричен, т.е. $i=j$ (откуда $N=2j$). Легко видеть, что тогда при **нечётных** $n$ всем условиям вида

$$\sum\limits_{k=-i}^j\alpha_k k^n=\sum\limits_{k=-j}^j\alpha_k k^n=0$$

можно удовлетворить, положив коэффициенты попарно равными: $\alpha_{j}=\alpha_{-j}$, $\alpha_{j-1}=\alpha_{-(j-1)}$ и т.д. Так можно поступить только в том случае, если порядок производной $s$ **чётен**: $s=2r$, иначе мы придём к противоречию с условием

$$\sum\limits_{k=-i}^j\frac{\alpha_kk^s}{s!}=\sum\limits_{k=-j}^j\frac{\alpha_kk^s}{s!}=1.$$

Выбрав коэффициенты попарно равными, мы переходим от СЛАУ из $N+1=2j+1$ уравнения с $2j+1$ неизвестной к СЛАУ меньшей размерности из $j+1$ уравнения (только с чётными $n$) с $j+1$ неизвестной ($\alpha_0$, ..., $\alpha_j$):

$$\sum\limits_{k=0}^j\alpha_k=0,\quad \sum\limits_{k=0}^j\alpha_kk^2=0,\quad ... \quad \sum\limits_{k=0}^j\alpha_kk^{2r-2}=0,\quad \sum\limits_{k=0}^j\frac{\alpha_kk^{2r}}{(2r)!}=\frac{1}{2},$$

$$\sum\limits_{k=0}^j\alpha_k k^n=0\quad\text{для }n=2r+2,...,2j.$$

Здесь учтена связь $p=N+1-s=2(j-r)+1$. Матрица СЛАУ меньшей размерности имеет вид

\begin{equation}
  \begin{pmatrix}
    1 & 1 & \dots & 1 & 1\\
    0 & 1 & \dots & (j-1)^2 & j^2 \\
    0 & 1 & \dots & (j-1)^4 & j^4 \\
    \vdots & \vdots & \ddots & \vdots & \vdots\\
    0 & 1 & \dots & (j-1)^{2j} & j^{2j}
  \end{pmatrix}
\end{equation}

и, очевидно, не является вырожденной. Поскольку, кроме того, правая часть СЛАУ меньшей размерности отлична от нуля, существует единственное нетривиальное решение. Таким образом, коэффициенты $\alpha_0$, ..., $\alpha_j$ определены оставшимися уравнениями однозначно, а с учётом попарных равенств однозначно определены все коэффициенты $\alpha_k$ (в количестве $2j+1$).

В силу единственности решения исходной СЛАУ (с матрицей Вандермонда) найденные указанным способом коэффициенты $\alpha_k$ совпадают с теми, которые мы получили бы, не делая исходного предположения о симметричности коэффициентов.

Итак, мы увидели, что **в случае симметричного шаблона и чётного порядка производной искомые коэффициенты симметричны (попарно равны)**. Заметим теперь, что согласно **общей** формуле для $f_m^{(s)}$ первый нетривиальный член погрешности метода имеет вид:

$$\varphi^{(s+p)}(x_m)h^{p}\sum\limits_{k=-i}^j\frac{\alpha_k k^{s+p}}{(s+p)!}.$$

Но в интересующем нас частном случае $s+p=2j+1$, т.е. рассматриваемый член включает сумму

$$\sum\limits_{k=-j}^j\alpha_k k^{2j+1}.$$

Она равна нулю в силу существующих между коэффициентами попарных равенств (выше мы уже отмечали это для всех сумм с нечётными $n$). Таким образом, хотя ожидаемый порядок аппроксимации есть $p=N+1-s$, симметричность коэффициентов автоматически обеспечивает равенство нулю члена погрешности с $h^p$, так что реальный порядок аппроксимации повышается до $N+2-s$. Иначе говоря, благодаря специальному виду шаблона, удаётся удовлетворить не $p+s$ уравнениям на коэффициенты, а $p+s+1$ уравнению.

Аналогичными рассуждениями можно показать, что при **нечётном** порядке производной формула, построенная на симметричном шаблоне, имеет **антисимметричные коэффициенты**: $\alpha_k=-\alpha_{-k}$ (откуда, между прочим, $\alpha_0=0$). Но легко убедиться, что к повышению порядка аппроксимации на единицу это не приводит (см. задачу 1, в которой приведён пример как раз такого случая).

Итак, мы доказали следующую теорему.

**Теорема 2.** Пусть на отрезке $[a,b]$ определена гладкая функция $\varphi$ и введена (вообще говоря, неравномерная) сетка $G$. Тогда на сеточном шаблоне из $N+1$ точки (с фиксированной "текущей" точкой) можно построить *единственную* формулу численного дифференцирования, аппроксимирующую производную $\varphi^{(s)}$. При этом порядок аппроксимации $p$ оказывается равным $N+1-s$. Если шаблон симметричен, а порядок производной $s$ чётен, то порядок аппроксимации повышается до $N+2-s$.

Из сказанного следует простое правило, которым полезно пользоваться при решении задач: **если из числа точек в шаблоне вычесть порядок производной, то получится порядок аппроксимации формулы**, которую можно построить на данном шаблоне для данной производной. Нужно только помнить, что при симметричном шаблоне и чётном порядке производной найденный порядок аппроксимации надо увеличить на единицу.

Из теоремы 2 также вытекает, что **минимальный шаблон, необходимый для аппроксимации $s$-й производной, должен включать $s+1$ точку**.

## Задача 7

Методом неопределённых коэффициентов построить формулу для численного нахождения **второй** производной гладкой функции $\varphi$ в точке $x_m$ равномерной сетки по значениям функции в точках

$$x_{m-1},\quad x_{m},\quad x_{m+1}.$$

Оценить погрешность численного дифференцирования и оптимальный шаг сетки.

**Решение.** Эта задача аналогична задаче 1. Заметим, что рассматриваемый шаблон симметричен (и включает три точки), а порядок производной чётен (и равен двум). Поэтому порядок аппроксимации будет $3-2+1=2$ (а не $3-2=1$). Вводим неопределённые коэффициенты:

$$f''_m=\frac{Af_{m-1}+Bf_{m}+Cf_{m+1}}{h^2}.$$

Поскольку формула имеет второй порядок аппроксимации, первый нетривиальный член погрешности будет содержать $h^2$. Для определения коэффициентов этот член удерживать не требуется, но нам надо также найти ошибку метода, т.е. необходимо знать коэффициент при $h^2$. Поэтому слагаемые числителя будем раскладывать по формуле Тейлора вплоть до $h^4$: тогда после деления на $h^2$ получим как раз члены до $h^2$ включительно. Имеем:

$$f_{m\pm 1}=\varphi(x_m\pm h)=f_m\pm hf'_m+\frac{h^2}{2}f''_m\pm\frac{h^3}{6}f'''_m+\frac{h^4}{24}f^{(4)}_m;$$

$$f''_m=\frac{1}{h^2}(A+B+C)f_m+\frac{1}{h}(-A+C)f'_m+\frac{1}{2}(A+C)f''_m+\frac{h}{6}(-A+C)f'''_m+\frac{h^2}{24}(A+C)f^{(4)}_m.$$

Получаем три условия на коэффициенты:

\begin{equation}
  \begin{cases}
    A+B+C=0,\\
    -A+C=0,\\
    A+C=2.\\
  \end{cases}
\end{equation}

Отсюда

$$A=C=1,\quad B=-2.$$

Видно, что коэффициент при $f'''_m$ занулился автоматически (поэтому порядок аппроксимации и повысился на единицу). Ошибка метода составляет

$$\Delta^{(\text{m})}=\frac{M_4h^2}{12}.$$

Искомая формула имеет вид:

$$f''_m=\frac{f_{m+1}-2f_m+f_{m-1}}{h^2},$$

так что ошибка округления есть

$$\Delta^{(\text{r})}=\frac{1+2+1}{h^2}M_0\varepsilon_\text{маш}=\frac{4}{h^2}M_0\varepsilon_\text{маш}.$$

Полная погрешность:

$$\Delta(h)=\Delta^{(\text{m})}+\Delta^{(\text{r})}=\frac{M_4h^2}{12}+\frac{4}{h^2}M_0\varepsilon_\text{маш}.$$

Минимизируя $\Delta(h)$, находим оптимальный шаг:

$$\Delta'(h)=\frac{M_4h}{6}-\frac{8}{h^3}M_0\varepsilon_\text{маш}=0,\quad h_\text{opt}=\bigg(\frac{48M_0\varepsilon_\text{маш}}{M_4}\bigg)^{1/4}.$$

## Задача 8

Для функции $y=f(x)$, заданной таблично, вычислить значение **второй** производной в точке $x=3$ с максимально возможной точностью.

\begin{align}
  x   &\quad& 1   &\quad& 1,5  &\quad& 2    &\quad& 2,5  &\quad& 3\\
  y&\quad& 0,50&\quad& 0,25 &\quad& 0,25 &\quad& 0,20 &\quad& 0,10
\end{align}

**Решение.** Имеем равномерную сетку с $h=0,5$. Шаблон пятиточечный, несимметричный ($x_m=3$, $i=4$, $j=0$). Значит, максимальный порядок аппроксимации есть $5-2=3$. Вводим неопределённые коэффициенты:

$$f''_m=\frac{Af_{m-4}+Bf_{m-3}+Cf_{m-2}+Df_{m-1}+Ef_{m}}{h^2}.$$

Поскольку в знаменателе стоит $h^2$, **для определения коэффициентов** в разложениях по формуле Тейлора достаточно удерживать первые четыре члена (ошибку метода оценивать не требуется, поэтому члены пятого порядка нам не нужны). Имеем:

\begin{align}
  f_{m-4}&=f_m-4hf'_m+\frac{(4h)^2}{2}f''_m-\frac{(4h)^3}{6}f'''_m+\frac{(4h)^4}{24}f^{(4)}_m,\\[0.2cm]
  f_{m-3}&=f_m-3hf'_m+\frac{(3h)^2}{2}f''_m-\frac{(3h)^3}{6}f'''_m+\frac{(3h)^4}{24}f^{(4)}_m,\\[0.2cm]
  f_{m-2}&=f_m-2hf'_m+\frac{(2h)^2}{2}f''_m-\frac{(2h)^3}{6}f'''_m+\frac{(2h)^4}{24}f^{(4)}_m.\\[0.2cm]
  f_{m-1}&=f_m-hf'_m+\frac{h^2}{2}f''_m-\frac{h^3}{6}f'''_m+\frac{h^4}{24}f^{(4)}_m.
\end{align}

Подставляем эти разложения в формулу для $f''_m$. Группируем члены с одинаковыми степенями $h$:

\begin{align}
  f''_m&=\frac{1}{h^2}(A+B+C+D+E)f_m+\frac{1}{h}(-4A-3B-2C-D)f'_m\,+\\[0.2cm]
  &+\frac{1}{2}(16A+9B+4C+D)f''_m+\frac{h}{6}(-64A-27B-8C-D)f'''_m\,+\\[0.2cm]
  &+\frac{h^3}{24}(256A+81B+16C+D)f^{(4)}_m.
\end{align}

Сравнивая левую и правую части, составляем систему уравнений:

\begin{equation}
  \begin{cases}
    A+B+C+D+E=0,\\
    -4A-3B-2C-D=0,\\
    16A+9B+4C+D=2,\\
    -64A-27B-8C-D=0,\\
    256A+81B+16C+D=0.
  \end{cases}
\end{equation}

Решая систему (см. код ниже), находим:

$$A=\frac{11}{12},\quad B=-\frac{14}{3},\quad C=\frac{19}{2},\quad D=-\frac{26}{3},\quad E=\frac{35}{12}.$$

Искомая формула имеет вид:

$$f''_m=\frac{11f_{m-4}-56f_{m-3}+114f_{m-2}-104f_{m-1}+35f_m}{12h^2}.$$

Подставляя табличные значения и шаг (см. код ниже), находим:

$$f''_m=\frac{11\cdot 0,5-56\cdot 0,25+114\cdot 0,25-104\cdot 0,2+35\cdot 0,1}{12\cdot(0,5)^2}=0,9.$$

## Дифференцирование в Python

### Реализация численных методов

Библиотека ```numpy``` содержит несколько функций, которые можно использовать для вычисления конечно-разностных аналогов производных.

Функция ```diff``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.diff.html)) вычисляет разности между соседними элементами поданного на вход массива, возвращая массив **на единицу меньшей длины**. Поэтому её можно использовать для быстрого вычисления производной вперёд. Пример для линейной функции $y(x)=2x+3$:

````python
import numpy as np
import matplotlib.pyplot as plt

x, h = np.linspace(0, 20, 11, retstep=True) # дополнительно возвращает шаг сетки
y = 2 * x + 3
y_deriv = np.diff(y) / h # делим на шаг сетки

print(y_deriv, len(y_deriv))
````

Чтобы получить массив той же длины, что и исходный, необходимо задать один из необязательных аргументов ```prepend``` (элемент, добавляемый в начало **перед вычислением разностей**) или ```append``` (элемент, добавляемый в конец **перед вычислением разностей**):

````python
last = 2 * 20 + 3

# для примера добавляем last+h, чтобы вставленный элемент был заметен
y_deriv_app = np.diff(y, append=last+h) / h

print(y_deriv_app, len(y_deriv_app))
````

С помощью функции ```diff``` можно вычислять **разности высших порядков**. Например, вторые разности определяются как разности соседних элементов в массиве первых разностей. Иначе говоря, если

$$\delta_i^{(1)} = f_{i+1}-f_i,$$

то

$$\delta_i^{(2)}=\delta_{i+1}^{(1)}-\delta_{i}^{(1)}=f_{i+2}-2f_{i+1}+f_i.$$

Нетрудно заметить, что это числитель формулы, аппроксимирующей вторую производную (ср. с задачей 3). Поэтому делением на $h^2$ получаем массив приближённых значений второй производной (в данном примере они совпадают с точными значениями):

````python
y_deriv2 = np.diff(y, n=2) / (h**2)

print(y_deriv2, len(y_deriv2))
````

Заметим, что при сопоставлении этого массива с сеткой нужно сдвинуть индекс аргумента, связав разность $\delta_i^{(2)}$ с ($i-1$)-м узлом, поскольку в числителе формулы из задачи 3 стоит всё-таки $f_{m+1}-2f_m+f_{m-1}$. Естественно, что чем выше порядок разности, тем сильнее укорачивается исходный массив.

Пример использования функции ```diff``` с визуализацией:

````python
from ipywidgets import interact, IntSlider

# ползунок для интерактивного графика (виджета)
N_w = IntSlider(min=10, max=500, step=10, value=50)

def sin_diff(N: int):
    x, h = np.linspace(0, 2*np.pi, N, retstep=True)
    y = np.sin(x)
    sin_d = np.diff(y) / h
    sin_dd = np.diff(y, n=2) / h ** 2

    fig, ax = plt.subplots()
    ax.scatter(x[:N-1], sin_d, s=10, label="$y'(x)=\\cos x$") # не удлиняем массив разностей, а укорачиваем массив аргументов
    ax.scatter(x[1:N-1], sin_dd, s=10, label="$y''(x)=-\\sin x$") # массив сдвинут и укорочен

    ax.legend()

    x_plot = np.linspace(0, 2*np.pi, 1000) # сетка для графиков точных производных
    ax.plot(x_plot, np.cos(x_plot))
    ax.plot(x_plot, -np.sin(x_plot))

    plt.title("Производные функции $y(x)=\\sin x$ (с помощью diff)")
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.grid()
    plt.show()

interact(sin_diff, N=N_w);
````

Существует также функция ```ediff1d``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.ediff1d.html)), представляющая собой, по сути, усечённую версию функции ```diff```: она тоже вычисляет разности между последовательными элементами массива, но с её помощью можно получить только разности первого порядка. Кроме того, отсутствует возможность указать ось массива, вдоль которой должны вычисляться разности, поскольку массив, подаваемый на вход функции ```ediff1d```, автоматически приводится к одномерному. В случае же функции ```diff``` можно выбрать ось, задав необязательный аргумент ```axis```. В силу перечисленных причин функция ```ediff1d```, очевидно, малополезна.

Более удобной является функция ```gradient``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.gradient.html)). Она возвращает массив конечных разностей **той же длины, что и исходный**, причём расчёт ведётся по формуле центральной разности. Точнее: по значениям заданного массива $f_i$ рассчитывается величина $(f_{i+1}-f_{i-1})/2$. На границах массива, где применение центральной разности невозможно, она автоматически заменяется на одностороннюю разность (слева — на числитель формулы производной вперёд, справа — на числитель формулы производной назад). Сказанное легко проверить на простом примере:

````python
g = [5., 7., 8., 12., 4., 3., 0., -1.]

print(g)
print(np.gradient(g).tolist())
````

Таким образом, делением элементов выходного массива функции ```gradient``` на $h$ (в случае равномерной сетки) мы получаем массив значений производной, приближенной со вторым порядком аппроксимации (во всех точках, кроме граничных). Удобно, что сигнатура функции предусматривает возможность автоматического деления. Если сетка равномерная, то достаточно просто указать величину шага. Будем считать, например, что значения функции $g$ из предыдущей ячейки отнесены к равномерной сетке с шагом $h=2$. Тогда приближение для производной можно получить так:

````python
g_d = np.gradient(g, 2)
g_d.tolist()
````

Нетрудно проверить, что вычисления выполнены в точности по формуле центральной разности

$$f'_m=\frac{f_{m+1}-f_{m-1}}{2h}$$

(для всех значений, исключая граничные).

Если же сетка неравномерная, то для автоматического деления на (переменный) шаг необходимо вторым аргументом передать в функцию ```gradient``` массив из узлов используемой сетки. Разумеется, так же можно поступить и в случае равномерной сетки. Иначе говоря, численное дифференцирование с помощью функции ```gradient``` предельно упрощается: **в функцию надо передать массив значений и массив аргументов** (узлов сетки).

Более того, если требуется получить второй порядок аппроксимации и для граничных точек, можно указать необязательный аргумент ```edge_order=2``` (по умолчанию он равен 1). В этом случае значения производных в граничных точках будут вычисляться по односторонним формулам второго порядка:

$$f'_m=\frac{-3f_m + 4f_{m+1}-f_{m+2}}{2h},\quad f'_m=\frac{f_{m-2} - 4f_{m-1}+3f_{m}}{2h}.$$

Например:

````python
x = np.array([0, 1, 2, 3, 4])
f = x**2

print(np.gradient(f, edge_order=1))
print(np.gradient(f, edge_order=2))
````

Вычислим снова производные синуса с использованием функции ```gradient```:

````python
# создаём ползунок заново, чтобы не менять предыдущий график
N_again = IntSlider(min=10, max=500, step=10, value=50)

def sin_gradient(N: int):
    x = np.linspace(0, 2*np.pi, N)
    y = np.sin(x)

    sin_d = np.gradient(y, x, edge_order=2)
    sin_dd = np.gradient(sin_d, x, edge_order=2)

    fig, ax = plt.subplots()
    ax.scatter(x, sin_d, s=10, c='b', label="$y'(x)=\\cos x$")
    ax.scatter(x, sin_dd, s=10, c='r', label="$y''(x)=-\\sin x$",)

    ax.legend()

    x_plot = np.linspace(0, 2*np.pi, 1000) # сетка для графиков точных производных
    ax.plot(x_plot, np.cos(x_plot), c='b')
    ax.plot(x_plot, -np.sin(x_plot), c='r')

    plt.title("Производные функции $y(x)=\\sin x$ (с помощью gradient)")
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.grid()
    plt.show()

interact(sin_gradient, N=N_again);
````

Хорошо видно, что при одинаковых (и небольших) $N$ этот график и график, построенный выше с использованием функции ```diff```, отличаются: ```gradient``` даёт точки, подходящие к кривым ближе.

**Упражнение 3.** Из графика при малых $N$ ясно видно, что значения второй производной (красные точки) в граничных узлах вычисляются с меньшей точностью, чем в остальных узлах, несмотря на то, что ```edge_order=2```. Почему так получается? Как это исправить?

В заключение это раздела сделаем ряд важных **замечаний**:

1. Если при решении некоторой задачи часть уравнений аппроксимирована с порядком $p_1$, а другая часть уравнений — с порядком $p_2$, то итоговый порядок аппроксимации будет равен $\min (p_1,p_2)$. Если производная в граничных точках вычисляется с первым порядком аппроксимации, а во внутренних точках — со вторым, то общий порядок аппроксимации будет равен одному. Если же вычислять производные в граничных точках со вторым порядком аппроксимации, то итоговый порядок аппроксимации будет равен двум; это грамотный вариант с точки зрения вычислительной математики.

### Символьные вычисления

До сих пор мы обсуждали задачу численного дифференцирования в её классической (для вычислительной математики) постановке. Можно, однако, изначально рассматривать другой подход к вычислению производных с помощью компьютера.

Мы уже говорили, что задача **аналитического** дифференцирования сводится к последовательному применению **ряда правил**, по которым одни функции заменяются другими. Дифференцируя, например, выражение $x^2$, мы заменяем его на $2x$. Мы не выполняем при этом никаких арифметических действий (не вычисляем конечные разности) и даже не особенно вникаем в тот факт, что $x^2$ есть функция, обладающая какими-то свойствами. Вместо этого мы работаем с выражением $x^2$ как с некоторым **символом**, который, согласно правилам, должен быть заменён **другим символом** $2x$. Сталкиваясь с более сложной функцией, например, $x\sin (x^2)$, мы рассматриваем её как **комбинацию (суперпозицию) символов** и тоже применяем некоторые правила замены одних символов другими. В упомянутом примере цепочка замен может быть записана следующим образом:

\begin{gather}
  \big[x\sin (x^2)\big]'\quad\Rightarrow\quad(x)' \sin (x^2)+x\big[\sin (x^2)\big]'\quad\Rightarrow\quad\sin (x^2)+x\cos(x^2)\cdot (x^2)'\quad\Rightarrow\\
  \Rightarrow\quad\sin (x^2)+x\cos(x^2)\cdot 2x \quad\Rightarrow\quad \sin (x^2)+2x^2\cos(x^2).
\end{gather}

Очевидно, что аналитический способ вычисления производных можно алгоритмизовать и реализовать в виде некоторой программы. При исполнении такой программы компьютер не будет выполнять арифметические операции, относящиеся непосредственно к дифференцированию. Вместо этого он будет анализировать заданное выражение, составленное в рамках некоторого синтаксиса из определённых **символов**, и преобразовывать (упрощать) его в соответствии с системой правил. Другими словами, он будет дифференцировать функции так же, как их дифференцируют в математическом анализе.

Описанный подход составляет суть так называемых **символьных вычислений**. Они могут применяться не только в задачах дифференцирования, но и во многих других математических задачах. Отличительная особенность символьных вычислений заключается в том, что **компьютер работает с математическими объектами формально**, выполняя над ними не арифметические операции, а заранее запрограммированные процедуры преобразований (точнее — компьютер выполняет то одно, то другое).

Очевидным достоинством символьных  вычислений является **аналитическая точность результата** (нет ошибок метода и ошибок округления). Однако при этом **теряется основное преимущество компьютерных расчётов — возможность получить решение в тех случаях, когда его не удаётся найти аналитически**. По существу, при выполнении символьных вычислений компьютер превращается в математика, который действует теми же аналитическими методами, что и математик-человек, но имеет гораздо большую производительность, знает наизусть все первообразные, все специальные функции, не допускает ошибок, описок и т.д. Однако задачи, не имеющие аналитического решения, оказываются для такого компьютера-математика недоступными.

Из сказанного следует, что сфера применения символьных вычислений не так широка, как сфера применения численных методов. Тем не менее существуют специальные пакеты программ для символьных вычислений (их называют **системами компьютерной алгебры**), которые в ряде случаев оказываются очень полезными. Чаще всего в таких программах комбинируются **и аналитические, и численные возможности**. Известным примером служит система компьютерной алгебры Mathematica.

В Python символьные вычисления можно выполнять с помощью средств достаточно мощной библиотеки ```sympy```. Подробно изучить её читатель сможет с помощью [документации](https://docs.sympy.org/latest/index.html). Мы приведём только несколько примеров с пояснениями в контексте дифференцирования функций.

Для выполнения любых символьных вычислений и преобразований необходимо предварительно задать **символьные переменные**. Это можно сделать, например, с помощью функции ```Symbol``` (конструктора одноимённого класса). Она преобразовывает поданную на вход строку в символьную переменную — объект класса ```Symbol```:

````python
from sympy import Symbol

x = Symbol('x') # переменная x
type(x)
````

Если требуется задать сразу несколько переменных, то можно применить функцию ```symbols``` (или её аналог — функцию ```var```). При этом отдельные переменные во входных строках должны разделяться пробелами или запятыми. Можно указывать типы переменных, накладывать на них ограничения и снабжать их индексами:

````python
from sympy import symbols, var

a, b, t = symbols('a b t') # переменные a, b и t
n = symbols('n', integer=True) # целочисленная переменная n
y1, z32 = var('y_1, z^3_2', positive=True) # положительнозначные переменные y_1 и z^3_2
````

Буквы латинского и греческого алфавитов, часто использующиеся в формулах, можно импортировать из модуля ```symby.abc```. Отметим, что имена символьной переменной и переменой, использующейся в коде, не обязаны совпадать:

````python
from sympy.abc import phi, omega

f = phi # переменная фи
w = omega # переменная омега
````

Алгебраические выражения можно составлять из объявленных переменных, пользуясь обычным синтаксисом языка Python. Например:

````python
expr1 = 2*a / (a - 5) - 5 / (a + 5) + 2*a**2 / (25 - a**2)
expr2 = (2*b**2 - 3*b + 4) / (b-1)**2 - (b-2) / (b-1)
````

Очень удобно организован вывод символьных выражений — в ipynb-файле они автоматически преобразуются к TeX-формуле:

````python
expr1
````

Можно вызвать и встроенную функцию ```print```. Тогда математическая формула будет напечатана в строку из "обычных" символов:

````python
print(expr2)
````

Если выражение содержит элементарные функции или константы, то их нужно предварительно импортировать:

````python
from sympy import sin, cos, pi

expr3 = (sin(w*t + f))**2 + (cos(w*t + f + 2*pi*n))**2
expr3
````

Обратим внимание на то, что выражение ```expr3``` было частично упрощено: ```sympy``` автоматически отбросил из аргумента косинуса слагаемое $2\pi n$. Если бы заданная выше переменная $n$ не являлась целочисленной, это преобразование не было бы выполнено. Можно упрощать выражения и целенаправленно — с помощью функции ```simplify```:



````python
from sympy import simplify

simplify(expr1)
````

````python
simplify(expr2)
````

````python
simplify(expr3)
````

Для символьного дифференцирования заданных выражений применяется функция ```diff```. Она принимает на вход функцию, производную которой надо вычислить, и аргумент (символ), по которому требуется продифференцировать:

````python
from sympy import diff

y = sin(x)
diff(y, x)
````

Если требуется найти производные высоких порядков, то аргумент нужно указать несколько раз — столько, сколько раз надо продифференцировать:

````python
diff(y, x, x)
````

Точно так же можно находить производные функций нескольких переменных, в том числе и смешанные:

````python
g = sin(x) * sin (y1) * z32**2 / (x * y1)
g
````

````python
diff(g, x, z32, y1, y1)
````

Если нужно создать **символ производной** некоторой функции, а не выполнять дифференцирование (для некоторых операций это бывает необходимо), то можно воспользоваться конструктором класса ```Derivative```:

````python
from sympy import Derivative

y_d = Derivative(y, x)
y_d
````

````python
g_d = Derivative(g, x, z32, y1, y1)
g_d
````

Из подобного символа можно получить сам результат дифференцирования при помощи метода ```doit```:

````python
y_d.doit()
````

К результату дифференцирования (полученному как с помощью ```diff```, так и с помощью ```doit```) можно применить метод ```evalf``` для вычисления значения производной в конкретной точке:

````python
y_d.doit().evalf(subs={x: 0})
````

Большое преимущество библиотеки ```sympy``` по сравнению с другими системами компьютерной алгебры состоит в том, что **она не существует обособленно, а реализована именно как библиотека**, и поэтому может взаимодействовать с остальным кодом. В частности, результаты, полученные в ```sympy```, после некоторого преобразования можно обрабатывать средствами библиотеки ```numpy```. Покажем, как это делается, на примере дифференцирования.

Зададим какую-нибудь сложную функцию:

$$f(x)=\Bigg(\frac{\arcsin\sqrt{x}+\cos^2(3x-1)}{\pi^{x^3}+\ln^2\big(1+e^x\big)}\Bigg)^{x^2\text{ch}\,x}.$$

Составим её символьное выражение:

````python
from sympy import asin, sqrt, E, ln, cosh

numerator = asin(sqrt(x)) + (cos(3*x-1))**2
denominator = pi**(x**3) + (ln(1+E**x))**2
dimension = x**2 * cosh(x)

f = (numerator / denominator) ** dimension
f
````

Выполним дифференцирование:

````python
f_d = diff(f, x)
f_d
````

Результат получился весьма громоздким. Применение функции ```simplify``` в данном случае помогает мало: обработка выражения занимает слишком много времени. Однако результат можно и не упрощать, а просто преобразовать в векторизованную функцию ```numpy``` (например, если в дальнейшем с этой производной требуется выполнять какие-то расчёты). Для этого сначала с помощью команды ```lambdify``` получаем лямбда-функцию, а затем стандартным образом выполняем её векторизацию:

````python
from sympy import lambdify
import numpy as np

f_deriv = np.vectorize(lambdify(x, f_d, 'numpy'))
````

Теперь с этой функцией можно обычным (для ```numpy```) способом выполнять любые расчёты. Построим, например, график найденной производной. Для сравнения запрограммируем также исходную функцию и продифференцируем её численно:

````python
def composed_func(x: float):
    numerator = np.arcsin(np.sqrt(x)) + (np.cos(3*x - 1))**2
    denominator = np.power(np.pi, x**3) + (np.log(1 + np.exp(x)))**2
    dimension = x**2 * np.cosh(x)

    return np.power(numerator/denominator, dimension)
````

````python
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# ползунок для интерактивного графика (виджета)
N_w = IntSlider(min=10, max=500, step=10, value=20)

def cf_deriv(f_deriv, N: int):
    f = np.vectorize(composed_func)

    x = np.linspace(0.0, 0.5, N)
    y = f(x)
    y_deriv = np.gradient(y, x) # численное дифференцирование

    grid = np.linspace(0.0, 0.5, 1000) # сетка для производной, найденной символьно
    y_d = f_deriv(grid) # производная, найденная символьно

    fig, ax = plt.subplots()
    ax.scatter(x, y_deriv, c='b', s=10)
    ax.plot(grid, y_d, c='r', zorder=0)

    plt.title("Производная функции $y=f(x)$")
    plt.xlabel("$x$")
    plt.ylabel("$f(x)$")

    plt.grid()
    plt.show()

interact(lambda N: cf_deriv(f_deriv, N), N=N_w);
````

Таким образом, сочетание средств библиотек ```numpy``` и ```sympy``` даёт нам в руки довольно мощный инструмент аналитического дифференцирования на компьютере. Другое дело, что применимость этого инструмента очень ограничена: символьные вычисления нельзя использовать в весьма распространённом случае таблично заданных функций. Кроме того, очевидно, что невозможно аналитически продифференцировать неизвестную функцию. Построить же конечно-разностную аппроксимацию производной в общем виде (не конкретизируя значения массива) можно всегда. Именно этим и объясняется ведущая роль численных методов дифференцирования. Как мы увидим в дальнейшем, они широко используются при построении разностных схем.

3. *А.А. Амосов, Ю.А. Дубинский, Н.В. Копченова. Вычислительные методы для инженеров.* Подробное введение в предмет с большим количеством примеров дано в первых трёх главах. 4. *А.А. Самарский, А.П. Михайлов. Математическое моделирование: Идеи. Методы. Примеры.* Серьёзная монография, в которой детально разбираются методы построения и анализа математических моделей. Не является пособием по курсу вычислительной математики, но представляет интерес как книга, один из авторов которой являлся идеологом математического моделирования.

5. *Р.С. Пастушков. Задачи и упражнения по вычислительной математике. V семестр. Решения, указания, ответы.* Глава 1 (стр. 3—5) содержит примеры типовых задач по теме "Погрешности вычислений" с решениями. **Рекомендуемая литература (с комментариями):**

В подавляющем большинстве учебников и монографий по вычислительной математике задача численного дифференцирования рассматривается не в начале (как в нашем курсе), а в середине, после построения теории интерполяции. Следуя такому плану изложения, почти все авторы выводят формулы численного дифференцирования из интерполяционных формул (чаще всего пользуются интерполяционными полиномами Ньютона или Лагранжа). Иной подход, в котором центральное место отводится методу неопределённых коэффициентов, принят лишь в небольшом числе учебников. Среди них:

1.   *Н.С. Бахвалов, Н.П. Жидков, Г.М. Кобельков. Численные методы.* Классический учебник по вычислительной математике, достаточно строгий и, вместе с тем, вполне доступный. Лучше ориентироваться на восьмое (электронное) издание 2015 г. Численному дифференцированию посвящены параграфы 15 и 16 главы 2 (стр. 76—84). Хотя сама глава посвящена интерполяции, в указанных параграфах, как ни парадоксально, почти не используются интерполяционные полиномы, а формулы численного дифференцирования выводятся именно через неопределённые коэффициенты.

2.   *Л.И. Турчак. Основы численных методов.* Расширенное издание типичного курса лекций по вычислительной математике. Численному дифференцированию посвящён первый параграф третьей главы (стр. 78—92). Теория интерполяции используется (пункт 3 параграфа), но для понимания изложенного не требуется особенного углубления в эту теорию (пункт 3 можно просто пропустить). Интересны пункты 4 и 5 про метод Рунге—Ромберга и частные производные.

3.   *А.И. Лобанов, И.Б. Петров. Вычислительная математика: курс лекций.* Формулам для вычисления производных отведён пункт 1.7 первой лекции (стр. 24—29). Ссылки на теорию интерполяции отсутствуют. Логика изложения идентична той, которая использовалась в тексте семинара. Изложение весьма сжатое. Напомним, что помимо указанного учебника существует книга тех же авторов с похожим названием: "Лекции по вычислительной математике" (издание 2006 г.). Лучше пользоваться названной выше книгой (издания 2021 г.). В более новом учебнике устранены замеченные опечатки, а материал отдельных лекций значительно переработан.

Типовое изложение темы численного дифференцирования (с опорой на теорию интерполяции) можно найти почти в любом учебнике (любой монографии). Обратим внимание на следующие книги:

4.   *А.А. Амосов, Ю.А. Дубинский, Н.В. Копченова. Вычислительные методы для инженеров.* Численное дифференцирование обсуждается довольно бегло в сравнительно короткой 12 главе (стр. 364—374). Вопрос о выводе этих формул рассматривается с привлечением интерполяции.

5.   *Н.Н. Калиткин. Численные методы.* Хорошая монография известного специалиста по вычислительной математике. Численное дифференцирование рассмотрено довольно подробно в третьей главе (стр. 70—84). Изложение опирается на теорию аппроксимации функций. Рассмотрены некоторые дополнительные вопросы: некорректность задачи численного дифференцирования, способы регуляризации и др. Разобран в том числе случай неравномерных сеток.

6.   *Н.Н. Калиткин, Е.А. Альшина. Численный анализ.* Первый том учебника "Численные методы" (в двух книгах). Дифференцированию функций посвящена глава 6 (стр. 215—233). Общий план изложения в целом тот же, что и в упомянутой выше монографии Н.Н. Калиткина, однако заметно, что текст полностью переработан и дополнен.

7.   *О.В. Локуциевский, М.Б. Гавриков. Начала численного анализа.* Серьёзный учебник по вычислительной математике, заметно выделяющийся по сложности изложения среди других книг. Численные методы излагаются в духе "чистой" математики: активно используются элементы теории приближений и функционального анализа. Книга даёт новый взгляд на многие вопросы с единых позиций, но требует времени и сил для понимания. Численному дифференцированию посвящён параграф 17 главы 4 (стр. 222—246).
