# Введение

На прошлом семинаре мы уделили все время квадратурным формулам Ньютона-Котеса, обсудили сами квадратурные формулы, способы их уточнения, ошибку при вычислении по этим квадратурным формулам. На этом семинаре мы будем обсуждать второй класс методов численного интегрирования - квадратурные формулы Гаусса. Пусть необходимо вычислить интеграл

$$
I= \int \limits _a^b f(x) dx.
$$

Считаем, что оба предела интегрирования $a$ и $b$ конечны, а подынтегральная функция не имеет особенностей на отрезке $[a,b]$ (интеграл рассматривается в собственном смысле).

# Квадратурные формулы Гаусса

Для начала предложим способ приближенного вычисления интеграла на отрезке $[-1;1]$. Напомним, что $x_i$ - узлы квадратурной формулы, $c_i$ - веса квадратурной формулы. Формула примет вид


$$
\int \limits _{-1}^{1} f(x) dx \approx \sum \limits _{i=1}^n c_i f(x_i).
$$

Веса $c_i$ и узлы $x_i$ выбираются так, чтобы квадратурная формула была точна для многочленов степени не выше $2n - 1$. Справедлива следующая теорема


**Теорема.** Если в качестве узлов квадратурной формулы взять нули
полинома Лежандра $q_n(x)$ степени $n$:  
$$ q_n(x_i)=0,\quad i=1,\ldots,n, $$
а в качестве весов — интегралы от базисных функций многочленов Лагранжа
$l_i(x)$ степени $n-1$:
$$
c_i=\int_{-1}^{1}
\frac{(x-x_1)(x-x_2)\cdots(x-x_{i-1})(x-x_{i+1})\cdots(x-x_n)}
{(x_i-x_1)(x_i-x_2)\cdots(x_i-x_{i-1})(x_i-x_{i+1})\cdots(x_i-x_n)}
\, dx,
$$
то квадратурная формула будет точна при подстановке в неё вместо $f(x)$ любого
многочлена степени не выше $2n-1$.


Остается вопрос, как пользоваться предложенной квадратурной формулой, если интеграл требуется вычислить на произвольном конечном отрезке $[a;b]$. В таком случае

$$
\int \limits_{a}^{b} f(x)\,dx
\approx \frac{b-a}{2} \sum_{i=1}^{n} c_i \,
f\!\left( \frac{a+b}{2} + \frac{b-a}{2}\, t_i \right),
$$

где $c_i$, $t_i$ — веса и узлы на $[-1,1]$.


Для оценки ошибки интегрирования функции $f\in C^{2n}[a,b]$ по формулам Гаусса можно воспользоваться формулой

$$
R_n
= \frac{(b-a)^{2n+1}\,(n!)^4}{(2n+1)\,\bigl((2n)!\bigr)^3}\,
f^{(2n)}(\xi), \qquad \xi \in [a,b].
$$

Тогда ошибка $\Delta_n=\max|R_n|$.


Узлы и веса квадратурной формулы Гаусса нет необходимости вычислять вручную, их можно найти в готовых таблицах. Мы приведем небольшую таблицу, которой будет достаточно в учебных целях

$$
\begin{array}{c|c|c|c}
n & c_i\ - \text{веса на}\ [-1,1] & t_i\ -  \text{узлы на}\ [-1,1] & \Delta_n \\ \hline
2 & c_{1,2}=1 &
t_{1,2}=\pm\frac{\sqrt{3}}{3} &
\dfrac{(b-a)^5}{2^5}\cdot\dfrac{1}{135}\,\max|f^{(4)}| \\[6pt] \hline
3 &
c_{1,3}=\frac{5}{9},\quad c_2=\frac{8}{9} &
t_{1,3}=\pm\sqrt{\frac{3}{5}},\quad t_2=0 &
\dfrac{(b-a)^7}{2^7}\cdot\dfrac{1}{15750}\,\max|f^{(6)}| \\[6pt] \hline
4 &
c_{1,4}\approx0.347855,\quad c_{2,3}\approx0.652145 &
t_{1,4}\approx\pm 0.861136,\quad t_{2,3}\approx\pm 0.339981 &
\dfrac{(b-a)^9}{2^9}\cdot\dfrac{1}{3472875}\,\max|f^{(8)}|
\end{array}
$$

## Задача 1

Вычислите интеграл по формулам Гаусса с точностью $ \varepsilon = 10^{-3} $

$$
I=\int \limits_{-1}^{1} \ln\!\left(1+\frac{x^{2}}{3}\right)\,dx
$$



**Решение.** В первую очередь заметим, что подынтегральная функция является четной, тогда верно, что
$$
I = 2 \int_{0}^{1} \ln\!\left(1+\frac{x^{2}}{3}\right)\,dx.
$$

Начнем с определения числа узлов, которые необходимо задать для вычисления с заданной точностью. Для начала проверим вариант $n=2$

$$\Delta_{2} = \dfrac{(b-a)^5}{2^5}\cdot\dfrac{1}{135}\,\max|f^{(4)}|$$

$$
\Delta_{2} = \frac{1}{2^{5}}\cdot \frac{1}{135}\cdot \frac{4}{3} \approx 3\cdot 10^{-4} < \frac{\varepsilon}{2}
\;\Rightarrow\; \text{достаточно}
$$

Константа $M_4$ может быть найдена различными способами, в прошлом семинаре мы получили формулу

$$
f(x)=\ln\!\left(1+\frac{x^2}{a^2}\right),\qquad
\bigl|f^{(n)}(x)\bigr|\le \frac{2\,(n-1)!}{a^{\,n}},
$$
откуда $M_4=4/3$

Если бы вариант с двумя узлами не подошел, мы бы последовательно увеличивали бы $n$, пока не нашли бы подходящий вариант.

Теперь осталось записать квадратурную формулу

$$
I \approx 2 \cdot \frac{b-a}{2}\,(c_{1} f(x_{1}) + c_{2} f(x_{2}))
$$

Узлы и веса берем из таблицы
$$
c_{1} = c_{2} = 1,\qquad
t_{1,2} = \pm \frac{\sqrt{3}}{3}
$$

$$
x_{1,2} = \frac12 + \frac12\, t_{1,2}
       = \frac12 \pm \frac{\sqrt{3}}{6}
$$
В итоге получаем
$$
I \approx f(x_{1}) + f(x_{2})
   = \ln\!\left(1+\frac{x_{1}^{2}}{3}\right)
   + \ln\!\left(1+\frac{x_{2}^{2}}{3}\right)
   \approx 0.2031928
$$
Точное значение интеграла
$$
I^{*} \approx 0.2029629
$$
хорошо совпадает с полученным в результате вычислений по квадратурной формуле

### Квадратуры Гаусса—Лежандра в Python



Узлы и веса квадратуры Гаусса—Лежандра можно получить с помощью функции `leggauss` из модуля `numpy.polynomial.legendre` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.polynomial.legendre.leggauss.html)). Функция возвращает узлы и веса на стандартном отрезке $[-1,1]$.

Сначала реализуем переход от стандартного отрезка к произвольному отрезку $[a,b]$. В функции ниже `t` и `c` — узлы и веса на $[-1,1]$, а `x` — перенесённые узлы на $[a,b]$.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import HTML, display
from numpy.polynomial.legendre import leggauss


def show_table(table) -> None:
    """Выводит таблицу как обычный HTML во всех средах выполнения."""
    display(HTML(table.to_html(index=False)))


def gauss_legendre(f, a: float, b: float, n: int) -> float:
    """Вычисляет интеграл от f на [a, b] по n-точечной формуле Гаусса."""
    # Узлы t_i и веса c_i квадратуры на стандартном отрезке [-1, 1].
    t, c = leggauss(n)

    # Линейное отображение узлов из [-1, 1] в [a, b].
    x = (a + b) / 2 + (b - a) / 2 * t

    # Множитель (b-a)/2 появляется из замены переменной.
    return (b - a) / 2 * np.dot(c, f(x))
```

Сравним узлы и веса, вычисленные библиотечной функцией, с приведённой выше таблицей. Симметричные узлы и соответствующие им равные веса выводятся отдельными строками.

```python
# Собираем узлы и веса для квадратур порядков n = 2, 3, 4.
node_rows = []
for n in (2, 3, 4):
    nodes, weights = leggauss(n)
    for i, (node, weight) in enumerate(zip(nodes, weights), start=1):
        node_rows.append(
            {"n": n, "i": i, "Узел t_i": node, "Вес c_i": weight}
        )

nodes_table = pd.DataFrame(node_rows)
show_table(nodes_table)
```

Различий с таблицей выше не обнаруживаем.

Теперь проверим утверждение об алгебраической точности. При $n=3$ формула должна точно интегрировать многочлены до степени $2n-1=5$. Для сравнения добавим многочлен степени $6$, для которого точность уже не гарантируется.

```python
# Проверяем трехточечную формулу на мономах x^k, k = 0, ..., 6.
n = 3
nodes, weights = leggauss(n)
exactness_rows = []

for degree in range(2 * n + 1):
    approximation = np.dot(weights, nodes**degree)

    # Интеграл нечетного монома по [-1, 1] равен нулю.
    exact_value = 0.0 if degree % 2 else 2.0 / (degree + 1)

    exactness_rows.append(
        {
            "Степень k": degree,
            "Квадратурная сумма": approximation,
            "Точное значение": exact_value,
            "Абсолютная погрешность": abs(approximation - exact_value),
        }
    )

exactness_table = pd.DataFrame(exactness_rows)
show_table(exactness_table)
```

Для $k \le 5$ значение абсолютной погрешности получилось порядка ошибки округления, это и демонстрирует проверяемое утверждение.

Наконец, повторим вычисление интеграла из задачи 1 при различном числе узлов. Его точное значение можно записать аналитически:

$$
I^*=2\ln\frac43-4+\frac{2\pi}{\sqrt3}.
$$

В таблице и на графике проследим зависимость погрешности от $n$.

```python
# Подынтегральная функция из задачи 1 и точное значение интеграла.
f_gauss = lambda x: np.log1p(x**2 / 3)
I_gauss_exact = 2 * np.log(4 / 3) - 4 + 2 * np.pi / np.sqrt(3)

orders = np.arange(1, 9)

# Как и в решении задачи, используем четность функции:
# интегрируем на [0, 1] и удваиваем результат.
approximations = np.array(
    [2 * gauss_legendre(f_gauss, 0.0, 1.0, int(n)) for n in orders]
)
errors = np.abs(approximations - I_gauss_exact)

gauss_results = pd.DataFrame(
    {
        "Число узлов n": orders,
        "Приближенное значение": approximations,
        "Абсолютная погрешность": errors,
    }
)
show_table(gauss_results)

# Логарифмический масштаб позволяет увидеть быстрое уменьшение погрешности.
plt.figure(figsize=(7, 4))
plt.semilogy(orders, errors, "o-", label="Формула Гаусса")
plt.axhline(1e-3, color="tab:red", linestyle="--", label=r"$\varepsilon=10^{-3}$")
plt.xlabel("Число узлов n")
plt.ylabel("Абсолютная погрешность")
plt.title("Формула Гаусса для интеграла из задачи 1")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.show()
```

При $n=2$ погрешность уже меньше $10^{-3}$, что согласуется с аналитической оценкой из решения задачи. При дальнейшем увеличении $n$ погрешность быстро уменьшается, пока не достигает уровня ошибок округления.

### Пример: квадратура Гаусса—Лежандра для функции Рунге

Рассмотрим стандартную функцию Рунге

$$
f(x)=\frac{1}{1+25x^2},\qquad x\in[-1,1].
$$

Точное значение интеграла равно

$$
I=\int_{-1}^{1}\frac{dx}{1+25x^2}
=\frac{2}{5}\arctan 5
\approx0.5493603068.
$$

Квадратуры Гаусса—Лежандра для этой функции сходятся к точному значению. Однако увеличение числа узлов повышает степень точности формулы, но не гарантирует уменьшения погрешности на каждом отдельном шаге.

Например, для двухточечной формулы

$$
Q_2=\frac{2}{1+25/3}=\frac{3}{14}\approx0.2142857143,
\qquad
|Q_2-I|\approx0.3350745925.
$$

Трёхточечная формула содержит узел $x=0$ с весом $8/9$. Поэтому

$$
Q_3=\frac{8}{9}
+2\cdot\frac{5}{9}\cdot\frac{1}{1+25\cdot3/5}
=\frac{23}{24}\approx0.9583333333,
$$

$$
|Q_3-I|\approx0.4089730266>|Q_2-I|.
$$

Таким образом, при переходе от двух узлов к трём погрешность увеличивается.

```python
# Функция Рунге и точное значение интеграла на [-1, 1].
runge = lambda x: 1 / (1 + 25 * x**2)
I_runge_exact = 2 / 5 * np.arctan(5)

# Вычисляем квадратуры с числом узлов n = 1, ..., 30.
n_values = np.arange(1, 31)
runge_approximations = np.array(
    [gauss_legendre(runge, -1, 1, int(n)) for n in n_values]
)
runge_errors = np.abs(runge_approximations - I_runge_exact)

# Отмечаем шаги, на которых добавление одного узла увеличило ошибку.
error_increased = ["—"] + [
    "да" if runge_errors[i] > runge_errors[i - 1] else "нет"
    for i in range(1, len(runge_errors))
]

runge_results = pd.DataFrame(
    {
        "Число узлов n": n_values,
        "Приближённое значение": runge_approximations,
        "Абсолютная погрешность": runge_errors,
        "Ошибка выросла относительно n-1": error_increased,
    }
)

# Для таблицы достаточно первых двенадцати строк; на графике покажем все 30.
show_table(runge_results.head(12))

plt.figure(figsize=(7, 4))
plt.semilogy(n_values, runge_errors, "o-", markersize=4)
plt.scatter(
    [2, 3],
    runge_errors[[1, 2]],
    color="tab:red",
    zorder=3,
    label="Ошибка выросла при переходе 2 → 3",
)
plt.xlabel("Число узлов n")
plt.ylabel("Абсолютная погрешность")
plt.title("Квадратура Гаусса—Лежандра для функции Рунге")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.show()
```

Увеличение погрешности при переходе $n=2\to3$ не противоречит росту алгебраической степени точности: точность на многочленах до степени $2n-1$ ничего не говорит о монотонности ошибки для произвольной неполиномиальной функции.

Немонотонность в начале последовательности связана с формой функции Рунге. Она имеет узкий максимум при $x=0$; нечётные формулы содержат центральный узел, а чётные — нет. Поэтому последовательные приближения сначала колеблются относительно точного значения. При дальнейшем увеличении $n$ погрешность в целом убывает.

# Квадратурные формулы Гаусса-Кристоффеля

Квадратурные формулы Гаусса-Кристоффеля позволяют вычислять интеграл с известной весовой функцией $p(x)$ как на конечном отрезке $[a,b]$, так и в пределах $(-\infty; +\infty)$. Формула мало отличается от предложенной выше, ее узлы и веса выбираются по похожему принципу

$$
\int \limits_{a}^{b} f(x)\,p(x)\,dx \;\approx\; \sum_{i=1}^{n} c_{i}\, f(x_{i})
$$


**Теорема.** Предполагается, что весовая функция $p(x)$ неотрицательна на $(a,b)$, её интеграл по этому промежутку положителен и конечен, а моменты $\int_a^b |x|^k p(x)\,dx$ конечны при $k=1,\ldots,2n$. Тогда квадратурная формула точна для многочленов степени не выше $2n-1$, если её узлами $x_i$ являются корни многочлена $Q_n(x)$ из семейства многочленов, ортогональных на промежутке интегрирования $(a,b)$ с весом $p(x)$, а весовыми коэффициентами являются числа

$$
c_i = \int_{a}^{b} l_i(x)\, p(x)\, dx.
$$


## Задача 2

Постройте формулу для вычисления интеграла по квадратурным формулам Гаусса-Кристоффеля с двумя узлами.

$$
I(f) = \int \limits _{-\infty}^{+\infty} f(x)\,\exp(-x^{2})\,dx
$$


**Решение.** Так как в условии сказано использовать формулу с двумя узлами, то ответ ищем в виде

$$
I \approx c_1\,f(x_1) + c_2\,f(x_2)
$$

Формула должна быть точна для многочленов
$$
P_{0}(x),\, P_{1}(x),\, P_{2}(x),\, P_{3}(x)
$$

Выпишем эти условия. Сначала подставим многочлен нулевой степени

$$
P_{0}(x)=1,\qquad
I=\int\limits_{-\infty}^{+\infty} 1\cdot e^{-x^{2}}\,dx=\sqrt{\pi},
$$
$$
I=c_{1}\cdot 1+c_{2}\cdot 1 \;\Longrightarrow\; c_{1}+c_{2}=\sqrt{\pi}.
$$

Далее многочлен первой степени
$$
P_{1}(x)=x,\qquad
I=\int\limits_{-\infty}^{+\infty} x e^{-x^{2}}\,dx=0,
$$
$$
I=c_{1}x_{1}+c_{2}x_{2} \;\Longrightarrow\; c_{1}x_{1}+c_{2}x_{2}=0.
$$

Далее многочлен второй степени
$$
P_{2}(x)=x^{2},\qquad
I=\int\limits_{-\infty}^{+\infty} x^{2} e^{-x^{2}}\,dx=\frac{\sqrt{\pi}}{2},
$$
$$
I=c_{1}x_{1}^{2}+c_{2}x_{2}^{2} \;\Longrightarrow\;
c_{1}x_{1}^{2}+c_{2}x_{2}^{2}=\frac{\sqrt{\pi}}{2}.
$$
И многочлен третьей степени
$$
P_{3}(x)=x^{3},\qquad
I=\int\limits_{-\infty}^{+\infty} x^{3} e^{-x^{2}}\,dx=0,
$$
$$
I=c_{1}x_{1}^{3}+c_{2}x_{2}^{3} \;\Longrightarrow\;
c_{1}x_{1}^{3}+c_{2}x_{2}^{3}=0.
$$

Записываем получившуюся систему уравнений и решаем

$$
\begin{cases}
c_{1}+c_{2}=\sqrt{\pi}, \\[6pt]
c_{1}x_{1}+c_{2}x_{2}=0, \\[6pt]
c_{1}x_{1}^{2}+c_{2}x_{2}^{2}=\dfrac{\sqrt{\pi}}{2}, \\[6pt]
c_{1}x_{1}^{3}+c_{2}x_{2}^{3}=0
\end{cases}
$$
Воспользуемся соображениями симметрии, тогда положим

$$c_1 = c_2 = c, \quad x_1 = -x_2 $$

Из первого уравнения найдем
$$c = \frac{\sqrt{\pi}}{2}$$
А из третьего
$$2 \cdot \frac{\sqrt{\pi}}{2} \, x_1^2 = \frac{\sqrt{\pi}}{2}
\quad\Rightarrow\quad
x_{1,2} = \pm \frac{1}{\sqrt{2}}$$

В итоге получаем
$$
I(f) \approx \frac{\sqrt{\pi}}{2}\,\bigl(f(\tfrac{1}{\sqrt{2}}) + f(-\tfrac{1}{\sqrt{2}})\bigr)
$$



### Квадратура Гаусса—Эрмита в Python



Для весовой функции $p(x)=e^{-x^2}$ квадратура Гаусса—Кристоффеля называется квадратурой Гаусса—Эрмита. Её узлы и веса вычисляет функция `scipy.special.roots_hermite` ([документация](https://docs.scipy.org/doc/scipy/reference/generated/scipy.special.roots_hermite.html)).

Важно, что вес $e^{-x^2}$ уже учтён в квадратурной формуле. Поэтому при вычислении

$$
\int_{-\infty}^{+\infty} f(x)e^{-x^2}\,dx
$$

в квадратурную сумму подставляются значения только функции $f(x)$; повторно умножать их на $e^{-x_i^2}$ не нужно.

```python
from scipy.special import gamma, roots_hermite

# Узлы и веса двухточечной квадратуры Гаусса—Эрмита.
hermite_nodes_2, hermite_weights_2 = roots_hermite(2)

hermite_two_point = pd.DataFrame(
    {
        "i": np.arange(1, 3),
        "Узел x_i": hermite_nodes_2,
        "Вес c_i": hermite_weights_2,
    }
)
show_table(hermite_two_point)
```

Полученные значения совпадают с результатом задачи 2:

Проверим точность этой формулы на мономах. Для $n=2$ точность должна сохраняться до третьей степени включительно; моном четвёртой степени добавим для сравнения.

```python
hermite_exactness_rows = []

for degree in range(5):
    approximation = np.dot(hermite_weights_2, hermite_nodes_2**degree)

    # Для четного k интеграл x^k exp(-x^2) равен Gamma((k+1)/2),
    # а для нечетного k он равен нулю по симметрии.
    exact_value = 0.0 if degree % 2 else gamma((degree + 1) / 2)

    hermite_exactness_rows.append(
        {
            "Степень k": degree,
            "Квадратурная сумма": approximation,
            "Точное значение": exact_value,
            "Абсолютная погрешность": abs(approximation - exact_value),
        }
    )

hermite_exactness_table = pd.DataFrame(hermite_exactness_rows)
show_table(hermite_exactness_table)
```

В качестве неполиномиального примера вычислим интеграл

Проследим, как меняется погрешность при увеличении числа узлов.

```python
I_hermite_exact = np.sqrt(np.pi) * np.exp(-1 / 4)
hermite_orders = np.arange(1, 9)
hermite_approximations = []

for n in hermite_orders:
    nodes, weights = roots_hermite(int(n))

    # Вес exp(-x^2) уже включен в weights, поэтому вычисляем только cos(nodes).
    hermite_approximations.append(np.dot(weights, np.cos(nodes)))

hermite_approximations = np.array(hermite_approximations)
hermite_errors = np.abs(hermite_approximations - I_hermite_exact)

hermite_results = pd.DataFrame(
    {
        "Число узлов n": hermite_orders,
        "Приближенное значение": hermite_approximations,
        "Абсолютная погрешность": hermite_errors,
    }
)
show_table(hermite_results)

plt.figure(figsize=(7, 4))
plt.semilogy(hermite_orders, hermite_errors, "o-")
plt.xlabel("Число узлов n")
plt.ylabel("Абсолютная погрешность")
plt.title(r"Квадратура Гаусса—Эрмита для $f(x)=\cos x$")
plt.grid(True, which="both", alpha=0.3)
plt.show()
```

# Интеграл от быстро осциллирующей функции

На практике можно столкнуться с интегралами, в которых подынтегральная функция совершает множество "осцилляций" на отрезке интегрирования. Это интегралы вида

$$
I = \int\limits_a^b f(x)\sin(\omega x)\,dx, \qquad \omega(b-a)\gg 1
$$

или

$$
I = \int\limits_a^b f(x)\cos(\omega x)\,dx, \qquad I = \int\limits_a^b f(x)\exp(i\omega x)\,dx,
$$

где $f(x)$ - функция, задающая гладкую медленно меняющуюся амплитуду. В таком случае прямое интегрирование по предложенным в прошлом семинаре квадратурным формулам Ньютона-Котеса и по квадратурным формулам Гаусса при больших частотах может требовать чрезмерно мелкой сетки. Поступим иначе. Построим $\tilde f(x)$ - аппроксимацию $ f(x)$, например сплайном. Тогда

$$\int\limits_a^b \tilde f(x)e^{i\omega x}dx$$

интегрируется аналитически. Ошибка интегрирования

$$
\Delta = \left| \int \limits_a^b (f(x) - \tilde f(x)) e^{i\omega x}\, dx \right|.
$$

Ошибка интегрирования на каждом отрезке

$$
\Delta_n = \left| \int_{x_n}^{x_{n+1}} f(x)e^{i\omega x}\,dx - \int_{x_n}^{x_{n+1}} \tilde S_3(x,n)e^{i\omega x}\,dx \right|
=
\left| \int_{x_n}^{x_{n+1}} R_n(x)e^{i\omega x}\,dx \right|
\le
\int_{x_n}^{x_{n+1}} |R_n(x)|\,dx,
$$

где $ R_n(x)=f(x)-\tilde S_3(x,n)$ - остаточный член интерполяции.

### Интеграл от быстро осциллирующей функции в Python

Рассмотрим интеграл из практической задачи

$$
I=\int_0^\pi x^3\sin(50x)\,dx
=-\frac{\pi^3}{50}+\frac{6\pi}{50^3}.
$$

Сравним составную формулу Симпсона с алгоритмом, который учитывает осциллирующий множитель отдельно. Для формулы Симпсона воспользуемся функцией `scipy.integrate.simpson` ([документация](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.simpson.html)). Функция `scipy.integrate.quad` поддерживает специальные режимы `weight="sin"` и `weight="cos"` ([документация](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.quad.html)).

При конечных пределах интегрирования `quad` в таком режиме применяет модифицированный метод Кленшоу—Кёртиса и при необходимости адаптивно разбивает отрезок. Число точек заранее не задаётся: алгоритм выбирает его, ориентируясь на параметры точности `epsabs` и `epsrel`. Если указать `full_output=True`, функция дополнительно возвращает словарь с диагностической информацией. Значение `infodict["neval"]` показывает фактическое число вычислений переданной функции.

```python
from scipy.integrate import quad, simpson

omega = 50.0
I_osc_exact = -np.pi**3 / omega + 6 * np.pi / omega**3


def simpson_oscillatory(number_of_intervals: int) -> float:
    """Вычисляет осциллирующий интеграл на равномерной сетке."""
    # Формуле Симпсона передаются значения полной подынтегральной функции
    # в N + 1 узлах равномерной сетки.
    x = np.linspace(0.0, np.pi, number_of_intervals + 1)
    y = x**3 * np.sin(omega * x)
    return simpson(y, x=x)


# Несколько характерных сеток, включая резонансный случай N = 50.
selected_N = [20, 40, 50, 60, 100, 200]
oscillatory_rows = []

for N in selected_N:
    approximation = simpson_oscillatory(N)
    oscillatory_rows.append(
        {
            "Метод": "Составная формула Симпсона",
            "Заданное число интервалов": N,
            "Вычислений функции": N + 1,
            "Приближенное значение": approximation,
            "Абсолютная погрешность": abs(approximation - I_osc_exact),
        }
    )

# В quad передаём только медленно меняющийся множитель x^3.
# Множитель sin(omega*x) задаётся параметрами weight и wvar.
# full_output=True позволяет получить число вычислений функции.
weighted_value, weighted_error_estimate, weighted_info = quad(
    lambda x: x**3,
    0.0,
    np.pi,
    weight="sin",
    wvar=omega,
    full_output=True,
)
weighted_neval = weighted_info["neval"]
weighted_actual_error = abs(weighted_value - I_osc_exact)

oscillatory_rows.append(
    {
        "Метод": "quad с весом sin(ωx)",
        "Заданное число интервалов": "не задаётся",
        "Вычислений функции": weighted_neval,
        "Приближенное значение": weighted_value,
        "Абсолютная погрешность": weighted_actual_error,
    }
)

oscillatory_table = pd.DataFrame(oscillatory_rows)
show_table(oscillatory_table)
print(f"Оценка погрешности, возвращённая quad: {weighted_error_estimate:.3e}")
print(f"Число вычислений функции в quad: {weighted_neval}")

# Сначала сохраняем прежний график: погрешность формулы Симпсона
# как функция числа интервалов равномерной сетки.
N_values = np.arange(10, 202, 2)
simpson_errors = np.array(
    [abs(simpson_oscillatory(int(N)) - I_osc_exact) for N in N_values]
)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

axes[0].semilogy(
    N_values,
    simpson_errors,
    "o-",
    markersize=3,
    label="Формула Симпсона",
)
axes[0].axvline(50, color="tab:red", linestyle="--", label="N=50")
axes[0].set_xlabel("Число интервалов N")
axes[0].set_ylabel("Абсолютная погрешность")
axes[0].set_title("Формула Симпсона")
axes[0].grid(True, which="both", alpha=0.3)
axes[0].legend()

# Для сопоставления разных алгоритмов используем общую меру затрат —
# количество вычисленных значений функции. Результат quad отмечаем одной точкой,
# поскольку число вычислений выбирает сам адаптивный алгоритм.
axes[1].semilogy(
    N_values + 1,
    simpson_errors,
    "o-",
    markersize=3,
    label="Формула Симпсона",
)
axes[1].scatter(
    weighted_neval,
    weighted_actual_error,
    marker="D",
    s=70,
    color="tab:orange",
    label=r"quad с весом $\sin(\omega x)$",
    zorder=3,
)
axes[1].set_xlabel("Число вычислений функции")
axes[1].set_ylabel("Абсолютная погрешность")
axes[1].set_title("Сравнение вычислительных затрат")
axes[1].grid(True, which="both", alpha=0.3)
axes[1].legend()

fig.suptitle(r"Интегрирование $x^3\sin(50x)$")
plt.tight_layout()
plt.show()
```

При $N=50$ все узлы сетки имеют вид $x_k=k\pi/50$, поэтому $\sin(50x_k)=\sin(k\pi)=0$ с точностью до ошибок округления. Формула Симпсона «не замечает» осцилляции между узлами и получает значение, близкое к нулю. Левый график также показывает, что погрешность при сгущении сетки может изменяться немонотонно: результат зависит не только от шага, но и от взаимного расположения узлов и осцилляций функции.

В правой части рисунка методы сравниваются по числу вычислений функции. В данном примере `quad` понадобилось 25 вычислений медленно меняющегося множителя $x^3$, после чего была достигнута погрешность порядка ошибки округления. Число 25 не является постоянной характеристикой метода: для более сложной функции алгоритм может разбить отрезок и выполнить больше вычислений.

Задать в `quad` точное число точек нельзя. Параметры `epsabs` и `epsrel` определяют требуемую точность, а параметр `limit` задаёт только верхнюю границу числа подынтервалов адаптивного разбиения. Поэтому результат `quad` показан на графике одной точкой, а не кривой с заранее выбранным числом интервалов.

# Несобственный интеграл

Рассмотрим подходы, позволяющие вычислить несобственные интегралы с заданной точностью. Конечно, задача имеет смысл, только если интеграл сходится абсолютно или условно. Далее рассмотрим несколько способов вычисления абсолютно сходящихся несобственных интегралов. Для начала рассмотрим несколько простых подходов на примере интеграла

$$
I = \int\limits_{0}^{1} \frac{\cos x}{\sqrt{x}}\,dx
$$

1) Замена переменной

$$
x = t^{2}, \qquad I = 2 \int\limits_{0}^{1} \cos(t^{2})\, dt.
$$

Полученный интеграл является собственным.

2) Интегрирование по частям

$$
I = \int\limits_{0}^{1} \frac{\cos x}{\sqrt{x}}\, dx
    = 2 \cos x\, \sqrt{x}\Big|_{0}^{1}
    + 2 \int\limits_{0}^{1} \sqrt{x}\, \sin x\, dx
$$

После первого интегрирования по частям интеграл становится собственным, первая производная подынтегральной функции ограничена, а вторая — неограничена. Поэтому можно применять, например, формулы прямоугольников с правой или левой точкой со стандартной оценкой погрешности. Чтобы воспользоваться стандартной оценкой погрешности формулы трапеций, интегрирование по частям требуется продолжить до получения подынтегральной функции с ограниченной второй производной. В данном примере после первого интегрирования для этого требуются ещё два таких преобразования.

$$
I = 2\cos 1
 + 2\sin x \cdot \frac{2}{3} x^{3/2}\Big|_{0}^{1}
 - 2 \int\limits_{0}^{1} \cos x \cdot \frac{2}{3} x^{3/2}\, dx
$$
$$
= 2\cos 1 + \frac{4}{3}\sin 1
- \frac{8}{15}x^{5/2}\cos x\Big|_{0}^{1}
- \frac{8}{15}\int\limits_{0}^{1}x^{5/2}\sin x\,dx
$$

$$
= \frac{22}{15}\cos 1 + \frac{4}{3}\sin 1
- \frac{8}{15}\int\limits_{0}^{1}x^{5/2}\sin x\,dx.
$$

У подынтегральной функции $x^{5/2}\sin x$ вторая производная ограничена и стремится к нулю при $x\to0+$.

3) Метод явного выделения особенности
$$
I=\int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx
    =\int\limits_{0}^{\delta}\frac{\cos x}{\sqrt{x}}\,dx
    +\int\limits_{\delta}^{1}\frac{\cos x}{\sqrt{x}}\,dx
$$

Первый интеграл $I_{1}$ вычисляется приближённо аналитически, второй интеграл $I_{2}$ — численно.

4) Метод Канторовича

$$
\int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx
=
\int\limits_{0}^{1}\frac{\cos x - 1}{\sqrt{x}}\,dx
    +\int\limits_{0}^{1}\frac{dx}{\sqrt{x}}
$$

Первый интеграл не содержит особенности и может быть найден численно, второй интеграл легко вычисляется аналитически.

Заметим, что в первом интеграле из $\cos x$ вычтен первый член её разложения в ряд Тейлора. Полученная подынтегральная функция непрерывно продолжается в точку $x=0$ значением $0$ и имеет ограниченную первую производную. Поэтому можно применять формулы прямоугольников с правой или левой точками со стандартной оценкой погрешности. Вторая производная при этом остаётся неограниченной. Чтобы воспользоваться стандартной оценкой погрешности формулы трапеций, вычтем из $\cos x$ два первых ненулевых члена разложения, то есть $1-x^2/2$. Тогда остаточная подынтегральная функция после доопределения нулём при $x=0$ будет иметь ограниченную вторую производную. Проделаем этот шаг
$$
I = \int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx
= \int\limits_{0}^{1}
\frac{\cos x - 1 + \frac{x^2}{2}}{\sqrt{x}}\,dx
+ \int\limits_{0}^{1}
\frac{1 - \frac{x^2}{2}}{\sqrt{x}}\,dx
$$

$$
= \int\limits_{0}^{1}
\frac{\cos x - 1 + \frac{x^2}{2}}{\sqrt{x}}\,dx
+ \frac{9}{5}.
$$



## Задача 3

Предложить алгоритм, как пользуясь программой, реализующей метод трапеций для численного интегрирования регулярных функций с произвольным шагом, вычислить следующий интеграл с точностью $\varepsilon=10^{-3}$ (использовать метод явного выделения особенности)

$$
I= \int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx$$

**Решение.**

$$
I= \int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx
=
\underbrace{\int\limits_{0}^{\delta}\frac{\cos x}{\sqrt{x}}\,dx}_{I_{1}}
+
\underbrace{\int\limits_{\delta}^{1}\frac{\cos x}{\sqrt{x}}\,dx}_{I_{2}}
$$

Интеграл $I_{1}$ вычисляется приближённо аналитически с точностью $\varepsilon/2$. Интеграл $I_{2}$ находится численно с точностью  $\varepsilon/2$.


*Вариант 1.*

1) подсчёт $I_{1}$, $\delta = ?$

Пусть $\cos x \approx 1$, первое отброшенное слагаемое $\frac{x^{2}}{2}$. Интеграл $I_{1}$ приближенно вычисляется следующим образом

$$
I_{1} \approx \int\limits_{0}^{\delta}\frac{1}{\sqrt{x}}\,dx = 2\sqrt{\delta}
$$

Оценим ошибку
$$
\Delta =
\int\limits_{0}^{\delta} \frac{x^{2}}{2\sqrt{x}}\,dx
= \frac12 \int\limits_{0}^{\delta} x^{3/2}\,dx
= \frac{\delta^{5/2}}{5}
$$

Потребуем, чтобы ошибка не превышала $\frac{\varepsilon}{2}$

$$
\Delta \le \frac{\varepsilon}{2}
$$

и найдём наибольшее допустимое значение $\delta$:

$$
\delta \le \left(\frac{5\varepsilon}{2}\right)^{2/5}
= 0.0025^{\,2/5}
\approx 0.091028.
$$

Выбираем $\delta=\delta_{\max}\approx0.091028$.

2) подсчёт $I_{2}$, $h = ?$

$$
I_{2} = \int\limits_{\delta}^{1} \frac{\cos x}{\sqrt{x}} \, dx,
\qquad
\max_{x\in[\delta,1]} |f''(x)| = M_{2} \approx 298.77.
$$

Потребуем, чтобы погрешность вычисления интеграла по формуле трапеций не превышала  $\frac{\varepsilon}{2}$:
$$
\frac1{12} \max |f''(x)| (b-a)\, h^{2} \le \frac{\varepsilon}{2},
$$
то есть
$$
\frac1{12} \cdot 298.77 \cdot (1-\delta)\, h^{2} \le \frac12 \cdot 10^{-3},
$$
откуда
$$
h \le h_{\max}\approx 4.700\cdot 10^{-3}.
$$

Найдем количество интервалов равномерной сетки

$$
\frac{b-a}{h_{\max}} = \frac{1-\delta_{\max}}{h_{\max}} \approx 193.38
\quad\Rightarrow\quad
N = 194,\qquad
h = \frac{1-\delta_{\max}}{194}\approx 4.685\cdot10^{-3}.
$$

Итог:

$$
I \approx 2\sqrt{\delta} + I_{2}, \qquad
\delta = \delta_{\max}\approx0.091028,
$$

в программу передаём

$$
f(x) = \frac{\cos x}{\sqrt{x}},\qquad
a = \delta_{\max}\approx0.091028,\quad
b = 1,\quad
N = 194.
$$

Приближение $\cos x \approx 1$ является довольно грубым, поэтому $\delta_{\max}\approx0.091028$ довольно мало, а также из-за малости $\delta$ оказывается довольно большой константа $M_2$. Это в свою очередь приводит к сравнительно большому количеству точек $N+1 = 195$. Рассмотрим другой вариант вычисления интеграла $I_{1}$.


*Вариант 2.*

1) подсчёт $ I_{1},\ \delta - ?$

Теперь положим $\cos x \approx 1 - \frac{x^{2}}{2}$, первый отброшенный член $\frac{x^{4}}{4!}$

Приближенно вычислим интеграл

$$
I_{1} \approx \int\limits_{0}^{\delta} \frac{1 - \frac{x^{2}}{2}}{\sqrt{x}}\,dx
= 2\sqrt{\delta} - \frac{\delta^{5/2}}{5}.
$$

Оценим ошибку и найдем $\delta$

$$
\Delta = \int\limits_{0}^{\delta} \frac{x^{4}}{4!\sqrt{x}}\,dx
= \frac{\delta^{9/2}}{108}, \qquad \Delta \le \frac{\varepsilon}{2}.
$$

$$
\frac{\delta^{9/2}}{108} \le \frac{1}{2}\cdot10^{-3}
\quad\Rightarrow\quad
\delta \le (54\varepsilon)^{2/9}\approx0.522769.
$$

Поэтому выбираем наибольшее допустимое значение

$$
\delta=\delta_{\max}=(54\varepsilon)^{2/9}\approx0.522769.
$$

2) подсчет $I_{2},\ h-?$

Оценим погрешность вычисления интеграла $I_{2}$ по формуле трапеций и найдем ограничение на шаг сетки, $M_{2}=\max_{x\in[\delta_{\max},1]}|f''(x)|\approx3.411,$

$$
\frac{1}{12}\max|f''|\,(b-a)\,h^{2}\le\frac{\varepsilon}{2}
$$

$$
\frac{1}{12}\cdot3.411\cdot(1-\delta_{\max})\,h^{2}\le\frac{1}{2}\cdot10^{-3}
$$

$$
h\le h_{\max}\approx6.071\cdot10^{-2}
$$

Найдем количество интервалов равномерной сетки
$$
\frac{b-a}{h_{\max}}=
\frac{1-\delta_{\max}}{h_{\max}}\approx7.861
\quad\Rightarrow\quad
N=8,\qquad
h=\frac{1-\delta_{\max}}{8}\approx5.965\cdot10^{-2}.
$$

Итог:
$$
I\approx2\sqrt{\delta}-\frac{\delta^{5/2}}{5}+I_{2},\qquad \delta=\delta_{\max}\approx0.522769.
$$

Передаём в программу:
$$
f(x)=\frac{\cos x}{\sqrt{x}},\qquad a=\delta_{\max}\approx0.522769,\quad b=1,\quad N=8.
$$


### Обработка особенности при $x=0$ в Python

Сначала воспроизведём аналитические преобразования средствами библиотеки `SymPy`, предназначенной для символьных вычислений ([документация](https://docs.sympy.org/latest/tutorials/intro-tutorial/calculus.html)). В отличие от NumPy, она работает не с приближёнными числовыми значениями, а с формулами: строит ряд Тейлора, вычисляет интегралы и упрощает выражения.

Затем сравним численно два варианта явного выделения особенности из задачи 3 и замену переменной $x=t^2$. Для составной формулы трапеций воспользуемся функцией `numpy.trapezoid` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.trapezoid.html)).

Точное значение здесь используется только для контроля. Его вычислим с высокой точностью после замены переменной, то есть для регулярного интеграла

$$
I=2\int_0^1\cos(t^2)\,dt,
$$

с помощью `scipy.integrate.quad` ([документация](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.quad.html)).

```python
# Если SymPy не установлен, его можно добавить командой: %pip install sympy
import sympy as sp

# Положительность переменных помогает SymPy правильно упростить sqrt(t**2).
x_sym, t_sym, delta_sym = sp.symbols("x t delta", positive=True)
singular_integrand_symbolic = sp.cos(x_sym) / sp.sqrt(x_sym)

# Ряд Тейлора показывает оба приближения, использованные в задаче:
# cos(x) ≈ 1 и cos(x) ≈ 1 - x**2/2.
cos_series = sp.series(sp.cos(x_sym), x_sym, 0, 6)

cos_approximation_1 = sp.Integer(1)
cos_approximation_2 = 1 - x_sym**2 / 2

# SymPy точно интегрирует приближения на отрезке [0, delta].
I1_symbolic_1 = sp.integrate(
    cos_approximation_1 / sp.sqrt(x_sym),
    (x_sym, 0, delta_sym),
)
I1_symbolic_2 = sp.integrate(
    cos_approximation_2 / sp.sqrt(x_sym),
    (x_sym, 0, delta_sym),
)

# Интегрируем первые отброшенные члены ряда: эти выражения
# используются в аналитических оценках погрешности.
remainder_integral_1 = sp.integrate(
    (x_sym**2 / 2) / sp.sqrt(x_sym),
    (x_sym, 0, delta_sym),
)
remainder_integral_2 = sp.integrate(
    (x_sym**4 / sp.factorial(4)) / sp.sqrt(x_sym),
    (x_sym, 0, delta_sym),
)

# Выполняем замену x=t**2: подставляем x(t) и умножаем на dx/dt.
transformed_integrand = sp.simplify(
    singular_integrand_symbolic.subs(x_sym, t_sym**2)
    * sp.diff(t_sym**2, t_sym)
)

# Преобразуем готовые SymPy-выражения в функции для следующего численного блока.
I1_numeric_1 = sp.lambdify(delta_sym, I1_symbolic_1, "numpy")
I1_numeric_2 = sp.lambdify(delta_sym, I1_symbolic_2, "numpy")
singular_integrand_numeric = sp.lambdify(
    x_sym, singular_integrand_symbolic, "numpy"
)
transformed_integrand_numeric = sp.lambdify(
    t_sym, transformed_integrand, "numpy"
)

print("Ряд Тейлора для cos(x):")
display(cos_series)

print("Приближение интеграла I_1, вариант 1:")
display(I1_symbolic_1)
print("Интеграл от первого отброшенного члена, вариант 1:")
display(remainder_integral_1)

print("Приближение интеграла I_1, вариант 2:")
display(I1_symbolic_2)
print("Интеграл от первого отброшенного члена, вариант 2:")
display(remainder_integral_2)

print("Подынтегральная функция после замены x=t**2:")
display(transformed_integrand)
```

Символьные вычисления дают те же выражения, что и аналитические выкладки:

$$
I_1^{(1)}=2\sqrt{\delta},\qquad
\Delta_1=\frac{\delta^{5/2}}{5},
$$

$$
I_1^{(2)}=2\sqrt{\delta}-\frac{\delta^{5/2}}{5},\qquad
\Delta_2=\frac{\delta^{9/2}}{108}.
$$

После замены $x=t^2$ выражение $\frac{\cos x}{\sqrt{x}}\,dx$ преобразуется в $2\cos(t^2)\,dt$, поэтому особенность при нуле исчезает. `SymPy` выполняет алгебраическую часть рассуждений, но выбор $\delta$, шага сетки и числа интервалов остаётся частью численного алгоритма.

Теперь определим параметры численного алгоритма из заданной точности, а не будем задавать их заранее. Для двух вариантов явного выделения особенности получены оценки

$$
\Delta_1(\delta)=\frac{\delta^{5/2}}{5},
\qquad
\Delta_2(\delta)=\frac{\delta^{9/2}}{108}.
$$

В предыдущей кодовой ячейке эти выражения уже сохранены в переменных `remainder_integral_1` и `remainder_integral_2`. Далее будем использовать их непосредственно, не записывая те же формулы вручную.

Потребуем

$$
\Delta_j(\delta)\leq \frac{\varepsilon}{2}
$$

и решим эти неравенства с помощью `sympy.solve_univariate_inequality` ([документация](https://docs.sympy.org/latest/modules/solvers/inequalities.html)). Точность $\varepsilon=10^{-3}$ зададим как рациональное число, поэтому до перехода к десятичным приближениям SymPy будет выполнять точные вычисления.

```python
# Используем переменную delta_sym и оценки remainder_integral_1,
# remainder_integral_2, полученные в предыдущей кодовой ячейке.
epsilon_symbolic = sp.Rational(1, 1000)

# Физически параметр delta должен принадлежать промежутку (0, 1].
delta_domain = sp.Interval(0, 1, left_open=True)


def solve_error_inequality(error_bound, variable, domain, tolerance):
    """Решает неравенство error_bound <= tolerance."""
    return sp.solve_univariate_inequality(
        sp.Le(error_bound, tolerance),
        variable,
        relational=False,
        domain=domain,
    )


# Решаем два неравенства относительно delta.
admissible_delta_1 = solve_error_inequality(
    remainder_integral_1, delta_sym, delta_domain,
    tolerance=epsilon_symbolic / 2,
)

admissible_delta_2 = solve_error_inequality(
    remainder_integral_2, delta_sym, delta_domain,
    tolerance=epsilon_symbolic / 2,
)

# Берём наибольшие допустимые delta: так численное интегрирование
# начинается дальше от особой точки x=0.
delta_1_exact = admissible_delta_1.sup
delta_2_exact = admissible_delta_2.sup
delta_1 = float(sp.N(delta_1_exact, 16))
delta_2 = float(sp.N(delta_2_exact, 16))

print("Допустимые значения delta для варианта 1:")
display(admissible_delta_1)
print("Допустимые значения delta для варианта 2:")
display(admissible_delta_2)

delta_results = pd.DataFrame(
    {
        "Вариант": [1, 2],
        "Максимальное допустимое delta": [delta_1, delta_2],
        "Оценка ошибки при выбранном delta": [
            float(remainder_integral_1.subs(delta_sym, delta_1_exact)),
            float(remainder_integral_2.subs(delta_sym, delta_2_exact)),
        ],
    }
)
show_table(delta_results)
```

Теперь выберем шаг равномерной сетки. Для составной формулы трапеций справедлива оценка

$$
\Delta_T\leq \frac{1}{12}M_2(1-\delta)h^2,
\qquad
M_2=\max_{x\in[\delta,1]}|f''(x)|.
$$

SymPy найдёт вторую производную функции $f(x)=\cos x/\sqrt{x}$. Для рассматриваемых значений $\delta$ функция $|f''(x)|$ убывает на $[\delta,1]$, поэтому максимум достигается в левой точке: $M_2=|f''(\delta)|$.

После этого снова решим неравенство, теперь относительно $h$:

$$
\frac{1}{12}M_2(1-\delta)h^2\leq\frac{\varepsilon}{2}.
$$

Полученное значение $h_{\max}$ не обязано точно делить отрезок $[\delta,1]$. Поэтому выберем

$$
N=\left\lceil\frac{1-\delta}{h_{\max}}\right\rceil,
\qquad
h=\frac{1-\delta}{N}\leq h_{\max}.
$$

```python
# Символьно вычисляем вторую производную подынтегральной функции.
x_symbol = sp.symbols("x", positive=True)
f_symbolic = sp.cos(x_symbol) / sp.sqrt(x_symbol)
f_second_symbolic = sp.simplify(sp.diff(f_symbolic, x_symbol, 2))
f_second = sp.lambdify(x_symbol, f_second_symbolic, "numpy")

print("Вторая производная f(x):")
display(f_second_symbolic)

h_symbol = sp.symbols("h", positive=True)
h_domain = sp.Interval(0, 1, left_open=True)


def choose_trapezoid_grid(delta_value: float) -> dict:
    """Находит допустимый шаг и минимальное число равных интервалов."""
    # Максимум |f''| на [delta, 1] достигается при x=delta.
    M2 = float(abs(f_second(delta_value)))

    # Априорная оценка погрешности составной формулы трапеций.
    error_bound = (
        sp.Float(M2, 16)
        * (1 - sp.Float(delta_value, 16))
        * h_symbol**2
        / 12
    )

    # Решаем неравенство error_bound <= epsilon/2 относительно h.
    admissible_h = solve_error_inequality(
    error_bound, h_symbol, h_domain,
    tolerance=epsilon_symbolic / 2,
    )
    h_max = float(sp.N(admissible_h.sup, 16))

    # Фактический шаг выбираем так, чтобы сетка точно заканчивалась в x=1.
    N = int(np.ceil((1 - delta_value) / h_max))
    h_actual = (1 - delta_value) / N

    return {
        "M2": M2,
        "h_max": h_max,
        "N": N,
        "h": h_actual,
        "trapezoid_error_bound": M2 * (1 - delta_value) * h_actual**2 / 12,
    }


grid_1 = choose_trapezoid_grid(delta_1)
grid_2 = choose_trapezoid_grid(delta_2)

grid_results = pd.DataFrame(
    [
        {
            "Вариант": 1,
            "delta": delta_1,
            "M2": grid_1["M2"],
            "Максимальный шаг": grid_1["h_max"],
            "Число интервалов N": grid_1["N"],
            "Фактический шаг": grid_1["h"],
        },
        {
            "Вариант": 2,
            "delta": delta_2,
            "M2": grid_2["M2"],
            "Максимальный шаг": grid_2["h_max"],
            "Число интервалов N": grid_2["N"],
            "Фактический шаг": grid_2["h"],
        },
    ]
)
show_table(grid_results)
```

```python
from scipy.integrate import quad


def composite_trapezoid(f, a: float, b: float, N: int) -> float:
    """Составная формула трапеций на N равных интервалах."""
    x = np.linspace(a, b, N + 1)
    return np.trapezoid(f(x), x)


# Эталонное значение получаем после устранения особенности заменой x=t^2.
# Подынтегральная функция уже построена SymPy в предыдущем блоке.
I_singular_reference, _ = quad(
    transformed_integrand_numeric,
    0.0,
    1.0,
    epsabs=1e-12,
    epsrel=1e-12,
)

# Вариант 1: cos(x) заменяется единицей на [0, delta_1].
N_1 = grid_1["N"]
I_variant_1 = (
    float(I1_numeric_1(delta_1))
    + composite_trapezoid(singular_integrand_numeric, delta_1, 1.0, N_1)
)

# Вариант 2: используются два первых ненулевых члена ряда cos(x).
N_2 = grid_2["N"]
I_variant_2 = (
    float(I1_numeric_2(delta_2))
    + composite_trapezoid(singular_integrand_numeric, delta_2, 1.0, N_2)
)

# После замены x=t^2 интегрируем регулярную функцию 2*cos(t^2).
# Оценка |g''(t)| <= 12 даёт ошибку трапеций не более 1/N^2.
N_symbol = sp.symbols("N", positive=True)
N_domain = sp.Interval(0, sp.oo, left_open=True)
admissible_N = solve_error_inequality(
    1 / N_symbol**2, N_symbol, N_domain,
    tolerance=epsilon_symbolic,
)
N_substitution = int(sp.ceiling(admissible_N.inf))
I_substitution = composite_trapezoid(
    transformed_integrand_numeric, 0.0, 1.0, N_substitution
)

# Априорные оценки для выбранных параметров.
analytical_bound_1 = float(
    remainder_integral_1.subs(delta_sym, delta_1_exact)
)
analytical_bound_2 = float(
    remainder_integral_2.subs(delta_sym, delta_2_exact)
)
substitution_bound = 1 / N_substitution**2

improper_results = pd.DataFrame(
    [
        {
            "Способ": "Выделение особенности: вариант 1",
            "delta": delta_1,
            "Число интервалов": N_1,
            "Вычислений функции": N_1 + 1,
            "Априорная оценка": (
                analytical_bound_1 + grid_1["trapezoid_error_bound"]
            ),
            "Приближенное значение": I_variant_1,
            "Фактическая погрешность": abs(I_variant_1 - I_singular_reference),
        },
        {
            "Способ": "Выделение особенности: вариант 2",
            "delta": delta_2,
            "Число интервалов": N_2,
            "Вычислений функции": N_2 + 1,
            "Априорная оценка": (
                analytical_bound_2 + grid_2["trapezoid_error_bound"]
            ),
            "Приближенное значение": I_variant_2,
            "Фактическая погрешность": abs(I_variant_2 - I_singular_reference),
        },
        {
            "Способ": "Замена x=t^2",
            "delta": np.nan,
            "Число интервалов": N_substitution,
            "Вычислений функции": N_substitution + 1,
            "Априорная оценка": substitution_bound,
            "Приближенное значение": I_substitution,
            "Фактическая погрешность": abs(I_substitution - I_singular_reference),
        },
    ]
)

print(f"Контрольное значение интеграла: {I_singular_reference:.12f}")
show_table(improper_results)
```

Во всех трёх случаях априорная оценка не превышает требуемой точности $10^{-3}$. Фактическая погрешность оказывается меньше оценки, поскольку оценки остатка ряда Тейлора и погрешности формулы трапеций не обязаны достигаться одновременно.

Во втором варианте дополнительный член разложения позволяет выбрать большее $\delta$ и начать численное интегрирование дальше от особой точки, поэтому требуется меньше интервалов. И в основном решении, и в демонстрации используется наибольшее значение, найденное из неравенства: $\delta_{\max}\approx0.522769$. Замена $x=t^2$ полностью устраняет особенность и сводит задачу к интегрированию регулярной функции.

## Задача 5

Предложить алгоритм, как пользуясь программой, реализующей метод трапеций для численного интегрирования регулярных функций с произвольным шагом, вычислить следующий интеграл с точностью $\varepsilon=10^{-3}$

$$I = \int \limits _{0}^{\pi/2} \ln(\sin x)\,dx$$

**Решение.**

Проведем следующие преобразования интеграла (воспользуемся методом Канторовича)

$$
I = \int \limits _{0}^{\pi/2} \ln(\sin x)\,dx
= \int\limits_{0}^{\pi/2} \ln(\sin x)\,dx
- \int\limits_{0}^{\pi/2} \ln x\,dx
+ \int\limits_{0}^{\pi/2} \ln x\,dx
$$

$$
= \int\limits_{0}^{\pi/2} \ln\!\left(\frac{\sin x}{x}\right)\!dx
+ \frac{\pi}{2}\left(\ln\frac{\pi}{2} - 1\right)
$$

Первый интеграл вычислим по методу трапеций (численно). Шаг найдем, воспользовавшись оценкой ошибки метода
$$
M_2 = 0.595
$$

$$
\frac{1}{12} M_2 (b-a) h^2 \le \varepsilon
$$

$$
\frac{1}{12}\cdot 0.595 \cdot \left(\frac{\pi}{2}-0\right) h^2
\le 10^{-3}
$$

$$
h \le 0.114
$$

Найдем число интервалов равномерной сетки

$$
\frac{b-a}{h}=\frac{\pi/2}{0.0114} = 13.8 \;\Rightarrow\; N = 14
$$

Итог:

$$
